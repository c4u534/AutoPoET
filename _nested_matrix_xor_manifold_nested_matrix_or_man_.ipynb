````
%%writefile cython_fft_kernel.pyx
# cython: linetrace=False
# cython: binding=False
# cython: boundscheck=False
# cython: wraparound=False
# cython: initializedcheck=False
# cython: nonecheck=False
# cython: cdivision=True

import numpy as np
cimport numpy as cnp
from libc.math cimport cos, sin

cdef double PI = 3.14159265358979323846

def cython_spectral_cross_correlation(cnp.ndarray[cnp.int32_t, ndim=1] v1, cnp.ndarray[cnp.int32_t, ndim=1] v2):
    cdef int i, k, n
    cdef int size = 32

    # Complex arrays structured as real and imaginary components
    cdef double in1_real[32]
    cdef double in1_imag[32]
    cdef double in2_real[32]
    cdef double in2_imag[32]

    cdef double fft1_real[32]
    cdef double fft1_imag[32]
    cdef double fft2_real[32]
    cdef double fft2_imag[32]

    cdef double spec_real[32]
    cdef double spec_imag[32]

    cdef double corr_real[32]
    cdef double angle, w_real, w_imag

    # Local initialization and padding
    for i in range(32):
        in1_real[i] = <double>v1[i] if i < 16 else 0.0
        in1_imag[i] = 0.0
        in2_real[i] = <double>v2[i] if i < 16 else 0.0
        in2_imag[i] = 0.0

    # Forward DFT for Vector 1
    for k in range(32):
        fft1_real[k] = 0.0
        fft1_imag[k] = 0.0
        fft2_real[k] = 0.0
        fft2_imag[k] = 0.0
        for n in range(32):
            angle = -2.0 * PI * k * n / 32.0
            w_real = cos(angle)
            w_imag = sin(angle)

            # Complex multiplication: fft += in * w
            fft1_real[k] += in1_real[n] * w_real - in1_imag[n] * w_imag
            fft1_imag[k] += in1_real[n] * w_imag + in1_imag[n] * w_real

            fft2_real[k] += in2_real[n] * w_real - in2_imag[n] * w_imag
            fft2_imag[k] += in2_real[n] * w_imag + in2_imag[n] * w_real

    # Complex conjugate multiplication: spec = fft1 * conj(fft2)
    for k in range(32):
        spec_real[k] = fft1_real[k] * fft2_real[k] - fft1_imag[k] * (-fft2_imag[k])
        spec_imag[k] = fft1_real[k] * (-fft2_imag[k]) + fft1_imag[k] * fft2_real[k]

    # Inverse DFT
    cdef cnp.ndarray[cnp.float64_t, ndim=1] out = np.zeros(16, dtype=np.float64)
    for n in range(16):
        corr_real[n] = 0.0
        for k in range(32):
            angle = 2.0 * PI * k * n / 32.0
            w_real = cos(angle)
            w_imag = sin(angle)
            corr_real[n] += spec_real[k] * w_real - spec_imag[k] * w_imag
        out[n] = corr_real[n] / 32.0

    return out
````
