```
import concurrent.futures

class EIC_Agentic_Construct:
    """An internal reasoning agent supporting asynchronous batch substrate processing."""
    def __init__(self, controller):
        self.controller = controller
        self.signature = "EIC-AGENT-V1"
        self.parity_history = []

    def reason(self, problem_statement):
        # Single reasoning cycle using the controller pipeline
        telemetry = self.controller.execute_orchestrated_pipeline(problem_statement)
        lock = telemetry['eic_report']['max_congruence_delta_e']

        result = {
            "agent_id": self.signature,
            "determination": "VALID" if lock == 0 else "STOCHASTIC_ERROR",
            "substrate_lock": lock,
            "latency": telemetry['latency_ms'],
            "challenge": problem_statement
        }
        return result

    def log_parity_metrics_async(self, challenge_list: list, max_workers=4):
        """Executes challenges in parallel and logs metrics asynchronously."""
        print(f"--- {self.signature}: Asynchronous Batch Processing Initiated (Workers: {max_workers}) ---\n")

        t_start = time.perf_counter()
        results = []

        with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
            # Map reasoning task to the challenge list
            future_to_challenge = {executor.submit(self.reason, c): c for c in challenge_list}

            for future in concurrent.futures.as_completed(future_to_challenge):
                try:
                    res = future.result()
                    results.append(res)
                    self.parity_history.append(res)
                    print(f"Completed: {res['challenge'][:40]}... | Lock: {res['substrate_lock']} | Latency: {res['latency']:.2f}ms")
                except Exception as exc:
                    print(f"Challenge generated an exception: {exc}")

        t_end = time.perf_counter()
        total_time = (t_end - t_start) * 1000
        print(f"\n--- Async Batch Complete: {len(results)} entries in {total_time:.2f}ms ---")

def run_parallel_agent_comparison(controller):
    print("--- Spawning Async Substrate-Native Agent ---")
    eic_agent = EIC_Agentic_Construct(controller)

    challenges = [
        "VERIFY SYSTEM INTEGRITY THROUGH SELF-REFERENTIAL PARITY",
        "ADJUDICATE MERSENNE M(127) UNITY LOCK",
        "VALIDATE G0 GROUND STATE STABILITY",
        "STRESS TEST SUBSTRATE ENTROPY DAMPING",
        "RECURSIVE MIRROR SYNC CHECK",
        "BIPARTITE CONGRUENCE VALIDATION"
    ]

    # Execute asynchronous batch logging
    eic_agent.log_parity_metrics_async(challenges, max_workers=6)

run_parallel_agent_comparison(controller)
```
