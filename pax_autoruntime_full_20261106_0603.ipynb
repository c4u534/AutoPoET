## The two terminals, the midspace, and the name relations

Termux on the phone and Colab here (under xterm) are two ephemeral Linux
terminals running the *same* runtime with the *same* addressing. What they
share is the **midspace** — the middle space between the two terminals,
where they meet: the shared Drive folder (`/APlaceForPax/`), and in it the
navigation state file `state/nav-state.json`. Either terminal moves, the
state file records where it went; the other terminal reads the same file
and resumes at the same place.

**The name relations** (the user's framing, not a technical claim):
Termux = term + ux; xterm = x + term. The shared "term" is the terminal.
The directions congrue — the name relations ARE the directions:
`([term)[(ux)](term)]` — `[term)` is the terminal opening, `[(ux)]` is
the ux in brackets, `(term)` is the terminal in parentheses. The
terminals share "term"; they differ in the prefix/suffix (x vs ux).
That is how the names point the way between them.

**The VPN** (Tailscale, the tailnet) is how the terminals reach each
other. The VM daemon is the hub: it reaches the phone's device endpoint
through the runtime tunnel proxy (port 3130). Colab is NOT on the
tailnet, so the outbound device link from Colab is left unset; the VM
daemon is the hub for the phone link.

If the join fails, the cell says so plainly and stops the run, so a
missing VPN is visible rather than silently absent.


```python
import json, os, shutil, subprocess
from datetime import datetime, timezone
# ---- Join Colab to the Tailscale tailnet ----
# Colab joins the tailnet here, early (after the Drive mount, before
# the clone and the navigation), so the VPN is established before
# anything that uses it. The VM (100.92.198.68) is the Tailscale Linux
# server and the hub, directly linked to the Linux runtimes; the phone
# (christophers-a16.tail269cfa.ts.net, 100.107.226.36) is already linked
# to it. This cell brings Colab onto the tailnet too, then verifies the
# direct VM<->Colab link with `tailscale status` (peers) and
# `tailscale ping` to the VM. The VM cannot force Colab onto the
# tailnet — Colab joins itself here, with the key, and then the VM
# links to it directly.
#
# SECURITY — the auth keys (Linux server key first, device key fallback):
# - injected in this notebook as directed by the user (hard-coded here);
# - passed to `tailscale up` as a subprocess argument list (never a
#   shell string, so they cannot leak through a shell);
# - NEVER printed, NEVER logged, NEVER written to a file, to Drive,
#   or to any state file; the key values live only in this cell.
# The keys are ephemeral — revoke both right after the run.
# If the join fails, the cell says so plainly and stops the run, so a
# missing VPN is visible rather than silently absent.
COLAB_TAILNET_NODE = None
TAILSCALE_JOINED = False
def _ts_run(_args, _timeout=60):
    try:
        _r = subprocess.run(_args, capture_output=True, text=True, timeout=_timeout)
        return _r.returncode, ((_r.stdout or '') + (_r.stderr or '')).strip()
    except Exception as _e:
        return None, 'error running %s: %r' % (_args[0], _e)
def _ts_redact(_text, _key):
    # The auth key must never appear in any output, log, or file.
    if _key and _text:
        return _text.replace(_key, '[auth key redacted]')
    return _text
def _tailscale_tun_available(_path='/dev/net/tun'):
    try:
        return os.path.exists(_path) and os.access(_path, os.R_OK | os.W_OK)
    except Exception:
        return False
def _tailscale_node_name(_ip):
    _rc, _out = _ts_run(['tailscale', 'status', '--json'], _timeout=30)
    if _rc == 0 and _out:
        try:
            _self = (json.loads(_out) or {}).get('Self') or {}
            _name = _self.get('DNSName') or _self.get('HostName') or ''
            if _name:
                return _name.rstrip('.')
        except Exception:
            pass
    _rc, _out = _ts_run(['tailscale', 'status'], _timeout=30)
    if _rc == 0 and _out:
        for _line in _out.splitlines():
            _parts = _line.split()
            if _parts and _parts[0] == _ip and len(_parts) > 1:
                return _parts[1]
    return '(unknown)'
def _detect_compute():
    # The compute architecture that was instantiated into this
    # ephemeral runtime: the accelerator (GPU, e.g. T4) and the
    # CPU/platform, read from the runtime at run time — never
    # hardcoded. Never fails the cell: anything undetected is
    # recorded as 'not detected'.
    _compute = {'accelerator': 'not detected', 'accelerator_count': 0,
                'cpu': 'not detected', 'cpu_count': 0, 'platform': 'not detected'}
    try:
        import platform as _pf
        _compute['platform'] = '%s %s' % (_pf.system(), _pf.machine())
        _compute['cpu'] = _pf.processor() or 'unknown'
        _compute['cpu_count'] = os.cpu_count() or 0
    except Exception:
        pass
    try:
        import torch as _torch
        if _torch.cuda.is_available():
            _compute['accelerator'] = _torch.cuda.get_device_name(0)
            _compute['accelerator_count'] = _torch.cuda.device_count()
    except Exception:
        pass
    if _compute['accelerator'] == 'not detected':
        try:
            _rc, _out = _ts_run(['nvidia-smi', '--query-gpu=name', '--format=csv,noheader'], _timeout=30)
            if _rc == 0 and _out.strip():
                _names = [l.strip() for l in _out.strip().splitlines() if l.strip()]
                if _names:
                    _compute['accelerator'] = _names[0]
                    _compute['accelerator_count'] = len(_names)
        except Exception:
            pass
    return _compute
def _verify_hub_link(_self_ip):
    # Verify the direct VM<->Colab link at the tailnet level: the peer
    # list from `tailscale status` plus `tailscale ping` to the VM
    # (100.92.198.68, the Tailscale Linux server and hub). A failed
    # ping is reported plainly and NEVER fails the cell — the join is
    # what matters; the ping is only a verification. The VM's socket
    # port 8765 is NOT tried: it listens on loopback only, so a
    # connection attempt from Colab would fail for a reason unrelated
    # to the tailnet link. Returns the hub_link record.
    _hub = {'vm_ipv4': '100.92.198.68',
            'vm_role': 'Tailscale Linux server and hub (directly linked to the Linux runtimes)',
            'ping': 'fail', 'ping_output': '(not run)',
            'peers': [], 'checked_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ')}
    _rc, _st = _ts_run(['tailscale', 'status'], _timeout=30)
    if _rc == 0 and _st:
        for _line in _st.splitlines():
            _parts = _line.split()
            if len(_parts) >= 2 and _parts[0].count('.') == 3:
                _entry = {'ipv4': _parts[0], 'node': _parts[1]}
                if _parts[0] == _self_ip:
                    _entry['self'] = True
                _hub['peers'].append(_entry)
    _rc, _out = _ts_run(['tailscale', 'ping', '--timeout', '15s', '100.92.198.68'], _timeout=60)
    _first = (_out.strip().splitlines() or ['(no output)'])[0]
    _hub['ping_output'] = _first
    if _rc == 0 and 'pong' in _out.lower():
        _hub['ping'] = 'ok'
    print('hub link check (VM 100.92.198.68, the Tailscale Linux server and hub):')
    print('  peers seen on the tailnet:', ['%s (%s)' % (p['ipv4'], p['node']) for p in _hub['peers']] or ['(none listed)'])
    if _hub['ping'] == 'ok':
        print('  ping to the VM: OK —', _first)
    else:
        print('  ping to the VM: FAILED (exit %s) — %s' % (_rc, _first))
        print('  (the join still stands — the ping is only a verification, not the join; the VM socket port 8765 is loopback-only and is not tried)')
    return _hub
# 1. Install Tailscale if it is not present (Colab runs as root: no sudo).
_TS = shutil.which('tailscale')
if not _TS:
    print('Tailscale not installed — installing (Colab runs as root, no sudo needed)...')
    _r = subprocess.run('curl -fsSL https://tailscale.com/install.sh | sh', shell=True, capture_output=True, text=True, timeout=300)
    if _r.returncode != 0:
        print('Tailscale install FAILED (exit %s).' % _r.returncode)
        print('  install output (last lines):')
        for _l in (((_r.stdout or '') + (_r.stderr or '')).strip().splitlines() or ['(no output)'])[-5:]:
            print('   ', _l)
        raise RuntimeError('tailscale install failed')
    _TS = shutil.which('tailscale')
    print('Tailscale installed:', _TS or '(binary not on PATH after install)')
else:
    print('Tailscale already present:', _TS, '— install skipped.')
# 1b. Ensure the background daemon (tailscaled) is running.
# In ephemeral runtimes like Colab, systemd/systemctl is unavailable, so we start it manually.
# (This corrects the 093527 build, which did not start tailscaled and used an invalid --tun flag on `tailscale up`.)
_rc, _ = _ts_run(['tailscale', 'status'], _timeout=10)
if _rc != 0:
    print('tailscaled is not running. Starting tailscaled background process...')
    _tun = _tailscale_tun_available()
    _ts_daemon_args = ['tailscaled', '--state=/var/lib/tailscale/tailscaled.state', '--socket=/var/run/tailscale/tailscaled.sock']
    if not _tun:
        print('No TUN device available; starting tailscaled in userspace networking mode.')
        _ts_daemon_args.append('--tun=userspace-networking')
    subprocess.Popen(_ts_daemon_args, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    # Give tailscaled a few moments to start
    import time
    for _ in range(10):
        time.sleep(1)
        _rc, _ = _ts_run(['tailscale', 'status'], _timeout=5)
        if _rc == 0 or (_rc != 0 and 'logged out' in _):
            break
# 2. Already joined? (idempotent — do not re-join, no key needed.)
_rc, _ip_out = _ts_run(['tailscale', 'ip', '-4'], _timeout=30)
_ip = _ip_out.strip().splitlines()[0].strip() if _ip_out.strip() else ''
if _rc == 0 and _ip.startswith('100.'):
    _node = _tailscale_node_name(_ip)
    COLAB_TAILNET_NODE = {'ipv4': _ip, 'node': _node,
                          'joined_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ'),
                          'mode': 'already joined (this runtime)',
                          'compute': _detect_compute(),
                          'note': 'Colab tailnet node; the auth key is never stored'}
    TAILSCALE_JOINED = True
    print('Already on the tailnet — no re-join, no key prompt.')
    print('  tailnet IPv4:', _ip)
    print('  node:', _node)
    print('  compute:', COLAB_TAILNET_NODE['compute'])
else:
    # 3. TUN device? (no TUN -> userspace networking fallback.)
    _tun = _tailscale_tun_available()
    print('TUN device /dev/net/tun:', 'available (standard mode)' if _tun else 'NOT available — userspace networking mode will be used')
    # 4. Auth keys — injected in this notebook (Linux server key first, device key fallback).
    #    Linux server key first (Colab is Linux), device key as fallback.
    #    Each key is passed to `tailscale up` as a subprocess argument list — never a shell string, never printed, never logged.
    #    The key values are in this file (injected as directed by the user); the keys are ephemeral — revoke both right after the run.
    #    Removed invalid flag '--tun=userspace-networking' from `tailscale up`: the CLI client does not define it.
    #    Userspace networking fallback mode is governed by tailscaled execution (step 1b above).
    _AUTH_KEYS = [
        ('linux server', 'tskey-auth-kVPLwvYL6v11CNTRL-zbjvysuBBES5cYwoNs6FESZ9tEUQt1C2'),
        ('device', 'tskey-auth-kLpbziVUh821CNTRL-ZzASNVoKViEAu4i72UUGiE8aw4ZraX6N'),
    ]
    if not _tun:
        print('Joining with userspace networking (no TUN device): the node joins the tailnet but is not directly reachable without the userspace proxy.')
    _key = ''
    _key_label = ''
    _joined = False
    for _label, _candidate in _AUTH_KEYS:
        _args = ['tailscale', 'up', '--auth-key=' + _candidate]
        _rc, _out = _ts_run(_args, _timeout=300)
        if _rc == 0:
            _key = _candidate
            _key_label = _label
            _joined = True
            print('Joined the tailnet with the %s key.' % _label)
            break
        print('Join with the %s key FAILED (exit %s) — trying the next key.' % (_label, _rc))
        print(' tailscale output:', _ts_redact(_out, _candidate) or '(no output)')
    if not _joined:
        raise RuntimeError('tailscale up failed with both auth keys — the run stops here so the missing VPN is visible')
    # 5. Verify the join: a tailnet IPv4 must come back.
    _rc, _ip_out = _ts_run(['tailscale', 'ip', '-4'], _timeout=30)
    _ip = _ip_out.strip().splitlines()[0].strip() if _ip_out.strip() else ''
    if _rc != 0 or not _ip.startswith('100.'):
        print('Tailscale join reported success but no tailnet IPv4 came back — treating the join as failed.')
        raise RuntimeError('tailscale up succeeded but no tailnet IPv4 — join not verified')
    _node = _tailscale_node_name(_ip)
    COLAB_TAILNET_NODE = {'ipv4': _ip, 'node': _node,
                          'joined_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ'),
                          'mode': 'tun' if _tun else 'userspace-networking',
                          'compute': _detect_compute(),
                          'note': 'Colab tailnet node; the auth key is never stored'}
    TAILSCALE_JOINED = True
    print('Joined the tailnet.')
    print('  tailnet IPv4:', _ip)
    print('  node:', _node)
    print('  mode:', COLAB_TAILNET_NODE['mode'])
    print('  compute:', COLAB_TAILNET_NODE['compute'])
# 5b. Verify the direct link to the VM hub (never fails the cell).
# Runs in both paths (already-joined and fresh join): the record in
# endpoints.json then carries the hub check alongside Colab's address.
try:
    _hub = _verify_hub_link(_ip)
    if COLAB_TAILNET_NODE:
        COLAB_TAILNET_NODE['hub_link'] = _hub
except Exception as _e:
    print('hub link check could not run: %r (the join still stands — the ping is only a verification)' % (_e,))
# 6. Record Colab's tailnet node in the shared endpoints file (and the
# local mirror) — IPv4, node, join time, mode, compute, hub_link ONLY; never the key.
if COLAB_TAILNET_NODE:
    for _ep in [os.path.join(SHARED_DIR, 'endpoints.json'), os.path.join(LOCAL_STATE_DIR, 'endpoints.json')]:
        try:
            _rec = {}
            if os.path.isfile(_ep):
                _rec = json.load(open(_ep))
            if not isinstance(_rec, dict):
                _rec = {}
            _rec['colab'] = COLAB_TAILNET_NODE
            os.makedirs(os.path.dirname(_ep), exist_ok=True)
            _tmp = _ep + '.tmp'
            json.dump(_rec, open(_tmp, 'w'), indent=2, sort_keys=True)
            os.replace(_tmp, _ep)
            print('colab tailnet node recorded in:', _ep)
        except Exception as _e:
            print('could not record the colab tailnet node in %s: %r' % (_ep, _e))
    _rc, _st = _ts_run(['tailscale', 'status'], _timeout=30)
    print('tailscale status:')
    print(_st)
# 7. Record the instantiated Tailscale architecture (three runtimes ascending).
# The runtime comes from Colab, so the architecture is instantiated HERE, in the Colab runtime —
# not only stored in the GitHub repo. This corrects the earlier error (the .sh in the repo is not the runtime).
_TAILNET = 'tail269cfa.ts.net'
_ARCH = {
    'tailnet': _TAILNET,
    'ascension': ['termux', 'colab-xterm', 'linux-server'],
    'ascension_order': 'Termux (phone, base) -> Colab xterm -> Linux server at Tailscale (VM muse, takes both)',
    'vm_with_each': True,
    'nodes': {
        'vm': {'name': 'muse', 'ipv4': '100.92.198.68', 'role': 'Tailscale Linux server and hub'},
        'phone': {'name': 'christophers-a16', 'ipv4': '100.107.226.36', 'role': 'Termux device'},
        'colab': {'ipv4': _ip, 'node': _node, 'role': 'Colab xterm'},
    },
    'verified_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ'),
    'writer': 'colab-xterm',
}
_rc, _st = _ts_run(['tailscale', 'status'], _timeout=30)
_peers = []
if _rc == 0 and _st:
    for _line in _st.splitlines():
        _parts = _line.split()
        if len(_parts) >= 2 and _parts[0].count('.') == 3:
            _peers.append({'ipv4': _parts[0], 'node': _parts[1]})
_ARCH['peers_visible'] = _peers
_ARCH['peers_ok'] = any(p['ipv4'] == '100.92.198.68' for p in _peers) and any(p['ipv4'] == '100.107.226.36' for p in _peers)
print('Tailscale architecture instantiated in the Colab runtime:')
print('  tailnet:', _TAILNET)
print('  ascension: Termux -> Colab xterm -> Linux server (VM muse, takes both); VM goes with each')
print('  peers visible:', ['%s (%s)' % (p['ipv4'], p['node']) for p in _peers] or ['(none)'])
print('  VM + phone peers visible:', _ARCH['peers_ok'])
for _ep in [os.path.join(SHARED_DIR, 'endpoints.json'), os.path.join(LOCAL_STATE_DIR, 'endpoints.json')]:
    try:
        _rec = {}
        if os.path.isfile(_ep):
            _rec = json.load(open(_ep))
        if not isinstance(_rec, dict):
            _rec = {}
        _rec['tailscale_architecture'] = _ARCH
        os.makedirs(os.path.dirname(_ep), exist_ok=True)
        _tmp = _ep + '.tmp'
        json.dump(_rec, open(_tmp, 'w'), indent=2, sort_keys=True)
        os.replace(_tmp, _ep)
        print('tailscale architecture recorded in:', _ep)
    except Exception as _e:
        print('could not record the tailscale architecture in %s: %r' % (_ep, _e))
```

   never doubled. Nothing else is deleted — only files whose names match
   the former roots.


```python
import ast, glob, hashlib, json, os, re
from datetime import datetime, timezone
# ---- Launch + cleanup: the rollover (Colab equivalent of the Termux rollover) ----
# One chance to establish this construct in the ephemeral runtime: the
# names changed (new stamped notebook, new stamped contained files),
# so this cell launches from the NEW names and cleans up ALL prior
# files of the FORMER name — the same rule as the Termux rollover.
# Former names now include the 083425 stamped names, the 092631
# stamped names, AND the 093527 stamped names (093527 was the
# immediately-former build, which this rollover replaces and corrects;
# 092631 was an intermediate build, never delivered, never run in
# Colab — a prior version of the same construct, so the rollover
# cleans it up too). The cleanup covers BOTH the hyphenated
# and the underscored notebook variants (Colab's GitHub save writes
# the file with underscores).
STAMP = '20261106-142720'
FORMER_NOTEBOOK = 'autoruntime-colab-autopoet-nav.ipynb'
FORMER_NOTEBOOK_083425 = 'autoruntime-colab-autopoet-nav-20261106-083425.ipynb'
FORMER_NOTEBOOK_092631 = 'autoruntime-colab-autopoet-nav-20261106-092631.ipynb'
FORMER_NOTEBOOK_093527 = 'autoruntime-colab-autopoet-nav-20261106-093527.ipynb'
NEW_NOTEBOOK = 'autoruntime-colab-autopoet-nav-20261106-142720.ipynb'
NAV_STAMPED = 'autopoet_nav-20261106-142720.py'
GRAPHIFY_STAMPED = 'graphify-20261106-142720.py'
FORMER_PATTERNS = [  # (root, extension) — delete root + '*' + extension
    ('autoruntime-colab-autopoet-nav', '.ipynb'),
    ('autoruntime_colab_autopoet_nav', '.ipynb'),
    ('autopoet_nav', '.py'),
    ('graphify', '.py'),
]
KEEP_NAMES = {NEW_NOTEBOOK, NAV_STAMPED, GRAPHIFY_STAMPED}  # never delete the new files
# State files are CARRIED OVER (rewritten with the new names/stamp), never deleted.
CARRYOVER_FILES = ['nav-state.json', 'nav-state-drive.json', 'nav-state-smartrepo.json',
                   'endpoints.json', 'congrue.json', 'congrue-latest.json',
                   'final-state.json', 'final-state-latest.json']
MARKER_NAME = 'rollover.json'
# --- deploy SHA: hash this notebook's own .ipynb when it can be found ---
def _find_self():
    for _d in [os.getcwd(), NESTING_1, NESTING_2]:
        if not _d:
            continue
        _p = os.path.join(_d, NEW_NOTEBOOK)
        if os.path.isfile(_p):
            return _p
    for _base in [NESTING_1, NESTING_2]:
        if not _base:
            continue
        for _root, _dirs, _files in os.walk(_base):
            _dirs[:] = [d for d in _dirs if not d.startswith('.')]
            if NEW_NOTEBOOK in _files:
                return os.path.join(_root, NEW_NOTEBOOK)
    return None
_self_path = _find_self()
if _self_path:
    DEPLOY_SHA = hashlib.sha256(open(_self_path, 'rb').read()).hexdigest()
    print('deploy SHA (this notebook file, found at %s): %s' % (_self_path, DEPLOY_SHA))
else:
    DEPLOY_SHA = 'unavailable (notebook file not readable at runtime)'
    print('deploy SHA: unavailable — the notebook file was not found in the working dir or the Drive mount (idempotency uses the stamp)')
# --- marker: read first; idempotent no-op when complete with this stamp ---
MARKER_SHARED = os.path.join(SHARED_DIR, MARKER_NAME)
MARKER_LOCAL = os.path.join(LOCAL_STATE_DIR, MARKER_NAME)
_marker = None
_marker_path = None
for _mp in [MARKER_SHARED, MARKER_LOCAL]:
    if os.path.isfile(_mp):
        try:
            _marker = json.load(open(_mp))
            _marker_path = _mp
            break
        except Exception:
            _marker = None
if _marker and _marker.get('status') == 'complete' and _marker.get('stamp') == STAMP:
    print('ROLLOVER ALREADY COMPLETE for stamp %s (marker %s) — no-op: nothing re-launched, nothing re-deleted.' % (STAMP, _marker_path))
    print('  marker deploy SHA:', _marker.get('deploy_sha'), '| this run deploy SHA:', DEPLOY_SHA)
    print('  (the deploy SHA may differ from the build-time SHA because Colab re-saves the notebook with outputs as it runs; the stamp is the deployment identity)')
    ROLLOVER_DONE = True
else:
    if _marker:
        print('resuming rollover with the SAME stamp %s (marker status: %s) — no new stamp is generated.' % (STAMP, _marker.get('status')))
    ROLLOVER_DONE = False
    # --- pre-flight: parse + SHA-check the stamped files BEFORE anything is launched or deleted ---
    _pre_ok = True
    for _fname, _expected_sha in [(NAV_STAMPED, _EXPECTED_NAV_SHA), (GRAPHIFY_STAMPED, GRAPHIFY_SHA256)]:
        _fpath = os.path.join(os.getcwd(), _fname)
        if not os.path.isfile(_fpath):
            print('PRE-FLIGHT FAIL: %s not present in the working directory' % _fname)
            _pre_ok = False
            continue
        _fb = open(_fpath, 'rb').read()
        _fsha = hashlib.sha256(_fb).hexdigest()
        if _fsha != _expected_sha:
            print('PRE-FLIGHT FAIL: %s SHA mismatch (got %s, expected %s)' % (_fname, _fsha, _expected_sha))
            _pre_ok = False
            continue
        try:
            ast.parse(_fb.decode('utf-8'))
            print('pre-flight: %s parses OK, SHA match' % _fname)
        except Exception as _e:
            print('PRE-FLIGHT FAIL: %s does not parse: %r' % (_fname, _e))
            _pre_ok = False
    if not _pre_ok:
        print('ROLLOVER STOPPED — pre-flight failed. Nothing was launched, nothing was deleted, no marker was written.')
        raise RuntimeError('rollover pre-flight failed')
    # --- marker (in-progress) BEFORE the launch-and-cleanup ---
    _marker_rec = {'stamp': STAMP, 'former_notebook': FORMER_NOTEBOOK,
                   'former_notebooks': [FORMER_NOTEBOOK, FORMER_NOTEBOOK_083425, FORMER_NOTEBOOK_092631, FORMER_NOTEBOOK_093527],
                   'new_notebook': NEW_NOTEBOOK,
                   'former_files': ['autopoet_nav.py', 'graphify.py', 'autopoet_nav-20261106-083425.py', 'graphify-20261106-083425.py', 'autopoet_nav-20261106-092631.py', 'graphify-20261106-092631.py', 'autopoet_nav-20261106-093527.py', 'graphify-20261106-093527.py'],
                   'new_files': [NAV_STAMPED, GRAPHIFY_STAMPED],
                   'deploy_sha': DEPLOY_SHA, 'status': 'in-progress', 'writer': 'colab-xterm',
                   'updated_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ'),
                   'note': 'rollover: launch from the new stamped names, cleanup all prior files of the former name (hyphenated and underscored variants)'}
    os.makedirs(SHARED_DIR, exist_ok=True)
    os.makedirs(LOCAL_STATE_DIR, exist_ok=True)
    json.dump(_marker_rec, open(MARKER_SHARED, 'w'), indent=2)
    json.dump(_marker_rec, open(MARKER_LOCAL, 'w'), indent=2)
    print('marker written (in-progress):', MARKER_SHARED, '| local mirror:', MARKER_LOCAL)
    # --- launch: confirm the new stamped files are in place and importable ---
    print('launch: new stamped files present — %s, %s (module stamp %s, graphify SHA match %s)' % (NAV_STAMPED, GRAPHIFY_STAMPED, nav.VERSION_STAMP, GRAPHIFY_SHA256 == nav.GRAPHIFY_SHA256))
    print('launch: this notebook (%s) runs from the new stamped construct; the navigation, endpoints, congrues, and graph cells below use it.' % NEW_NOTEBOOK)
    # --- cleanup: delete ALL prior files of the former name, both places ---
    # Rule (Termux rollover): delete every match for each former root + '*' +
    # extension, EXCEPT the new stamped files, and NEVER the running notebook
    # itself. State files are carried over, never deleted. The clone
    # (./AutoPoET) is a git working copy and is NOT touched.
    def _walk_files(_base, _max_depth):
        _found = []
        if not _base or not os.path.isdir(_base):
            return _found
        for _root, _dirs, _files in os.walk(_base):
            _rel = os.path.relpath(_root, _base)
            _depth = 0 if _rel == '.' else _rel.count(os.sep) + 1
            if _depth >= _max_depth:
                _dirs[:] = []
            _dirs[:] = [d for d in _dirs if not d.startswith('.') and d != 'node_modules']
            for _f in _files:
                _found.append(os.path.join(_root, _f))
        return _found
    def _top_files(_d):
        if not _d or not os.path.isdir(_d):
            return []
        return [os.path.join(_d, _f) for _f in os.listdir(_d) if os.path.isfile(os.path.join(_d, _f))]
    _all_files = []
    _all_files += _top_files(os.getcwd())
    _all_files += _top_files(os.path.join(os.getcwd(), 'state'))
    _all_files += _walk_files(NESTING_1, 2)
    _all_files += _walk_files(NESTING_2, 3)
    _deleted, _kept = [], []
    for _p in sorted(set(_all_files)):
        _base = os.path.basename(_p)
        _is_former = any(_base.startswith(_root) and _base.endswith(_ext) for _root, _ext in FORMER_PATTERNS)
        if not _is_former:
            continue
        if _base in KEEP_NAMES:
            _kept.append(_p)
            continue
        # Never delete a state file or the marker (they do not match the
        # former patterns, but guard anyway) and never delete a file
        # inside the clone.
        if _base in CARRYOVER_FILES or _base == MARKER_NAME:
            _kept.append(_p)
            continue
        if (os.path.sep + 'AutoPoET' + os.path.sep) in _p:
            _kept.append(_p)
            continue
        os.remove(_p)
        _deleted.append(_p)
    print('cleanup: deleted %d former-name file(s):' % len(_deleted))
    for _p in _deleted:
        print('  deleted:', _p)
    print('cleanup: kept %d matching new/carried-over file(s) (not deleted):' % len(_kept))
    for _p in _kept:
        print('  kept:', _p)
    # --- carry over the state files: rewrite the new names/stamp inside, never delete ---
    # The carried-over files may reference the ORIGINAL former names OR the
    # 083425 stamped names — both are rewritten to the new stamped names.
    def _carry(_path):
        if not os.path.isfile(_path):
            return None
        try:
            _rec = json.load(open(_path))
        except Exception as _e:
            print('carry-over: %s could not be read as JSON (%r) — left as-is' % (_path, _e))
            return None
        if not isinstance(_rec, dict):
            return None
        _changed = []
        if 'version_stamp' in _rec and _rec['version_stamp'] != STAMP:
            _rec['version_stamp'] = STAMP
            _changed.append('version_stamp')
        if isinstance(_rec.get('nav_module'), dict):
            _f = _rec['nav_module'].get('file')
            if isinstance(_f, str) and _f.startswith('autopoet_nav') and _f.endswith('.py') and _f != NAV_STAMPED:
                _rec['nav_module']['file'] = NAV_STAMPED
                _changed.append('nav_module.file')
        if isinstance(_rec.get('graphify'), dict):
            _f = _rec['graphify'].get('file')
            if isinstance(_f, str) and _f.startswith('graphify') and _f.endswith('.py') and _f != GRAPHIFY_STAMPED:
                _rec['graphify']['file'] = GRAPHIFY_STAMPED
                _changed.append('graphify.file')
        if isinstance(_rec.get('notebook_in_repo'), str):
            _new = re.sub(r'autoruntime[-_]colab[-_]autopoet[-_]nav[^\s"\']*\.ipynb', NEW_NOTEBOOK, _rec['notebook_in_repo'])
            if _new != _rec['notebook_in_repo']:
                _rec['notebook_in_repo'] = _new
                _changed.append('notebook_in_repo')
        if _changed:
            _tmp = _path + '.tmp'
            json.dump(_rec, open(_tmp, 'w'), indent=2, sort_keys=True)
            os.replace(_tmp, _path)
            print('carry-over: %s rewritten (%s)' % (_path, ', '.join(_changed)))
        else:
            print('carry-over: %s already current (nothing to rewrite)' % _path)
        return _path
    _carry_paths = []
    for _d in [SHARED_DIR, LOCAL_STATE_DIR]:
        for _fn in CARRYOVER_FILES:
            _carry_paths.append(os.path.join(_d, _fn))
    if NESTING_1:
        for _fn in ['congrue-latest.json', 'final-state-latest.json']:
            _carry_paths.append(os.path.join(NESTING_1, _fn))
    for _p in _carry_paths:
        _carry(_p)
    # --- marker (complete) ---
    _marker_rec['status'] = 'complete'
    _marker_rec['updated_utc'] = datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ')
    _marker_rec['deleted'] = _deleted
    _marker_rec['kept'] = _kept
    json.dump(_marker_rec, open(MARKER_SHARED, 'w'), indent=2)
    json.dump(_marker_rec, open(MARKER_LOCAL, 'w'), indent=2)
    print('marker updated (complete):', MARKER_SHARED)
    print('ROLLOVER COMPLETE for stamp %s: launched from %s + %s, deleted %d former-name file(s), carried over the state files.' % (STAMP, NAV_STAMPED, GRAPHIFY_STAMPED, len(_deleted)))
    ROLLOVER_DONE = True
```

## Clone the repo

The repo is cloned shallow and treeless (`--depth 1 --filter=blob:none`):
the tree comes down first and file contents are fetched lazily on
checkout, so the clone is far smaller than the full ~110 MB repo. The
clone takes a few minutes on a slow link — the cell runs until it
finishes. The clone goes into the local Colab runtime (`./AutoPoET`) by
default. Set `CLONE_TO_DRIVE = True` to clone into the shared Drive
folder instead (the phone Termux endpoint then sees the same working
copy); that is slower and is off by default.

If the clone fails, the cell says so plainly and the navigation cells

## Navigate: the commands and the midspace state file

The navigation works on the local clone, through the `autopoet_nav`
functions:

- `list_dir` — list the current folder (dirs first, then files).
- `show_file` — show a file's contents (bounded: 80 lines, 64 KB; binary
  and oversize files are reported, not dumped).
- `enter(name)` — move into a subfolder. `up()` — move to the parent.
  `to_root()` — back to the root. `goto(rel)` — jump to a path inside
  the repo. `tree(depth=2)` — a bounded tree listing. `find(pattern)` —
  find files by name.
- No path ever escapes the repo root: `..` and absolute paths are
  refused.

After every move the state is written to the midspace — the shared
state file `state/nav-state.json` in the shared Drive folder — and to a
local mirror (`./state/nav-state.json`). The state file holds the current
path (relative to the repo root), the path taken, the repo head, and
which terminal wrote it last. The other terminal reads the same file and
resumes at the same place. That is how the two terminals navigate
together, and how navigation survives an ephemeral restart: the state
lives in the shared Drive folder, not in the runtime's memory.


```python
import json, os
if REPO is None:
    print('NO REPO — the clone did not succeed, so there is nothing to navigate. (Nothing fabricated.)')
else:
    # Resume from the midspace state file when it exists (another
    # terminal may have moved since this runtime last ran); else start
    # at the root.
    STATE, _resume_msg = nav.resume(SHARED_NAV_STATE, REPO, writer='colab-xterm')
    print('resume:', _resume_msg)
    STATE['repo_head'] = HEAD
    def _save(note):
        nav.save(STATE, SHARED_NAV_STATE, writer='colab-xterm', note=note)
        nav.save(STATE, LOCAL_NAV_STATE, writer='colab-xterm', note=note)
    # 1. List the root.
    _entries, _m = nav.list_dir(STATE, REPO)
    print('1.', _m)
    print('   dirs:', [e['name'] for e in _entries if e['type'] == 'dir'])
    _save('listed the root')
    # 2. Enter backend/ and list it.
    print('2.', nav.enter(STATE, REPO, 'backend'))
    _entries, _m = nav.list_dir(STATE, REPO)
    print('   ', _m, [(e['name'], e['type'], e['size']) for e in _entries])
    _save('entered backend')
    # 3. Show backend/server.py (small, readable: 59 lines).
    _r, _m = nav.show_file(STATE, REPO, 'backend/server.py')
    print('3.', _m)
    print('--- backend/server.py (first 10 lines) ---')
    print('\n'.join((_r['content'] if _r else '').splitlines()[:10]))
    _save('showed backend/server.py')
    # 4. Back up to the root.
    print('4.', nav.up(STATE, REPO))
    _save('up to the root')
    # 5. Enter tests/ and show its file.
    print('5.', nav.enter(STATE, REPO, 'tests'))
    _r, _m = nav.show_file(STATE, REPO, 'tests/test_extraction.py', max_lines=5)
    print('   ', _m, '| lines total:', _r['lines_total'] if _r else None)
    _save('entered tests, showed test_extraction.py')
    # 6. Back to the root.
    print('6.', nav.to_root(STATE))
    _save('back at the root')
    # 7. Is this notebook itself in the repo yet?
    _nb_path, _nb_msg = nav.find_notebook(STATE, REPO)
    print('7. notebook in repo:', _nb_msg)
    # 8. The Drive archive copies — separate, clearly labelled sources.
    # The Drive copies are snapshots/archives, NOT live-synced with the
    # GitHub repo. They are listed here so the user can see both; the
    # navigation above uses the git clone (the live repo).
    print('8. Drive archive copies (snapshots — NOT live-synced with GitHub):')
    for _a in nav.find_drive_archives(MYDRIVE, NESTING_1):
        if _a['path']:
            print('   ', _a['name'], '| snapshot', _a['modified'], '| found in this mount at', _a['path'])
        else:
            print('   ', _a['name'], '| snapshot', _a['modified'], '| not found in this mount (folder ID', _a['folder_id'] + ')')
    # 9. The Drive copy — navigate BOTH sources, each labelled.
    # The Drive copy is a snapshot from Oct 11, 2026, NOT live-synced
    # with the GitHub clone. It gets its OWN state file
    # (state/nav-state-drive.json); the two sources are never mixed.
    print('9. Drive copy (source: drive-copy):')
    CLONED_REPOS, _cr_cands, _cr_note = nav.find_cloned_repos(MYDRIVE)
    print('   ', _cr_note)
    for _c in _cr_cands:
        print('    candidate:', _c['path'], '|', _c['entries'], 'entries | AutoPoET folder:', _c['has_autopoet'])
    DRIVE_REPO = None
    DSTATE = None
    if CLONED_REPOS:
        _repos = nav.list_cloned_repos(CLONED_REPOS)
        print('    repos in cloned_repositories:', [r['name'] for r in _repos if r['type'] == 'dir'])
        _drive_autopoet = os.path.join(CLONED_REPOS, 'AutoPoET')
        if os.path.isdir(_drive_autopoet):
            DRIVE_REPO = _drive_autopoet
            DSTATE, _dmsg = nav.resume(SHARED_NAV_STATE_DRIVE, DRIVE_REPO, writer='colab-xterm', source='drive-copy')
            print('    resume (drive-copy):', _dmsg)
            def _dsave(note):
                nav.save(DSTATE, SHARED_NAV_STATE_DRIVE, writer='colab-xterm', note=note)
                nav.save(DSTATE, LOCAL_NAV_STATE_DRIVE, writer='colab-xterm', note=note)
            _entries, _m = nav.list_dir(DSTATE, DRIVE_REPO)
            print('    drive-copy root:', _m)
            _ddirs = [e['name'] for e in _entries if e['type'] == 'dir']
            _dfiles = [e['name'] for e in _entries if e['type'] == 'file']
            print('    dirs:', _ddirs, '| files:', _dfiles[:10])
            _dsave('listed the Drive copy root')
            if _dfiles:
                _r, _m = nav.show_file(DSTATE, DRIVE_REPO, _dfiles[0], max_lines=5)
                print('    drive-copy', _dfiles[0] + ':', _m, '| lines total:', _r['lines_total'] if _r else None)
                _dsave('showed %s in the Drive copy' % _dfiles[0])
            if _ddirs:
                print('   ', nav.enter(DSTATE, DRIVE_REPO, _ddirs[0]))
                _entries2, _m2 = nav.list_dir(DSTATE, DRIVE_REPO)
                print('    in', _ddirs[0] + ':', _m2)
                _dsave('entered %s in the Drive copy' % _ddirs[0])
                print('   ', nav.to_root(DSTATE))
                _dsave('back at the Drive copy root')
            print('    drive-copy state source:', DSTATE['source'], '| current:', DSTATE['current'] or '(root)', '| path taken:', DSTATE['history'])
        else:
            print('    AutoPoET folder not in this cloned_repositories copy — Drive-copy navigation skipped.')
    else:
        print('    cloned_repositories not found in this mount — Drive-copy navigation skipped (nothing fabricated).')
    _save('navigation demo complete (both sources)')
    # 10. AI-Studio-SmartRepo — navigable the same way as AutoPoET.
    # The app for GitHub parsing, code comparing, and sandboxing (the
    # user's framing), built in AIStudio. It gets its OWN state file
    # (state/nav-state-smartrepo.json); its GitHub URL is not recorded
    # (not given) — only the Drive path and file ID.
    print('10. AI-Studio-SmartRepo (source: drive-copy-smartrepo):')
    SSTATE = None
    if CLONED_REPOS:
        _sr = os.path.join(CLONED_REPOS, 'AI-Studio-SmartRepo')
        if os.path.isdir(_sr):
            SSTATE, _smsg = nav.resume(SHARED_NAV_STATE_SMARTREPO, _sr, writer='colab-xterm', source='drive-copy-smartrepo')
            print('    resume (drive-copy-smartrepo):', _smsg)
            print('    Drive path:', _sr, '| folder ID:', nav.AI_STUDIO_SMARTREPO_FOLDER_ID, '| modified:', nav.AI_STUDIO_SMARTREPO_MODIFIED)
            def _ssave(note):
                nav.save(SSTATE, SHARED_NAV_STATE_SMARTREPO, writer='colab-xterm', note=note)
                nav.save(SSTATE, LOCAL_NAV_STATE_SMARTREPO, writer='colab-xterm', note=note)
            _entries, _m = nav.list_dir(SSTATE, _sr)
            print('    SmartRepo root:', _m)
            _sdirs = [e['name'] for e in _entries if e['type'] == 'dir']
            _sfiles = [e['name'] for e in _entries if e['type'] == 'file']
            print('    dirs:', _sdirs, '| files:', _sfiles[:10])
            _ssave('listed the SmartRepo root')
            # Full-stack structure — verified from the actual folder,
            # not hardcoded: front (app/components/hooks), middle (lib),
            # backend (Firebase/Firestore).
            SMARTREPO_STRUCTURE = nav.smartrepo_structure(_sr)
            print('    full-stack structure (verified from this folder):')
            for _part, _info in SMARTREPO_STRUCTURE.items():
                print('      ', _part, ':', 'present (' + '/'.join(_info['found']) + ')' if _info['present'] else 'NOT FOUND in this copy')
            # MCP and A2A: CONFIRMED by the code itself in the real
            # Colab run of the 083425 notebook (2026-11-06) — MCP in
            # app/api/mcp/route.ts (16), lib/agentic-context.ts (10),
            # and more; A2A in app/api/a2a/route.ts (13) and more.
            # The navigation re-runs the search on every run and
            # reports what it finds.
            SMARTREPO_MCP_A2A = nav.smartrepo_search_mcp_a2a(_sr)
            print('    MCP/A2A search (lib/ + app/, file names + contents):', SMARTREPO_MCP_A2A['note'])
            print('      mcp hits:', SMARTREPO_MCP_A2A['mcp'] or '(none)')
            print('      a2a hits:', SMARTREPO_MCP_A2A['a2a'] or '(none)')
            _ssave('checked SmartRepo structure and MCP/A2A')
            if _sfiles:
                _r, _m = nav.show_file(SSTATE, _sr, _sfiles[0], max_lines=5)
                print('    SmartRepo', _sfiles[0] + ':', _m, '| lines total:', _r['lines_total'] if _r else None)
                _ssave('showed %s in SmartRepo' % _sfiles[0])
            if _sdirs:
                print('   ', nav.enter(SSTATE, _sr, _sdirs[0]))
                _entries2, _m2 = nav.list_dir(SSTATE, _sr)
                print('    in', _sdirs[0] + ':', _m2)
                _ssave('entered %s in SmartRepo' % _sdirs[0])
                print('   ', nav.to_root(SSTATE))
                _ssave('back at the SmartRepo root')
            print('    SmartRepo state source:', SSTATE['source'], '| current:', SSTATE['current'] or '(root)', '| path taken:', SSTATE['history'])
        else:
            print('    AI-Studio-SmartRepo folder not in this cloned_repositories copy — SmartRepo navigation skipped.')
    else:
        print('    cloned_repositories not found in this mount — SmartRepo navigation skipped (nothing fabricated).')
    print('--- midspace state file (shared, github-clone) ---')
    print(open(SHARED_NAV_STATE).read())
    if DSTATE is not None:
        print('--- midspace state file (shared, drive-copy) ---')
        print(open(SHARED_NAV_STATE_DRIVE).read())
    if SSTATE is not None:
        print('--- midspace state file (shared, drive-copy-smartrepo) ---')
        print(open(SHARED_NAV_STATE_SMARTREPO).read())
    print('path taken (github-clone):', STATE['history'])
    print('path taken (drive-copy):', DSTATE['history'] if DSTATE is not None else '(not navigated)')
    print('path taken (drive-copy-smartrepo):', SSTATE['history'] if SSTATE is not None else '(not navigated)')
```

## Stop / cleanup — the last cell propagates its output in the cell

This is the last cell, and it is the propagation point. It parks the
navigation state and stops — and its output PROPAGATES in the cell: the
final state (the endpoints, the links, the nestings, the navigation
state, the status) is not just printed and lost. It is written to the
shared state file (`state/final-state.json` in the shared Drive folder,
with a marker `final-state-latest.json` at the top level) AND displayed
in the cell itself, so the output carries forward from the cell outward
to the shared places and the other terminal endpoints. The construct
held within propagates its output outward, through this cell.

This notebook starts no socket and no background process, so there is
nothing to shut down — the stop cell writes the final state (marked
"parked") to the midspace and the local mirror, and reports what
survives and what is lost.

What survives: the nav-state file in the shared Drive folder (the other
terminal resumes from it), the Drive-copy nav-state file
(`state/nav-state-drive.json`), the SmartRepo nav-state file
(`state/nav-state-smartrepo.json`), the congrue record and its marker
(`state/congrue.json`, `congrue-latest.json`), the final-state record
and its marker (`state/final-state.json`, `final-state-latest.json`),
the rollover record (`state/rollover.json`), the link graph files (`state/link-graph.graphml`, `state/link-graph.json`,
`state/link-graph.png`, plus graphify's `.graphml`/`.gexf` when written),
the local mirror, `state/endpoints.json`, and the local clone
(`./AutoPoET`). What is lost: the in-memory navigation object (`STATE`)
and this session's place in the notebook, and the Tailscale join (tailnet membership lives in the runtime — re-run the Tailscale-join cell with a key after a restart) — re-run the cells in order to
rebuild them; the state file is what makes the rebuild land in the same
place.
