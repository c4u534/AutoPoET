```python
import hashlib, math, re

# ---------- core measures (the derivation) ----------
def whitespace_trace(text):
    voids, run = [], 0
    for ch in text:
        if ch == ' ':
            run += 1
        elif run: voids.append(run); run = 0
    if run: voids.append(run)
    lines = text.split('\n')
    return {'chars': len(text), 'words': len(text.split()),
            'whitespace': sum(1 for c in text if c.isspace()),
            'void_widths': voids, 'lines': len(lines),
            'line_void_counts': [ln.count(' ') for ln in lines]}

def volume(text):
    if not text: return None
    return sum(1 for c in text if c.isspace()) / len(text)

def volume_level(text):
    v = volume(text)
    if v is None: return 'VL-0 null'
    if v < 0.20:  return 'VL-1 word-carried'
    if v < 0.80:  return 'VL-2 open band'
    return 'VL-3 media-locked'

def parity_fold(text):
    f = 0
    for b in text.encode(): f ^= b
    return format(f, '08b')

def parity_hash(text):
    f = 0
    for b in text.encode(): f ^= b
    return hashlib.sha256(text.encode() + bytes([f])).hexdigest()

def addressment(text):
    h256 = hashlib.sha256(text.encode()).hexdigest()
    h512 = hashlib.sha512(text.encode()).hexdigest()
    return f'{h256}:{h512}', ':'.join(h256[:32][i:i+4] for i in range(0, 32, 4))

def media_of(text):
    return '\n'.join(' ' * ln.count(' ') for ln in text.split('\n'))

def invert_lines(text):
    def invert(line):
        out, last = [], 0
        for tok in re.findall(r'\S+| +', line):
            if tok.startswith(' '): out.append(' ' * (last if len(tok) == 1 else len(tok)))
            else: out.append('\u2588'); last = len(tok)
        return ''.join(out)
    return '\n'.join(invert(ln) for ln in text.split('\n'))

# ---------- the challenge harness (red/blue, built into the construct) ----------
def run_challenge(name, contract_tests, attacks):
    passed, failed = [], []
    for label, fn in contract_tests + attacks:
        try:
            fn(); passed.append(label)
        except AssertionError as e:
            failed.append((label, str(e) or 'assert failed'))
    receipt = {'component': name, 'tests': len(contract_tests) + len(attacks),
               'passed': len(passed), 'failed': failed,
               'verdict': 'CHALLENGE-CLEAN' if not failed else 'OPEN'}
    print(f"{name}: {len(passed)}/{receipt['tests']} green -> {receipt['verdict']}")
    for label, err in failed: print(f'   OPEN [{label}]: {err}')
    return receipt

RECEIPTS = []
print('core measures + challenge harness defined')
```

## Component 11 — the IPv6 relation (Cause's statement, with the standard correspondences beside it)

Cause's statement, as arrived: standard IPv6 has hashing architectures and parity hashing sequential, as they
relate and derive to memory addressment routings in the IPv6 construct, and positional sequences in the IPv6 hashing.

The checked standard facts that stand beside it (public record, unmerged):
- **IPv6 routing is positional.** Longest-prefix match walks the address's bit sequence in order; positions carry
  fixed roles — routing prefix, subnet, interface identifier. Where a bit sits decides where the packet goes.
- **Hash-derived identifiers are standard IPv6 practice.** RFC 7217 generates the interface identifier as a hash —
  F(prefix, interface, network ID, DAD counter, secret key) — placed positionally in the low 64 bits; temporary
  addresses (RFC 4941, now 8981) are likewise hash-derived and re-derived in sequence. A counter inside the hash
  makes the derivation sequential: same inputs, next count, next identifier.
- **IPv6 carries no header checksum.** The parity-like check was removed from the carrier layer and pushed to the
  ends / upper layers — as here, where parity rides in the envelope's echoback, not in the carrier addressment.
- **This construct's form of the same pattern:** parity hashing is sequential (the XOR fold is struck first, then
  SHA-256 is taken over raw + fold byte), and the hashing is positional — the leading 128 bits form the block,
  the leading hextets form the registry_id, and routing in the memory addressment goes by that position
