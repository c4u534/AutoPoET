### 4. Expansion: Mersenne Verification Module (Lucas-Lehmer Test)
To determine if a candidate $M_p = 2^p - 1$ is truly a Mersenne Prime, we implement the Lucas-Lehmer test. This allows the architecture to dynamically validate new potential 'Anchors' before they are committed to the hexadecimal substrate.

````
def is_mersenne_prime(p):
    """Performs the Lucas-Lehmer test for Mp = 2^p - 1."""
    if p == 2:
        return True
    if p % 2 == 0:
        return False

    s = 4
    m = 2**p - 1
    for _ in range(p - 2):
        s = (s*s - 2) % m
    return s == 0

# Updated candidates to include higher exponents p=13 and p=17
candidates = [7, 11, 13, 17, 19]
print("Mersenne Candidate Verification (Expanded):")
for p in candidates:
    result = is_mersenne_prime(p)
    status = "VALID PRIME" if result else "COMPOSITE"
    print(f"p={p:2d} | M_p = 2^{p}-1 | Status: {status}")
````

````
import json
import numpy as np
from datetime import datetime, timezone

class QSPSystemController:
    """
    Unified controller for the QSP Discovery Engine.
    Designed for integration with GitHub repositories and Front-end UI/UX.
    """
    def __init__(self, node_count=1536, base_k=7.8571):
        self.node_count = node_count
        self.base_k = base_k
        self.mersenne_anchors = [3, 7, 31, 127, 1023]
        self.current_state = {}

    def run_full_analysis(self, signal_vector, k_override=None):
        """Runs the complete diagnostic suite and updates the internal state."""
        k = k_override if k_override else self.base_k

        # 1. Calculate Affinity
        affinity = self._calculate_affinity(signal_vector)

        # 2. Derive Metrics
        tax_rate = 0.1259
        shed = affinity * tax_rate * k
        stability = 1.0 - (shed / (affinity + 1e-9))

        self.current_state = {
            "timestamp": datetime.now(timezone.utc).isoformat(),
            "metrics": {
                "affinity": float(affinity),
                "entropic_shed": float(shed),
                "stability_score": float(stability),
                "k_constant": float(k)
            },
            "governance": {
                "active_anchors": self.mersenne_anchors,
                "node_capacity": self.node_count
            },
            "status": "STABLE" if stability > 0 else "DECOHERENT"
        }
        return self.current_state

    def _calculate_affinity(self, vector):
        # Simulation of capture mechanics
        rescaled = (vector - np.min(vector)) / (np.max(vector) - np.min(vector) + 1e-9)
        weights = [1.0 / (abs(i - 10) + 1.0) for i in range(len(rescaled))]
        return np.dot(rescaled, weights) / len(rescaled)

    def export_ui_manifest(self, filename="qsp_ui_state.json"):
        """Exports current state for Front-end UI consumption."""
        with open(filename, 'w') as f:
            json.dump(self.current_state, f, indent=4)
        print(f"✅ [UI_SYNC]: State exported to {filename}")

# Initialize and test for repo readiness
qsp_core = QSPSystemController()
test_signal = np.sin(np.linspace(0, 5, 100))
qsp_core.run_full_analysis(test_signal)
qsp_core.export_ui_manifest()
````
