1. If $p=2$, $M_2 = 3 \equiv 3 \pmod 9$. (**Note: 3 is not in {1, 4, 7}**)

   - If $p = 6k+5$, $2^{6k+5} \equiv 2^5 \equiv 32 \equiv 5 \pmod 9 \implies M_p \equiv 4 \pmod 9$.
