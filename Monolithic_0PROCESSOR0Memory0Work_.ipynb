```
import json
import time

# --- System Check & Mock Fallback ---
try:
    substrate = AutoPoetMonolithicSubstrate()
except NameError:
    print("Note: AutoPoetMonolithicSubstrate not found. Using Mock Substrate for demonstration.")
    class MockSubstrate:
        def execute_cycle(self, cid, prompt):
            return {
                'a2ui_payload': json.dumps({
                    "surfaceId": f"mock_{cid}",
                    "components": [
                        {"id": "c1", "type": "Card", "props": {"title": "Mock Web Title"}},
                        {"id": "t1", "type": "TextField", "props": {"value": "Mock Value"}}
                    ]
                })
            }
        def close(self):
            pass
    substrate = MockSubstrate()

class WebFrontendAdapter:
    """Adapts A2UI payloads for standard Web JSON consumption."""
    @staticmethod
    def transform_for_react(a2ui_json: str):
        data = json.loads(a2ui_json)
        mapping = {
            "Card": "MuiCard",
            "TextField": "MuiTypography"
        }
        web_components = []
        for comp in data.get("components", []):
            web_components.append({
                "component": mapping.get(comp["type"], "div"),
                "key": comp["id"],
                "properties": comp["props"]
            })
        return {
            "status": "success",
            "surface_id": data["surfaceId"],
            "dom_tree": web_components,
            "timestamp": time.time()
        }

# --- Execute Adaptation ---
raw_cycle = substrate.execute_cycle(999, "WEB_ADAPTATION_TEST")
adapter = WebFrontendAdapter()
web_ready_payload = adapter.transform_for_react(raw_cycle['a2ui_payload'])

print("Adapted Web Payload for Frontend consumption:")
print(json.dumps(web_ready_payload, indent=2))
substrate.close()
```

### Part IV: Multi-Framework Mapping Registry
This registry provides a unified interface to map the internal A2UI types (`Card`, `TextField`, etc.) to specific component tags for different frontend ecosystems.

```
class MultiFrameworkRegistry:
    """Central registry for mapping A2UI types to various frontend frameworks."""

    # Mapping definitions for different targets
    TARGETS = {
        "react_mui": {
            "Card": "MuiCard",
            "TextField": "MuiTextField",
            "Button": "MuiButton"
        },
        "vue_vuetify": {
            "Card": "v-card",
            "TextField": "v-text-field",
            "Button": "v-btn"
        },
        "svelte": {
            "Card": "Card",
            "TextField": "TextInput",
            "Button": "Button"
        },
        "web_components": {
            "Card": "a2ui-card",
            "TextField": "a2ui-text-input",
            "Button": "a2ui-button"
        }
    }

    @classmethod
    def add_target(cls, name, mapping):
        """Dynamically add a new framework mapping at runtime."""
        cls.TARGETS[name] = mapping

    @classmethod
    def map_payload(cls, a2ui_payload, target="react_mui"):
        data = json.loads(a2ui_payload)
        mapping = cls.TARGETS.get(target, {})

        output = []
        for comp in data.get("components", []):
            output.append({
                "framework_tag": mapping.get(comp["type"], "div"),
                "internal_id": comp["id"],
                "props": comp["props"]
            })
        return output

# --- Example: Svelte and Dynamic Mapping ---
registry = MultiFrameworkRegistry()

print("--- SVELTE MAPPING ---")
print(json.dumps(registry.map_payload(raw_cycle['a2ui_payload'], "svelte"), indent=2))

# Adding a new target (e.g., Tailwind/HTML) dynamically
registry.add_target("tailwind", {
    "Card": "div.shadow-lg.p-4",
    "TextField": "input.border-b"
})

print("\n--- DYNAMIC TAILWIND MAPPING ---")
print(json.dumps(registry.map_payload(raw_cycle['a2ui_payload'], "tailwind"), indent=2))
```

```
def add_target_from_json(registry_class, config_json):
    """Parses a JSON config to add a new framework target mapping."""
    config = json.loads(config_json)
    for target_name, mapping in config.items():
        registry_class.add_target(target_name, mapping)
        print(f"[Registry] Successfully added target: {target_name}")

# --- Example: Adding a Mobile framework via JSON ---
mobile_config = """
{
    "react_native": {
        "Card": "View.Card",
        "TextField": "TextInput",
        "Button": "TouchableOpacity"
    }
}
"""

add_target_from_json(MultiFrameworkRegistry, mobile_config)

# Verify the new mapping
mobile_test = registry.map_payload(raw_cycle['a2ui_payload'], "react_native")
print("\n--- REACT NATIVE (JSON CONFIGURED) MAPPING ---")
print(json.dumps(mobile_test, indent=2))
```

### Part V: Automated Batch Export for Multi-Framework Testing
This section implements a utility to sweep through all registered frameworks and generate export artifacts for each simultaneously.

```
import os

def export_all_frameworks(registry, a2ui_payload, base_filename="frontend_payload"):
    """Maps and exports the payload for every target in the registry."""
    exported_files = []

    for target in registry.TARGETS.keys():
        # Map the payload for the specific target
        mapped_data = registry.map_payload(a2ui_payload, target=target)

        # Define file path
        filename = f"{base_filename}_{target}.json"

        # Write to JSON
        with open(filename, 'w') as f:
            json.dump(mapped_data, f, indent=2)

        exported_files.append(filename)
        print(f"[Export] Generated: {filename}")

    return exported_files

# --- Trigger Batch Export ---
print("Starting batch export for all registered frameworks...\n")
all_files = export_all_frameworks(registry, raw_cycle['a2ui_payload'])

# Optional: Zip and download all generated payloads
import zipfile
zip_name = 'all_framework_payloads.zip'
with zipfile.ZipFile(zip_name, 'w') as zipf:
    for file in all_files:
        zipf.write(file)

print(f"\nAll payloads bundled into: {zip_name}")
try:
    from google.colab import files
    files.download(zip_name)
except Exception:
    print("Download available in the file browser sidebar.")
```
