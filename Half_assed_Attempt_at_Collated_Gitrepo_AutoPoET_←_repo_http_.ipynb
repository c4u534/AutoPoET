### Component 01: The Self-Oscillating Two-Mass Glottal Engine

* **UID:** `APT-VOC-001`
* **Source Notebook:** `interactive_two_mass_model_glottal_pulse_source...`
* **Archetype:** `SUBSTRATE_SOURCE`
* **Subsystem Focus:** Non-linear vocal fold biomechanics & self-sustaining mucosal wave oscillations.
* **Mathematical Invariant:**

$$m_1 \ddot{x}_1 + r_1 \dot{x}_1 + k_1 x_1 + k_c (x_1 - x_2) + f_{c1}(x_1) = F_1(P_{g1})$$


$$m_2 \ddot{x}_2 + r_2 \dot{x}_2 + k_2 x_2 + k_c (x_2 - x_1) + f_{c2}(x_2) = F_2(P_{g2})$$


$$P_{g1} = P_s - \frac{1}{2} \rho \left(\frac{U_g}{A_{g1}}\right)^2 (1 + k_e), \quad U_g(t) = \sqrt{\frac{2 P_s}{\rho \left(\frac{1+k_e}{A_{g1}^2} + \frac{1}{A_{g2}^2}\right)}}$$


* **Inputs:** `P_sub: float` (Subglottal lung pressure, $800\text{ Pa}$), `f0_target: float` (Pitch in Hz), `duration: float` (Seconds).
* **Outputs:** `ug_flow: np.ndarray` (Instantaneous acoustic volume velocity $U_g(t)$ in $\text{m}^3/\text{s}$).
* **Dependencies:** `numpy`, `dataclasses`
* **Composable Tags:** `glottis, two_mass, biomechanics, bernoulli, limit_cycle, vocal_folds`
* **Silo Reference:** [NB- B00 - GeoMersPhonetiQmetricsModality - Word Sound Syllables Phonetic Sound Frequencies Patterning Geometry 44 English Phonemes](https://gemini.google.com/notebook/295cf8f8-f74c-4a19-b635-20a653ea660b)

````
from dataclasses import dataclass
import numpy as np


@dataclass
class TwoMassParameters:
  m1: float = 0.125e-3  # Lower mass (kg)
  m2: float = 0.025e-3  # Upper mass (kg)
  d1: float = 0.25e-2  # Lower thickness (m)
  d2: float = 0.05e-2  # Upper thickness (m)
  lg: float = 1.4e-2  # Glottal fold length (m)
  k1: float = 80.0  # Lower stiffness (N/m)
  k2: float = 8.0  # Upper stiffness (N/m)
  kc: float = 25.0  # Vertical shear coupling stiffness (N/m)
  x01: float = 0.01e-2  # Neutral half-width lower (m)
  x02: float = 0.01e-2  # Neutral half-width upper (m)
  c1: float = 240.0  # Collision stiffness 1
  c2: float = 24.0  # Collision stiffness 2
  zeta1: float = 0.15  # Damping ratio lower
  zeta2: float = 0.40  # Damping ratio upper
  p_sub: float = 850.0  # Lung pressure (Pa) ~8.5 cm H2O
  rho: float = 1.184  # Air density (kg/m^3)


class SovereignTwoMassGlottalSource:
  """Executes 4th-Order Runge-Kutta numerical integration of self-oscillating

  vocal folds under Bernoulli pressure drops and mucosal shear coupling.
  """

  def __init__(
      self, params: TwoMassParameters = None, sample_rate: int = 44100
  ):
    self.p = params or TwoMassParameters()
    self.fs = sample_rate
    self.dt = 1.0 / sample_rate
    self.state = np.zeros(4, dtype=np.float64)  # [x1, v1, x2, v2]

  def _derivatives(
      self, s: np.ndarray, q_tension: float, p_sub: float
  ) -> np.ndarray:
    x1, v1, x2, v2 = s
    p = self.p

    # Biomechanical scaling via cricothyroid tension factor Q
    m1 = p.m1 / q_tension
    m2 = p.m2 / q_tension
    k1 = p.k1 * (q_tension**2)
    k2 = p.k2 * (q_tension**2)
    kc = p.kc * (q_tension**2)
    x01 = p.x01 / np.sqrt(q_tension)
    x02 = p.x02 / np.sqrt(q_tension)

    # Glottal orifice areas
    ag1 = 2.0 * p.lg * max(0.0, x1 + x01)
    ag2 = 2.0 * p.lg * max(0.0, x2 + x02)

    # Bernoulli aerodynamics
    ke = 0.12
    if ag1 > 1e-8 and ag2 > 1e-8:
      denom = ((1.0 + ke) / (ag1**2)) + (1.0 / (ag2**2))
      ug = np.sqrt(max(0.0, 2.0 * p_sub / (p.rho * denom)))
      pg1 = p_sub - 0.5 * p.rho * ((ug / ag1) ** 2) * (1.0 + ke)
    else:
      ug = 0.0
      pg1 = p_sub if ag1 <= 1e-8 else 0.0

    # Nonlinear collision restoration forces
    fc1 = p.c1 * (x1 + x01) if (x1 + x01) < 0.0 else 0.0
    fc2 = p.c2 * (x2 + x02) if (x2 + x02) < 0.0 else 0.0

    # Mechanical damping
    r1 = 2.0 * p.zeta1 * np.sqrt(m1 * k1)
    r2 = 2.0 * p.zeta2 * np.sqrt(m2 * k2)

    a1 = (p.lg * p.d1 * pg1 - r1 * v1 - k1 * x1 - kc * (x1 - x2) - fc1) / m1
    a2 = (-r2 * v2 - k2 * x2 - kc * (x2 - x1) - fc2) / m2
    return np.array([v1, a1, v2, a2])

  def generate_flow(
      self, duration: float, f0_target: float = 120.0
  ) -> np.ndarray:
    n_samples = int(duration * self.fs)
    ug_out = np.zeros(n_samples, dtype=np.float64)
    q_tension = max(0.4, f0_target / 120.0)

    for n in range(n_samples):
      # 4th-Order Runge-Kutta Step
      k1 = self._derivatives(self.state, q_tension, self.p.p_sub)
      k2 = self._derivatives(
          self.state + 0.5 * self.dt * k1, q_tension, self.p.p_sub
      )
      k3 = self._derivatives(
          self.state + 0.5 * self.dt * k2, q_tension, self.p.p_sub
      )
      k4 = self._derivatives(
          self.state + self.dt * k3, q_tension, self.p.p_sub
      )
      self.state += (self.dt / 6.0) * (k1 + 2.0 * k2 + 2.0 * k3 + k4)

      # Extract instantaneous volume velocity
      x1, _, x2, _ = self.state
      ag1 = 2.0 * self.p.lg * max(0.0, x1 + self.p.x01 / np.sqrt(q_tension))
      ag2 = 2.0 * self.p.lg * max(0.0, x2 + self.p.x02 / np.sqrt(q_tension))
      if ag1 > 1e-8 and ag2 > 1e-8:
        denom = (1.12 / (ag1**2)) + (1.0 / (ag2**2))
        ug_out[n] = np.sqrt(max(0.0, 2.0 * self.p.p_sub / (self.p.rho * denom)))
      else:
        ug_out[n] = 0.0

    return ug_out
````

### Component 02: The Coupled 2.5D BEM-Webster Lossy Horn Propagator

* **UID:** `APT-VOC-002`
* **Source Notebook:** `Implement_Boundary_Element_Method_(BEM)_solvers_f_.ipynb`
* **Archetype:** `MODAL_WAVEGUIDE`
* **Subsystem Focus:** Viscothermal boundary layers, yielding vocal tract wall admittance, and longitudinal wave transfer.
* **Mathematical Invariant:**

$$\frac{1}{S(z)} \frac{d}{dz}\left(S(z) \frac{dP(z,\omega)}{dz}\right) + \Gamma^2(z,\omega) P(z,\omega) = 0$$


$$\Gamma^2(z,\omega) = \frac{\omega^2}{c^2} \left[1 + (\eta_v + \eta_t) \frac{\mathcal{P}(z)}{S(z)} \frac{1}{\sqrt{\omega}}\right] - i \omega \rho \frac{\mathcal{P}(z)}{S(z)} Y_w(z)$$


* **Inputs:** `slices: List[BEMSlice2D]` (Array of cross-sectional area contours from glottis $z=0$ to lips $z=17.5\text{ cm}$), `freqs: np.ndarray` ($50\text{ Hz}–4500\text{ Hz}$).
* **Outputs:** `H_mag_dB: np.ndarray` (Vocal Tract Transfer Function $\vert{}H(f)\vert{}$), `formant_peaks: List[float]` ($F_1, F_2, F_3, F_4$ in Hz).
* **Dependencies:** `numpy`, `scipy.special`, `scipy.signal`
* **Composable Tags:** `bem, webster_horn, boundary_elements, viscothermal_loss, transfer_matrix, vocal_tract`
* **Silo Reference:** [NB- B00 - GeoMersPhonetiQmetricsModality...](https://gemini.google.com/notebook/295cf8f8-f74c-4a19-b635-20a653ea660b)

````
from dataclasses import dataclass
import numpy as np
import scipy.special as sp


@dataclass
class BEMSlice2D:
  z_pos: float  # Axial distance from glottis (m)
  nodes: np.ndarray  # Coordinate array (N, 2) in meters

  def compute_invariants(self) -> tuple[float, float, float, float]:
    """Shoelace formula for Area S, boundary Perimeter P, Hydraulic Radius Rh, and Compactness."""
    x = self.nodes[:, 0]
    y = self.nodes[:, 1]
    x_next = np.roll(x, -1)
    y_next = np.roll(y, -1)
    area = 0.5 * np.abs(np.sum(x * y_next - x_next * y))
    perimeter = np.sum(np.hypot(x_next - x, y_next - y))
    rh = (2.0 * area) / (perimeter + 1e-12)
    compactness = (perimeter**2) / (4.0 * np.pi * area + 1e-12)
    return area, perimeter, rh, compactness


class CoupledBEMWebsterEngine:
  """Cascades 2D BEM-parameterized vocal tract slices into a lossy transmission matrix chain."""

  def __init__(self, slices: list[BEMSlice2D]):
    self.slices = slices
    self.n = len(slices)
    self.c_sound = 343.0
    self.rho_air = 1.184
    self.mu_air = 1.86e-5
    self.k_thermal = 0.026
    self.cp_air = 1005.0
    self.gamma_air = 1.40
    self.r_wall = 1600.0
    self.m_wall = 1.5
    self.c_wall = 3.0e-5
    self._extract_geometry()

  def _extract_geometry(self):
    self.areas = np.zeros(self.n)
    self.perimeters = np.zeros(self.n)
    self.rh = np.zeros(self.n)
    for k, s in enumerate(self.slices):
      a, p, rh, _ = s.compute_invariants()
      self.areas[k] = a
      self.perimeters[k] = p
      self.rh[k] = rh

  def _radiation_impedance(self, omega: float) -> complex:
    """Levine-Schwinger baffled piston radiation impedance at the lip aperture."""
    s_lip = self.areas[-1]
    ka = (omega / self.c_sound) * np.sqrt(s_lip / np.pi)
    r_rad = (self.rho_air * self.c_sound / s_lip) * (
        1.0 - sp.jv(1, 2 * ka) / (ka + 1e-12)
    )
    x_rad = (self.rho_air * self.c_sound / s_lip) * (
        sp.struve(1, 2 * ka) / (ka + 1e-12)
    )
    return complex(r_rad, x_rad)

  def solve_vttf(self, freqs: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
    h_db = np.zeros(len(freqs))
    for f_idx, freq in enumerate(freqs):
      omega = 2.0 * np.pi * freq
      k0 = omega / self.c_sound
      delta_v = np.sqrt(2.0 * self.mu_air / (self.rho_air * omega + 1e-12))
      y_w = 1.0 / (
          self.r_wall
          + 1j * omega * self.m_wall
          + 1.0 / (1j * omega * self.c_wall + 1e-15)
      )

      t_total = np.eye(2, dtype=np.complex128)
      for k in range(self.n - 1):
        dz = self.slices[k + 1].z_pos - self.slices[k].z_pos
        s_k = self.areas[k]
        p_k = self.perimeters[k]
        rh_k = self.rh[k]

        alpha_loss = (
            1.0 / (2.0 * rh_k * self.c_sound)
        ) * np.sqrt(omega) * (
            np.sqrt(self.mu_air / (2.0 * self.rho_air))
            + (self.gamma_air - 1.0)
            * np.sqrt(self.k_thermal / (2.0 * self.rho_air * self.cp_air))
        )
        gamma_k = (
            alpha_loss
            + 1j * k0
            + (self.rho_air * self.c_sound * p_k / (2.0 * s_k)) * y_w
        )
        z0_k = (self.rho_air * self.c_sound / s_k) * (
            1.0 + (1.0 - 1j) * delta_v / (2.0 * rh_k)
        )

        cosh_g = np.cosh(gamma_k * dz)
        sinh_g = np.sinh(gamma_k * dz)
        tk = np.array(
            [[cosh_g, z0_k * sinh_g], [sinh_g / z0_k, cosh_g]],
            dtype=np.complex128,
        )
        t_total = t_total @ tk

      z_rad = self._radiation_impedance(omega)
      h_omega = z_rad / (t_total[0, 0] * z_rad + t_total[0, 1] + 1e-15)
      h_db[f_idx] = 20.0 * np.log10(np.abs(h_omega) + 1e-12)

    return freqs, h_db
````

### Component 03: The Direct Helmholtz BEM 2D Boundary Collocation Solver

* **UID:** `APT-MOD-001`
* **Source Notebook:** `Implement_Boundary_Element_Method_(BEM)_solvers_f_.ipynb`
* **Archetype:** `MODAL_BOUNDARY_SOLVER`
* **Subsystem Focus:** 2D Green's function Helmholtz boundary integral equations over arbitrary non-circular geometries (lip protrusion, glottal slit, pyriform sinuses).
* **Mathematical Invariant:**

$$c(x) \psi(x) = \int_{\Gamma} \left[ G(x,y;k) \frac{\partial \psi}{\partial n_y}(y) - \frac{\partial G}{\partial n_y}(x,y;k) \psi(y) \right] d\Gamma(y)$$


$$G(x,y;k) = \frac{i}{4} H_0^{(1)}(k \Vert{}x - y\Vert{}), \quad H_{ii} = \frac{1}{2}, \quad G_{ii} = \frac{i L_i}{4}\left[1 + \frac{2i}{\pi}\left(\ln\left(\frac{k L_i}{4}\right) + \gamma_E - 1\right)\right]$$


* **Inputs:** `elements: List[BoundaryElement2D]` (1D boundary contour of the cross-section), `k_wavenumber: float`, `v_drive: np.ndarray` (Dipole velocity).
* **Outputs:** `psi_boundary: np.ndarray` (Surface acoustic pressure), `q_boundary: np.ndarray` (Normal velocity).
* **Dependencies:** `numpy`, `scipy.special`
* **Composable Tags:** `bem, helmholtz, hankel, greens_function, boundary_collocation, pyriform_sinus`
* **Silo Reference:** [NB-A00- Assembling all Python and Colab](https://notebooklm.google.com/notebook/f49e0447-f44c-46e7-a237-ee878534ee53)

````
from dataclasses import dataclass
import numpy as np
import scipy.special as sp


@dataclass
class BoundaryElement2D:
  x_start: float
  y_start: float
  x_end: float
  y_end: float

  @property
  def length(self) -> float:
    return np.hypot(self.x_end - self.x_start, self.y_end - self.y_start)

  @property
  def midpoint(self) -> np.ndarray:
    return np.array([
        (self.x_start + self.x_end) * 0.5,
        (self.y_start + self.y_end) * 0.5,
    ])

  @property
  def normal(self) -> np.ndarray:
    dx = self.x_end - self.x_start
    dy = self.y_end - self.y_start
    l = self.length + 1e-15
    return np.array([dy / l, -dx / l])


class HelmholtzBEMSolver2D:
  """Solves the 2D Helmholtz boundary integral equation via Gauss-Legendre quadrature

  and analytic logarithmic singularity self-integration.
  """

  def __init__(self, elements: list[BoundaryElement2D], k: float):
    self.elements = elements
    self.n = len(elements)
    self.k = float(k)
    self.euler_gamma = 0.5772156649015329
    self.g = np.zeros((self.n, self.n), dtype=np.complex128)
    self.h = np.zeros((self.n, self.n), dtype=np.complex128)
    self._assemble()

  def _assemble(self):
    gauss_pts = np.array(
        [-0.8611363116, -0.3399810436, 0.3399810436, 0.8611363116]
    )
    gauss_wts = np.array(
        [0.3478548451, 0.6521451549, 0.6521451549, 0.3478548451]
    )

    for i in range(self.n):
      xi = self.elements[i].midpoint
      for j in range(self.n):
        elem = self.elements[j]
        lj = elem.length
        nj = elem.normal

        if i == j:
          # Analytic Singular Self-Term Integration
          self.h[i, i] = 0.5 + 0.0j
          ln_term = (
              np.log((self.k * lj) / 4.0) + self.euler_gamma - 1.0
          )
          self.g[i, i] = (1j * lj / 4.0) * (
              1.0 + (2.0j / np.pi) * ln_term
          )
        else:
          # 4-Point Gauss-Legendre Quadrature
          g_val, h_val = 0.0 + 0.0j, 0.0 + 0.0j
          for p in range(4):
            s = gauss_pts[p]
            w = gauss_wts[p]
            y_pos = np.array([
                elem.x_start + (1.0 + s) * 0.5 * (elem.x_end - elem.x_start),
                elem.y_start + (1.0 + s) * 0.5 * (elem.y_end - elem.y_start),
            ])
            r_vec = y_pos - xi
            r = np.linalg.norm(r_vec) + 1e-15
            dr_dn = np.dot(r_vec, nj) / r

            g_kern = (1j / 4.0) * sp.hankel1(0, self.k * r)
            dg_kern = (-1j * self.k / 4.0) * sp.hankel1(1, self.k * r) * dr_dn
            jac = lj * 0.5
            g_val += g_kern * w * jac
            h_val += dg_kern * w * jac
          self.g[i, j] = g_val
          self.h[i, j] = h_val

  def solve(
      self, driving_v: np.ndarray, admittance: float = 0.05
  ) -> tuple[np.ndarray, np.ndarray]:
    """Robin boundary condition: q = -i*k*beta*psi + v_drive."""
    a_mat = self.h + (1j * self.k * admittance) * self.g
    rhs = self.g @ driving_v
    psi_b = np.linalg.solve(a_mat, rhs)
    q_b = -1j * self.k * admittance * psi_b + driving_v
    return psi_b, q_b
````

### Component 04: Dynamic CV Coarticulation & Acoustic Locus Formant Relaxer

* **UID:** `APT-ART-001`
* **Source Notebook:** `dynamic_consonant_vowel_(CV)_coarticulation.ipynb`
* **Archetype:** `SKILL_ARTICULATOR`
* **Subsystem Focus:** Consonant-vowel transitions, acoustic locus relaxation trajectories, and velar pinch coarticulation.
* **Mathematical Invariant:**

$$F_n(t) = F_n^{\text{vowel}} + (F_n^{\text{locus}} - F_n^{\text{vowel}}) e^{-\frac{t - t_{\text{rel}}}{\tau_n}}, \quad e_{\text{burst}}(t) = A_{\text{burst}} \left(\frac{t - t_{\text{rel}}}{t_{\text{rise}}}\right) e^{1 - \frac{t - t_{\text{rel}}}{t_{\text{rise}}}}$$


* **Inputs:** `consonant: str` (`'b'`, `'d'`, `'g'`), `vowel: str` (`'a'`, `'i'`, `'u'`), `f0: float`, `duration: float`.
* **Outputs:** `audio_pcm: np.ndarray` (Synthesized 16-bit speech samples).
* **Dependencies:** `numpy`, `scipy.signal`
* **Composable Tags:** `coarticulation, locus_equations, velar_pinch, plosive_burst, biquad_cascade, direct_form_ii`
* **Silo Reference:** [NB- B00 - GeoMersPhonetiQmetricsModality...](https://gemini.google.com/notebook/295cf8f8-f74c-4a19-b635-20a653ea660b)

````
import numpy as np
import scipy.signal as signal


class SovereignCoarticulationSynthesizer:
  """Synthesizes dynamic CV transitions using acoustic locus equations and

  sample-by-sample Direct Form II Transposed biquad filtering.
  """

  def __init__(self, sample_rate: int = 44100):
    self.sr = sample_rate
    self.vowels = {
        'a': {'F': [730, 1090, 2440, 3400], 'BW': [80, 90, 130, 200]},
        'i': {'F': [270, 2290, 3010, 3500], 'BW': [50, 100, 160, 220]},
        'u': {'F': [300, 870, 2240, 3400], 'BW': [60, 80, 110, 180]},
    }
    self.stops = {
        'b': {
            'locus': [180, 750, 2100, 3400],
            'burst_f': 800,
            'burst_bw': 400,
            'amp': 0.15,
        },
        'd': {
            'locus': [200, 1800, 2700, 3400],
            'burst_f': 4000,
            'burst_bw': 600,
            'amp': 0.35,
        },
        'g': {
            'locus': [220, 2400, 2300, 3400],
            'burst_f': 2200,
            'burst_bw': 300,
            'amp': 0.45,
        },
    }

  def _rosenberg_glottal_pulse(self, f0: float, n_samples: int) -> np.ndarray:
    period = int(self.sr / f0)
    t_open, t_close = int(0.40 * period), int(0.16 * period)
    pulse = np.zeros(n_samples)
    for p in range(0, n_samples, period):
      for i in range(min(period, n_samples - p)):
        if i < t_open:
          pulse[p + i] = 0.5 * (1.0 - np.cos(np.pi * i / t_open))
        elif i < t_open + t_close:
          pulse[p + i] = np.cos(np.pi * (i - t_open) / (2.0 * t_close))
    d_pulse = np.diff(pulse, prepend=0)
    return d_pulse / (np.max(np.abs(d_pulse)) + 1e-9)

  def synthesize_syllable(
      self, consonant: str = 'd', vowel: str = 'a', f0: float = 120.0
  ) -> np.ndarray:
    duration = 0.35
    n = int(self.sr * duration)
    t = np.linspace(0, duration, n, endpoint=False)
    t_rel = 0.045
    t_burst = 0.012

    c_spec = self.stops[consonant]
    v_spec = self.vowels[vowel]

    # Velar Pinch adaptation for /g/
    if consonant == 'g' and vowel != 'i':
      c_spec['locus'] = [220, 1400, 2300, 3400]
      c_spec['burst_f'] = 1600

    # 1. Voice Bar & Glottal Excitation
    glottal = self._rosenberg_glottal_pulse(f0, n)
    sos_voice_bar = signal.butter(2, 180, 'lowpass', fs=self.sr, output='sos')
    voice_bar = signal.sosfilt(sos_voice_bar, glottal) * 0.18

    excitation = np.zeros(n)
    for i in range(n):
      if t[i] < t_rel:
        excitation[i] = voice_bar[i]
      else:
        gain = np.clip((t[i] - t_rel) / 0.015, 0.0, 1.0)
        excitation[i] = glottal[i] * gain

    # 2. Transient Release Burst
    sos_burst = signal.butter(
        2,
        [
            c_spec['burst_f'] - c_spec['burst_bw'] / 2,
            c_spec['burst_f'] + c_spec['burst_bw'] / 2,
        ],
        'bandpass',
        fs=self.sr,
        output='sos',
    )
    burst_noise = signal.sosfilt(sos_burst, np.random.normal(0, 1, n))
    for i in range(n):
      if t_rel <= t[i] < t_rel + t_burst:
        dt = t[i] - t_rel
        env = (dt / (t_burst * 0.2)) * np.exp(1.0 - dt / (t_burst * 0.2))
        excitation[i] += burst_noise[i] * env * c_spec['amp']

    # 3. Dynamic Formant Trajectories
    f_trajs = np.zeros((4, n))
    bw_trajs = np.zeros((4, n))
    tau = 0.040 / 2.8
    for k in range(4):
      f_loc = c_spec['locus'][k]
      f_vow = v_spec['F'][k]
      for i in range(n):
        if t[i] < t_rel:
          f_trajs[k, i] = f_loc
        else:
          f_trajs[k, i] = f_vow + (f_loc - f_vow) * np.exp(-(t[i] - t_rel) / tau)
        bw_trajs[k, i] = v_spec['BW'][k]

    # 4. Direct Form II Transposed Cascade Filtering
    audio = np.copy(excitation)
    for k in range(4):
      w1, w2 = 0.0, 0.0
      for i in range(n):
        f = np.clip(f_trajs[k, i], 20.0, self.sr * 0.49)
        bw = np.clip(bw_trajs[k, i], 10.0, self.sr * 0.25)
        r = np.exp(-np.pi * bw / self.sr)
        theta = 2.0 * np.pi * f / self.sr
        a1 = -2.0 * r * np.cos(theta)
        a2 = r * r
        b0 = 1.0 - r

        x_in = audio[i]
        y_out = b0 * x_in + w1
        w1 = -a1 * y_out + w2
        w2 = -a2 * y_out
        audio[i] = y_out

    return audio / (np.max(np.abs(audio)) + 1e-9)
````

### Component 05: Unvoiced Aspirated Stops & Nasal Antizero Resonator

* **Silo Reference:** [NB- B00 - GeoMersPhonetiQmetricsModality...](https://gemini.google.com/notebook/295cf8f8-f74c-4a19-b635-20a653ea660b)

````
import numpy as np
import scipy.signal as signal


class AdvancedStopNasalSynthesizer:
  """Implements acoustic stub antiformants (zeroes) for nasals (/m, n/)

  and turbulent aspiration noise with Voice Onset Time (VOT) for /p, t, k/.
  """

  def __init__(self, sample_rate: int = 44100):
    self.sr = sample_rate
    self.specs = {
        'p': {
            'type': 'unvoiced',
            'locus': [180, 750, 2100, 3400],
            'burst_f': 800,
            'vot': 0.050,
            'amp': 0.25,
        },
        't': {
            'type': 'unvoiced',
            'locus': [200, 1800, 2700, 3400],
            'burst_f': 4200,
            'vot': 0.060,
            'amp': 0.40,
        },
        'k': {
            'type': 'unvoiced',
            'locus': [220, 2000, 2300, 3400],
            'burst_f': 2000,
            'vot': 0.075,
            'amp': 0.60,
        },
        'm': {
            'type': 'nasal',
            'locus': [250, 1000, 2200, 3400],
            'z1': 950.0,
            'fn1': 280.0,
            'closure': 0.08,
        },
        'n': {
            'type': 'nasal',
            'locus': [280, 1700, 2600, 3400],
            'z1': 1750.0,
            'fn1': 290.0,
            'closure': 0.075,
        },
    }

  def synthesize(
      self, phoneme: str = 'm', vowel: str = 'a', f0: float = 120.0
  ) -> np.ndarray:
    duration = 0.40
    n = int(self.sr * duration)
    t = np.linspace(0, duration, n, endpoint=False)
    spec = self.specs[phoneme]

    # Glottal Differentiated Rosenberg Pulse
    period = int(self.sr / f0)
    glottal = np.zeros(n)
    for p in range(0, n, period):
      for i in range(min(period, n - p)):
        if i < int(0.4 * period):
          glottal[p + i] = 0.5 * (1 - np.cos(np.pi * i / (0.4 * period)))
        elif i < int(0.56 * period):
          glottal[p + i] = np.cos(
              np.pi * (i - 0.4 * period) / (2 * 0.16 * period)
          )
    excitation = np.diff(glottal, prepend=0)

    if spec['type'] == 'unvoiced':
      t_rel = 0.05
      t_vot_end = t_rel + spec['vot']
      sos_tilt = signal.butter(1, 1500, 'lowpass', fs=self.sr, output='sos')
      asp_noise = (
          signal.sosfilt(sos_tilt, np.random.normal(0, 1, n)) * 0.25
      )

      # Silent occlusion and aspiration window
      for i in range(n):
        if t[i] < t_rel:
          excitation[i] = 0.0
        elif t_rel <= t[i] < t_vot_end:
          env = np.sin(np.pi * (t[i] - t_rel) / spec['vot'])
          excitation[i] = asp_noise[i] * env
        else:
          excitation[i] *= np.clip((t[i] - t_vot_end) / 0.02, 0.0, 1.0)

    elif spec['type'] == 'nasal':
      t_rel = spec['closure']
      for i in range(n):
        if t[i] < t_rel:
          excitation[i] *= 0.65

    # Cascade Formant Resonators
    audio = np.copy(excitation)
    for k in range(4):
      f_loc = spec['locus'][k]
      f_vow = [730, 1090, 2440, 3400][k]
      w1, w2 = 0.0, 0.0
      for i in range(n):
        f = (
            f_loc
            if t[i] < t_rel
            else f_vow + (f_loc - f_vow) * np.exp(-(t[i] - t_rel) / 0.018)
        )
        r = np.exp(-np.pi * 90.0 / self.sr)
        theta = 2.0 * np.pi * f / self.sr
        a1 = -2.0 * r * np.cos(theta)
        a2 = r * r
        b0 = 1.0 - r

        x_in = audio[i]
        y_out = b0 * x_in + w1
        w1 = -a1 * y_out + w2
        w2 = -a2 * y_out
        audio[i] = y_out

    # Apply 2nd-Order Oral Stub Antizero Notch Filter for Nasals
    if spec['type'] == 'nasal':
      fz = spec['z1']
      theta_z = 2.0 * np.pi * fz / self.sr
      r_z = 0.985
      b0_z, b1_z, b2_z = 1.0, -2.0 * r_z * np.cos(theta_z), r_z * r_z
      r_p = r_z - (110.0 / self.sr) * np.pi
      a1_p, a2_p = -2.0 * r_p * np.cos(theta_z), r_p * r_p

      notch_audio = np.zeros(n)
      for i in range(n):
        depth = 1.0 if t[i] < t_rel else max(0.0, 1.0 - (t[i] - t_rel) / 0.02)
        x0 = audio[i]
        x1 = audio[i - 1] if i > 0 else 0.0
        x2 = audio[i - 2] if i > 1 else 0.0
        y1 = notch_audio[i - 1] if i > 0 else 0.0
        y2 = notch_audio[i - 2] if i > 1 else 0.0
        y_filt = (b0_z * x0 + b1_z * x1 + b2_z * x2) - (a1_p * y1 + a2_p * y2)
        notch_audio[i] = (1.0 - depth) * x0 + depth * y_filt
      audio = notch_audio

    return audio / (np.max(np.abs(audio)) + 1e-9)
````

### Component 06: Glyphic Stroke Curvature Transducer

* **UID:** `APT-GLY-001`
* **Source Notebook:** `Omnilingual Morpho-Acoustic Transduction.ipynb`
* **Archetype:** `SKILL_TRANSDUCER`
* **Subsystem Focus:** Kinematic stroke curvature tensors mapped to formant resonance poles across uncatalogued and ancient scripts.
* **Mathematical Invariant:**

$$\mathcal{K}(s) = \frac{\vert{}\dot{x}\ddot{y} - \dot{y}\ddot{x}\vert{}}{(\dot{x}^2 + \dot{y}^2)^{3/2}} \implies F_1 = 200 + \frac{650}{1 + 0.15 \mathcal{K}}, \quad F_2 = 700 + 1700(1 - e^{-0.25 N_{\text{strokes}}})$$


* **Inputs:** `strokes: List[np.ndarray]` (List of 2D stroke coordinate trajectories $[(x_1, y_1), \dots]$).
* **Outputs:** `acoustic_formants: Dict[str, float]` ($F_1, F_2, F_3$ resonant frequencies).

* **Composable Tags:** `glyph, stroke_curvature, cuneiform, hieroglyphs, morpho_acoustics, ancient_scripts`
* **Silo Reference:** [NB 11 - Intelligence Poet - Poetry the The Semantics and the Function](https://gemini.google.com/notebook/321ac690-f064-4527-9f65-218a29aaa52e)

````
import numpy as np


class GlyphicAcousticTransducer:
  """Converts visual glyph strokes directly into vocal tract acoustic formants

  without linguistic text models or arbitrary token tables.
  """

  def __init__(self, c_sound: float = 343.0):
    self.c = c_sound

  def evaluate_stroke_curvature(self, stroke: np.ndarray) -> float:
    if len(stroke) < 3:
      return 0.0
    dx = np.gradient(stroke[:, 0])
    dy = np.gradient(stroke[:, 1])
    d2x = np.gradient(dx)
    d2y = np.gradient(dy)
    num = np.abs(dx * d2y - dy * d2x)
    den = (dx**2 + dy**2) ** 1.5 + 1e-12
    return float(np.mean(num / den))

  def transduce_glyph_to_sound(
      self, glyph_strokes: list[np.ndarray]
  ) -> dict[str, float]:
    total_k = sum(self.evaluate_stroke_curvature(st) for st in glyph_strokes)
    n_strokes = len(glyph_strokes)

    f1 = 200.0 + (650.0 / (1.0 + 0.15 * total_k))
    f2 = 700.0 + (1700.0 * (1.0 - np.exp(-0.25 * n_strokes)))
    f3 = 2100.0 + (1100.0 * np.sin(total_k * np.pi / 4.0) ** 2)

    return {
        "F1": round(f1, 2),
        "F2": round(f2, 2),
        "F3": round(f3, 2),
        "Curvature_Tensor": round(total_k, 4),
    }
````
