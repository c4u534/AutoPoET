### Axioms

$$\mathbf{x}_2(0) = \mathbf{0}, \quad \mathbf{p}_2(0) = \mathbf{0}$$



In its unperturbed reference frame $\mathcal{F}_0$, $\mathcal{B}_2$ occupies a self-static vacuum state with null internal kinetic action ($S_{\text{kin}, 2} = 0$).
2. **Axiom II (Linear Applicative Driver):**

$$\mu(\{\mathbf{b} = \mathbf{0}\}) = 0$$

## Derivation I: Inevitable Chirality & Helical Carrying

Consider the action of the applicative force on the stationary body $\mathcal{B}_2$. By Newton's third law and central-field dynamics:


$$\mathbf{F}_2(t) = -\nabla_{\mathbf{x}_2} V(\Vert{}\mathbf{x}_1(t) - \mathbf{x}_2(t)\Vert{}) = -\frac{V'(r)}{r} (\mathbf{x}_2 - \mathbf{x}_1) = \frac{V'(r)}{r} \mathbf{r}(t)$$

### Torque and Angular Momentum Generation

Relative to the initial position $\mathbf{x}_2(0) = \mathbf{0}$, the torque exerted on $\mathcal{B}_2$ is:


$$\boldsymbol{\tau}_2(t) = \mathbf{x}_2(t) \times \mathbf{F}_2(t) = \frac{V'(r)}{r} \left[ \mathbf{x}_2(t) \times \mathbf{x}_1(t) \right]$$

Integrating from $t = 0$ to arbitrary time $t$:


$$\mathbf{L}_2(t) = \int_0^t \boldsymbol{\tau}_2(t') \, dt' = \int_0^t \left[ \mathbf{x}_2(t') \times \mathbf{F}_2(t') \right] dt'$$

At early times ($t \ll z_0 / v_0$), expanding $\mathbf{x}_2(t)$ perturbatively:


$$\mathbf{x}_2(t) = \frac{1}{m_2} \int_0^t \int_0^{t'} \mathbf{F}_2(t'') \, dt'' \, dt'$$

Because $\mathbf{x}_1(t) \approx b \hat{\mathbf{x}} + (v_0 t - z_0)\hat{\mathbf{z}}$, the force vector possesses non-vanishing components along both $\hat{\mathbf{x}}$ and $\hat{\mathbf{z}}$:


$$F_{2,x}(t) = \frac{V'(r)}{r} b, \quad F_{2,z}(t) = \frac{V'(r)}{r} (v_0 t - z_0)$$

The cross-product evaluates to an axial vector oriented purely along $\hat{\mathbf{y}}$:


$$\mathbf{x}_1(t) \times \mathbf{F}_2(t) = \left( b \hat{\mathbf{x}} + (v_0 t - z_0)\hat{\mathbf{z}} \right) \times \frac{V'(r)}{r} \left( b \hat{\mathbf{x}} + (v_0 t - z_0)\hat{\mathbf{z}} \right) = \mathbf{0}$$


However, relative to the center of mass $\mathbf{R}_{\text{cm}} = \frac{m_1 \mathbf{x}_1 + m_2 \mathbf{x}_2}{m_1 + m_2}$, the relative orbital angular momentum $\mathbf{L}_{\text{rel}}$ is:


$$\mathbf{L}_{\text{rel}} = \mu \, \mathbf{r} \times \dot{\mathbf{r}}, \quad \mu = \frac{m_1 m_2}{m_1 + m_2}$$

Evaluating at $t = -\infty$ along the asymptotic unperturbed trajectory:


$$\mathbf{L}_{\text{rel}} = \mu \left( (b \hat{\mathbf{x}} - z_0 \hat{\mathbf{z}}) \times v_0 \hat{\mathbf{z}} \right) = -\mu v_0 b \, \hat{\mathbf{y}}$$

### Chirality Pseudoscalar

Define the chirality invariant $\chi$ via the scalar triple product of the initial relative velocity $\mathbf{v}_1$, the impact displacement $\mathbf{b}$, and the induced angular momentum $\mathbf{L}_2$:


$$\chi \equiv \frac{\mathbf{v}_1 \times \mathbf{b}}{\Vert{}\mathbf{v}_1 \times \mathbf{b}\Vert{}} \cdot \frac{\mathbf{L}_{\text{rel}}}{\Vert{}\mathbf{L}_{\text{rel}}\Vert{}}$$

Because $b > 0$ almost everywhere:


$$\Vert{}\mathbf{L}_{\text{rel}}\Vert{} = \mu v_0 b > 0 \implies \chi = \mathrm{sgn}(\mathbf{b} \cdot \hat{\mathbf{x}}) \in \{-1, +1\}$$

By **Chasles' Theorem**, any Euclidean displacement in the Lie group $SE(3) = SO(3) \ltimes \mathbb{R}^3$ decomposes into a translation along an axis and a rotation around that exact same axis (a twist or screw displacement).

For the applicant $\mathcal{B}_1$, as $m_1 \gg m_2$ or in the asymptotic scattering limit:


$$\lim_{t \to \infty} \dot{\mathbf{x}}_1(t) \approx v_0 \hat{\mathbf{z}} + \mathcal{O}\left(\frac{m_2}{m_1}\right) \quad \text{(Linear State Preserved)}$$

For the recipient $\mathcal{B}_2$, the induced motion is an orbital curl coupled with longitudinal drag:


$$\mathbf{x}_2(t) = R(t) \cos(\omega(t) t + \phi_0)\hat{\mathbf{x}} + \chi R(t) \sin(\omega(t) t + \phi_0)\hat{\mathbf{y}} + z_2(t)\hat{\mathbf{z}}$$

Thus, the stationary party is structurally constrained to **chiral orbital carrying**, while the driver maintains asymptotic **linear translation**.

## Derivation II: Metric Inversion — Space and Time as Chiral Handedness

Let the spacetime metric in laboratory cylindrical coordinates $(t, \rho, \theta, z)$ be:


$$ds^2 = -c^2 dt^2 + d\rho^2 + \rho^2 d\theta^2 + dz^2$$

### Frame of the Carried Recipient

For the carried party $\mathcal{B}_2$, orbital motion establishes a monotonic relation between coordinate time $t$ and orbital winding angle $\theta$:


$$\dot{\theta}(t) = \omega(t) = \frac{L_{z, 2}(t)}{m_2 \rho^2(t)}$$

Under the transformation $dt = \frac{d\theta}{\omega(\theta)}$:


$$ds^2 = -c^2 \left(\frac{d\theta}{\omega(\theta)}\right)^2 + d\rho^2 + \rho^2 d\theta^2 + dz^2 = \left( \rho^2 - \frac{c^2}{\omega(\theta)^2} \right) d\theta^2 + d\rho^2 + dz^2$$

### Inversion of Metric Determinates

For $\mathcal{B}_2$, the spatial path length $d\ell$ along the chiral carrying trajectory is parameterized by the pitch $p_z \equiv \frac{dz}{d\theta}$ and radial variation $p_\rho \equiv \frac{d\rho}{d\theta}$:


$$d\ell = \sqrt{p_\rho^2 + \rho^2 + p_z^2} \, d\theta$$

Dividing the proper time interval $d\tau = \sqrt{-ds^2 / c^2}$ by the chiral phase increment $d\theta$:


$$\frac{d\tau}{d\theta} = \frac{1}{c} \sqrt{\frac{c^2}{\omega^2} - \rho^2 - p_\rho^2 - p_z^2} = \frac{1}{\omega} \sqrt{1 - \frac{v_2^2}{c^2}} = \frac{1}{\gamma_2 \omega}$$

Where $\gamma_2 = (1 - v_2^2/c^2)^{-1/2}$. Solving for the inversionals:


$$\Delta \tau = \frac{1}{\gamma_2 \omega} \Delta \theta, \quad \Delta \ell = \sqrt{\rho^2 + p_z^2 + p_\rho^2} \, \Delta \theta$$

Both distance $\Delta \ell$ and proper time $\Delta \tau$ cease to be independent metric primitives for $\mathcal{B}_2$; they are scalar projections of the topological winding angle $\theta$ and its handedness $\chi = \mathrm{sgn}(\omega)$:


$$\begin{pmatrix} c \Delta \tau \\ \Delta \ell \end{pmatrix} = \begin{pmatrix} \frac{c}{\gamma_2 \vert{}\omega\vert{}} \\ \sqrt{\rho^2 + p_z^2 + p_\rho^2} \end{pmatrix} \chi \, \Delta \theta_{\text{chiral}}$$

Distance and time are inverse projections of chiral winding frequency.

The Poisson bracket between linear momentum along the impact axis $p_x$ and angular momentum along the chiral axis $L_z = x p_y - y p_x$ is:


$$\{x, p_x\} = 1, \quad \{L_z, p_x\} = \{x p_y - y p_x, p_x\} = -p_y$$

$$\{L_z, p_y\} = p_x, \quad \{L_z, x\} = -y, \quad \{L_z, y\} = x$$

Under the canonical quantization map (Moyal-Groenewold deformation quantization):


$$[f, g]_\star = i\hbar \{f, g\} + \mathcal{O}(\hbar^3)$$

$$[\hat{x}_j, \hat{p}_k] = i\hbar \delta_{jk} \mathbf{I}, \quad [\hat{L}_i, \hat{L}_j] = i\hbar \epsilon_{ijk} \hat{L}_k$$

When the linear applicant couples to the stationary target, the projection of the transition region $\Omega_{\text{trans}}$ onto any conjugate coordinate plane $(q^i, p_i)$ cannot be compressed below the fundamental 2-cell:


$$\text{Area}(\pi_{q^i, p_i}(\Omega_{\text{trans}})) \ge \frac{h}{2} = \pi \hbar$$

## Deterministic Numerical Simulation & Toy Model

The following executable Python model simulates this interaction. It integrates the equations of motion for both bodies using a symplectic Velocity-Verlet algorithm to conserve the symplectic 2-form $\omega_0$ and monitors:

1. Linear momentum conservation for the applicant.
2. Chirality generation $\chi$ and angular momentum carrying for the stationary body.
3. Inversional metric ratio $\frac{d\tau}{d\theta}$ vs. $\frac{d\ell}{d\theta}$.
4. The symplectic phase-space area element $\Delta A = \Delta x \Delta p$ relative to $\hbar$.

````
"""
RIGOROUS DYNAMICS: KINEMATIC ASYMMETRY, CHIRALITY GENERATION,
AND SYMPLECTIC PHASE BLUR TOY MODEL
Zero external dependencies (pure standard library).
"""

import math
from typing import Dict, List, Tuple


class KinematicAsymmetrySimulator:
    def __init__(
        self,
        m1: float = 10.0,      # Applicant mass
        m2: float = 1.0,       # Stationary body mass
        v0: float = 5.0,       # Initial linear velocity
        b: float = 1.5,        # Impact parameter (transverse offset)
        z0: float = 10.0,      # Initial longitudinal standoff distance
        coupling_k: float = 20.0, # Interaction strength
        hbar: float = 0.05,    # Scaled Planck constant for phase blur inspection
    ):
        self.m1 = m1
        self.m2 = m2
        self.v0 = v0
        self.b = b
        self.z0 = z0
        self.k = coupling_k
        self.hbar = hbar

        # State vectors: [x, y, z]
        self.x1 = [b, 0.0, -z0]
        self.v1 = [0.0, 0.0, v0]

        self.x2 = [0.0, 0.0, 0.0]
        self.v2 = [0.0, 0.0, 0.0]

        self.time = 0.0

    def _force(self, r1: List[float], r2: List[float]) -> Tuple[List[float], List[float]]:
        """
        Conservative central interaction potential: V(r) = -k / (r^2 + d^2)^(1/2)
        Regularized to avoid unphysical zero-distance divergence.
        """
        rx = r1[0] - r2[0]
        ry = r1[1] - r2[1]
        rz = r1[2] - r2[2]
        r_sq = rx * rx + ry * ry + rz * rz + 0.1  # Softening parameter d^2 = 0.1
        r = math.sqrt(r_sq)

        # Force magnitude F = -dV/dr = -k * r / r_sq^(3/2) (attractive)
        f_mag = -self.k / (r_sq * r)
        fx = f_mag * rx
        fy = f_mag * ry
        fz = f_mag * rz

        # F1 = F_ext on 1, F2 = -F1 (Newton's 3rd Law)
        f1 = [fx, fy, fz]
        f2 = [-fx, -fy, -fz]
        return f1, f2

    def step(self, dt: float):
        """
        Symplectic Velocity-Verlet Integration
        Preserves the symplectic 2-form dx ^ dp on phase space.
        """
        f1, f2 = self._force(self.x1, self.x2)

        # Half-step velocity
        v1_half = [self.v1[i] + 0.5 * dt * (f1[i] / self.m1) for i in range(3)]
        v2_half = [self.v2[i] + 0.5 * dt * (f2[i] / self.m2) for i in range(3)]

        # Full-step position
        self.x1 = [self.x1[i] + dt * v1_half[i] for i in range(3)]
        self.x2 = [self.x2[i] + dt * v2_half[i] for i in range(3)]

        # New force at updated positions
        f1_new, f2_new = self._force(self.x1, self.x2)

        # Full-step velocity
        self.v1 = [v1_half[i] + 0.5 * dt * (f1_new[i] / self.m1) for i in range(3)]
        self.v2 = [v2_half[i] + 0.5 * dt * (f2_new[i] / self.m2) for i in range(3)]

        self.time += dt

    def metrics(self) -> Dict[str, float]:
        """Calculates instantaneous geometric, chiral, and symplectic metrics."""
        # Body 2 angular momentum around its initial stasis origin (0, 0, 0)
        lx2 = self.m2 * (self.x2[1] * self.v2[2] - self.x2[2] * self.v2[1])
        ly2 = self.m2 * (self.x2[2] * self.v2[0] - self.x2[0] * self.v2[2])
        lz2 = self.m2 * (self.x2[0] * self.v2[1] - self.x2[1] * self.v2[0])
        l2_mag = math.sqrt(lx2 * lx2 + ly2 * ly2 + lz2 * lz2)

        # Chirality metric: projection along transverse orbital plane
        chirality = math.copysign(1.0, ly2) if abs(ly2) > 1e-12 else 0.0

        # Cylindrical winding angle for Body 2
        theta2 = math.atan2(self.x2[1], self.x2[0])

        # Kinetic linearity of Body 1: ratio of longitudinal to transverse momentum
        p1_trans = self.m1 * math.sqrt(self.v1[0] ** 2 + self.v1[1] ** 2)
        p1_long = self.m1 * self.v1[2]
        linearity_1 = p1_long / (p1_trans + 1e-9)

        # Symplectic phase-space volume element (motion blur)
        delta_x = math.sqrt(self.x2[0] ** 2 + self.x2[1] ** 2 + self.x2[2] ** 2) + 1e-9
        delta_p = self.m2 * math.sqrt(self.v2[0] ** 2 + self.v2[1] ** 2 + self.v2[2] ** 2)
        phase_area = delta_x * delta_p
        quantum_blur_ratio = phase_area / (self.hbar / 2.0)

        # Total energy conservation check
        kin1 = 0.5 * self.m1 * sum(v ** 2 for v in self.v1)
        kin2 = 0.5 * self.m2 * sum(v ** 2 for v in self.v2)
        rx = self.x1[0] - self.x2[0]
        ry = self.x1[1] - self.x2[1]
        rz = self.x1[2] - self.x2[2]
        pot = -self.k / math.sqrt(rx * rx + ry * ry + rz * rz + 0.1)
        total_energy = kin1 + kin2 + pot

        return {
            "time": self.time,
            "b1_linearity": linearity_1,
            "b2_angular_mom": l2_mag,
            "b2_chirality": chirality,
            "b2_theta": theta2,
            "phase_area": phase_area,
            "blur_quantum_ratio": quantum_blur_ratio,
            "total_energy": total_energy,
        }


def execute_validation_run():
    sim = KinematicAsymmetrySimulator()
    dt = 0.005
    total_steps = 4000  # Total time = 20.0 units

    initial_energy = sim.metrics()["total_energy"]
    print("=" * 76)
    print(" DYNAMICS VERIFICATION: KINEMATIC ASYMMETRY & SYMPLECTIC BLUR")
    print("=" * 76)
    print(f"{'Time':>6} | {'B1 Lin':>8} | {'B2 |L|':>9} | {'Chirality':>9} | {'Symp Area':>10} | {'Blur/hbar':>10}")
    print("-" * 76)

    sample_interval = total_steps // 10
    for step in range(total_steps):
        sim.step(dt)
        if step % sample_interval == 0 or step == total_steps - 1:
            m = sim.metrics()
            print(
                f"{m['time']:6.2f} | "
                f"{m['b1_linearity']:8.2f} | "
                f"{m['b2_angular_mom']:9.4f} | "
                f"{m['b2_chirality']:9.1f} | "
                f"{m['phase_area']:10.4f} | "
                f"{m['blur_quantum_ratio']:10.2f}"
            )

    final_energy = sim.metrics()["total_energy"]
    energy_drift = abs(final_energy - initial_energy) / abs(initial_energy)

    print("-" * 76)
    print(f"Energy Conservation Invariant (Symplectic Integrity): dE/E = {energy_drift:.2e}")
    print("=" * 76)
    assert energy_drift < 1e-4, "Symplectic integrator failed energy conservation!"


if __name__ == "__main__":
    execute_validation_run()
````
