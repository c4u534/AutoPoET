### 2. CORE ARCHITECTURAL PILLARS & MICRO-BENCHMARK SYNTHESIS

#### 2.1 Finite Field Arithmetic over $GF(2^{16})$
Arithmetic operations are performed over the finite Galois Field $GF(2^{16})$, defined modulo the irreducible primitive polynomial:

$$p(x) = x^{16} + x^5 + x^3 + x + 1 \quad (\text{Hexadecimal: } \mathtt{0x1002B})$$

* **Assembly-Optimized Multiplication**: Bypasses compiler-generated branch logic by utilizing inline assembly. Registers shift, mask, and perform XOR feedback loops dynamically within raw registers to prevent branch predictor poisoning.
* **Reversible State Bijection**: Employs an unkeyed involution swizzle $\pi(x)$ on 4-bit nibbles satisfying $\pi(\pi(x)) = x$ to guarantee $100.0\%$ mathematical state-reversibility.

#### 2.2 64-Byte Cache-Line Alignment & False Sharing Prevention
Standard heap allocators are vulnerable to page offset misalignments. If a 16-bit word overlaps a 64-byte boundary, the CPU executes a split memory transaction, doubling retrieval latency. Furthermore, when multiple threads concurrently write to variables occupying the same cache line, they trigger MESI cache-line invalidations (False Sharing), stalling the pipeline.

This engine resolves this by allocating memory with a strict `alignas(64)` constraint via `posix_memalign`, completely isolating distinct thread states and maximizing L1/L2 cache hit rates.

### 4. PRODUCTION-GRADE INTEGRATION BLUEPRINTS

#### Blueprint 4.1: InfiniBand RDMA Substrate Orchestration
This C code utilizes the OpenFabrics Enterprise Distribution (OFED) verbs library (`<infiniband/verbs.h>`) to register our 64-byte aligned VMM substrate as an RDMA-accessible Memory Region:

```c
#include <infiniband/verbs.h>
#include <stdlib.h>
#include <stdio.h>

#define SUBSTRATE_SIZE 393216 // 1536 nodes * 256 bytes

struct RdmaContext {
    struct ibv_context*   ib_ctx;
    struct ibv_pd*        pd;      // Protection Domain
    struct ibv_mr*        mr;      // Registered Memory Region
    void*                 buffer;  // Aligned VMM substrate pointer
};

struct RdmaContext* initialize_rdma_substrate() {
    struct RdmaContext* ctx = malloc(sizeof(struct RdmaContext));
    
    // 1. Allocate 64-byte aligned VMM substrate buffer
    if (posix_memalign(&ctx->buffer, 64, SUBSTRATE_SIZE) != 0) {
        return NULL;
    }
    
    // 2. Open first available InfiniBand device
    struct ibv_device** dev_list = ibv_get_device_list(NULL);
    ctx->ib_ctx = ibv_open_device(dev_list[0]);
    ibv_free_device_list(dev_list);
    
    // 3. Allocate Protection Domain
    ctx->pd = ibv_alloc_pd(ctx->ib_ctx);
    
    // 4. Register memory region with read/write permissions for RDMA operations
    ctx->mr = ibv_reg_mr(ctx->pd, ctx->buffer, SUBSTRATE_SIZE,
                         IBV_ACCESS_LOCAL_WRITE | IBV_ACCESS_REMOTE_WRITE | IBV_ACCESS_REMOTE_READ);
                         
    printf("[RDMA] Registered Memory Region: Local Key (lkey)=0x%x, Remote Key (rkey)=0x%x, VirtAddr=%p\n",
           ctx->mr->lkey, ctx->mr->rkey, ctx->buffer);
           
    return ctx;
}
```

### 🛡️ Part II: Production Rust PyO3 Boilerplate & InfiniBand RDMA Memory Registration
This safe implementation provides explicit Protection Domain (PD) and Memory Region (MR) abstractions. It wraps raw pointer address registration inside clean RAII boundaries, ensuring memory regions are properly deregistered from the InfiniBand hardware tables when Python garbage-collects the objects.

````
rust_src_content = """use pyo3::prelude::*;
use pyo3::types::PyByteArray;
use std::os::raw::{c_void, c_int};
use std::ptr;

// Simulated InfiniBand ibverbs bindings for environments without physical Infiniband HCAs
#[repr(C)]
pub struct ibv_pd {
    pub handle: u32,
}

#[repr(C)]
pub struct ibv_mr {
    pub addr: *mut c_void,
    pub length: usize,
    pub lkey: u32,
    pub rkey: u32,
}

/// Thread-safe, compile-time verified Protection Domain (PD) wrapper
#[pyclass]
pub struct ProtectionDomain {
    raw_pd: *mut ibv_pd,
}

#[pymethods]
impl ProtectionDomain {
    #[new]
    pub fn new() -> Self {
        // Simulate ibv_alloc_pd
        let simulated_pd = Box::into_raw(Box::new(ibv_pd { handle: 42 }));
        Self { raw_pd: simulated_pd }
    }
}

impl Drop for ProtectionDomain {
    fn drop(&mut self) {
        unsafe {
            if !self.raw_pd.is_null() {
                let _ = Box::from_raw(self.raw_pd);
            }
        }
    }
}

/// InfiniBand Registered Memory Region (MR) wrapper featuring automatic RAII de-allocation
#[pyclass]
pub struct MemoryRegion {
    #[pyo3(get)]
    pub lkey: u32,
    #[pyo3(get)]
    pub rkey: u32,
    #[pyo3(get)]
    pub virtual_address: u64,
    raw_mr: *mut ibv_mr,
    buffer_ptr: *mut c_void,
    size: usize,
}

impl Drop for MemoryRegion {
    fn drop(&mut self) {
        unsafe {
            // 1. Deregister MR from ibverbs subsystem
            if !self.raw_mr.is_null() {
                let _ = Box::from_raw(self.raw_mr);
            }
            // 2. Free aligned memory buffer to prevent leaks
            if !self.buffer_ptr.is_null() {
                libc::free(self.buffer_ptr);
            }
        }
    }
}

/// Sovereign Substrate wrapping both PyO3 wrappers and direct memory registration
#[pyclass]
pub struct SovereignRdmaSubstrate {
    pd: Py<ProtectionDomain>,
    region: Option<Py<MemoryRegion>>,
}

#[pymethods]
impl SovereignRdmaSubstrate {
    #[new]
    pub fn new(pd: Py<ProtectionDomain>) -> Self {
        Self { pd, region: None }
    }

    /// Registers a 64-byte aligned memory block as an RDMA active region
    pub fn register_region(&mut self, py: Python, size: usize) -> PyResult<Py<MemoryRegion>> {
        unsafe {
            let mut ptr: *mut c_void = ptr::null_mut();
            // Allocate strictly 64-byte cache-line aligned buffer (posix_memalign)
            let res = libc::posix_memalign(&mut ptr, 64, size);
            if res != 0 {
                return Err(pyo3::exceptions::PyOSError::new_err("Failed to allocate aligned heap memory"));
            }

            // Zero-initialize buffer to avoid garbage state
            ptr::write_bytes(ptr, 0, size);

            // Simulate ibv_reg_mr registration with verbs keys
            let simulated_lkey = 0xCAFEBABE;
            let simulated_rkey = 0xDEADBEEF;
            let simulated_mr = Box::into_raw(Box::new(ibv_mr {
                addr: ptr,
                length: size,
                lkey: simulated_lkey,
                rkey: simulated_rkey,
            }));

            let mr = Py::new(py, MemoryRegion {
                lkey: simulated_lkey,
                rkey: simulated_rkey,
                virtual_address: ptr as u64,
                raw_mr: simulated_mr,
                buffer_ptr: ptr,
                size,
            })?;

            self.region = Some(mr.clone());
            Ok(mr)
        }
    }

    /// Exposes a direct zero-copy bytearray view of the registered memory to Python
    pub fn get_buffer_view<'py>(&self, py: Python<'py>) -> PyResult<&'py PyByteArray> {
        match &self.region {
            Some(mr_py) => {
                let mr = mr_py.borrow(py);
                unsafe {
                    // Constructs a memory slice directly targeting physical RAM without copying data
                    let slice = std::slice::from_raw_parts_mut(mr.buffer_ptr as *mut u8, mr.size);
                    Ok(PyByteArray::from_slice(py, slice))
                }
            },
            None => Err(pyo3::exceptions::PyRuntimeError::new_err("No registered memory region found")),
        }
    }
}

/// Root PyO3 Module entry point definition
#[pymodule]
fn sovereign_rdma(_py: Python, m: &PyModule) -> PyResult<()> {
    m.add_class::<ProtectionDomain>()?;
    m.add_class::<MemoryRegion>()?;
    m.add_class::<SovereignRdmaSubstrate>()?;
    Ok(())
}
"""

(src_dir / "lib.rs").write_text(rust_src_content, encoding='utf-8')
print("✔ Rust boilerplate code files generated successfully.")
````

### 🧠 Cache-Line Hits & False Sharing Evaluation (`alignas(64)` Impact)

Modern CPU architectures fetch memory in **64-byte cache lines**. When multiple threads simultaneously modify independent variables that reside within the same 64-byte boundary, they trigger a hardware conflict known as **False Sharing**:
1. **Core A** modifies index 0.
2. **Core B** wants to modify index 1.
3. Because both indexes share the same 64-byte cache line, the CPU's cache coherency protocol (such as MESI) invalidates the cache line for Core B, forcing it to reload data from the much slower L3 cache or DRAM.

By forcing `alignas(64)`, we guarantee that each data structure starts on its own independent cache line, eliminating these false-sharing validation penalties.

````
import time
import ctypes
from concurrent.futures import ThreadPoolExecutor

# Class representing an unaligned struct layout where variables sit directly next to each other
class UnalignedCounter(ctypes.Structure):
    _fields_ = [
        ("value", ctypes.c_int64)
    ]

# Class representing a 64-byte cache-line aligned struct (padding to align to 64 bytes)
class AlignedCounter(ctypes.Structure):
    _fields_ = [
        ("value", ctypes.c_int64),
        ("padding", ctypes.c_byte * 56) # 8 bytes (value) + 56 bytes (padding) = 64 bytes exact
    ]

NUM_THREADS = 8
ITERATIONS = 1_000_000

def run_unaligned_benchmark():
    # All counters are packed sequentially inside the same cache line
    counters = (UnalignedCounter * NUM_THREADS)()

    def worker(tid):
        for _ in range(ITERATIONS):
            counters[tid].value += 1

    t_start = time.perf_counter()
    with ThreadPoolExecutor(max_workers=NUM_THREADS) as executor:
        executor.map(worker, range(NUM_THREADS))
    t_end = time.perf_counter()
    return (t_end - t_start) * 1000

def run_aligned_benchmark():
    # Each counter starts exactly on a 64-byte boundary
    counters = (AlignedCounter * NUM_THREADS)()

    def worker(tid):
        for _ in range(ITERATIONS):
            counters[tid].value += 1

    t_start = time.perf_counter()
    with ThreadPoolExecutor(max_workers=NUM_THREADS) as executor:
        executor.map(worker, range(NUM_THREADS))
    t_end = time.perf_counter()
    return (t_end - t_start) * 1000

print("=== Running Cache-Line Hit Rate & Alignment Analysis ===\n")
print(f"Threads: {NUM_THREADS} | Operations/Thread: {ITERATIONS:,}\n")

# Warm-up
_ = run_unaligned_benchmark()
_ = run_aligned_benchmark()

# Executing benchmarks
unaligned_time = run_unaligned_benchmark()
aligned_time = run_aligned_benchmark()
speedup = unaligned_time / aligned_time

print("--- Results ---")
print(f"❌ Packed (Unaligned - False Sharing Active): {unaligned_time:.2f} ms")
print(f"✔ Aligned (alignas(64) - Cache Line Isolated): {aligned_time:.2f} ms")
print(f"🚀 Performance Gain: {speedup:.2f}x speedup via Cache Boundary Isolation")
````

Because the Python GIL serializes execution, we must use a compiled C-ABI test with native POSIX threads (`pthread`) to see true concurrent cache-coherency effects.

We will compile a C script containing:
1. **Unaligned Counters**: Sequentially adjacent 64-bit integer values sharing a single 64-byte L1 cache line.
2. **Aligned Counters**: Counters explicitly aligned using `__attribute__((aligned(64)))` to isolate each variable to its own independent cache line.
3. **POSIX Threads**: Parallel OS threads directly mutating these structures simultaneously.

````
import os
import subprocess
from pathlib import Path

# Configure Workspace
workspace_dir = Path('/content/drive/MyDrive/Colab-Omni/RemediatedFilesWorkspace')
if not workspace_dir.exists():
    workspace_dir = Path('/tmp/autopoet_monolith_workspace')

c_benchmark_src = r"""
#define _GNU_SOURCE
#include <stdio.h>
#include <stdlib.h>
#include <stdint.h>
#include <pthread.h>
#include <sys/time.h>

#define NUM_THREADS 8
#define ITERATIONS 10000000ULL

typedef struct {
    uint64_t value;
} UnalignedCounter_t;

typedef struct __attribute__((aligned(64))) {
    uint64_t value;
} AlignedCounter_t;

UnalignedCounter_t unaligned_counters[NUM_THREADS];
AlignedCounter_t aligned_counters[NUM_THREADS];

void* run_unaligned(void* arg) {
    int tid = *(int*)arg;
    for (uint64_t i = 0; i < ITERATIONS; i++) {
        unaligned_counters[tid].value++;
    }
    return NULL;
}

void* run_aligned(void* arg) {
    int tid = *(int*)arg;
    for (uint64_t i = 0; i < ITERATIONS; i++) {
        aligned_counters[tid].value++;
    }
    return NULL;
}

double get_time_ms() {
    struct timeval tv;
    gettimeofday(&tv, NULL);
    return (tv.tv_sec * 1000.0) + (tv.tv_usec / 1000.0);
}

int main() {
    pthread_t threads[NUM_THREADS];
    int tids[NUM_THREADS];
    double start, end;

    printf("=== Native Hardware Cache-Line Alignment Benchmark ===\n");
    printf("Threads: %d | Iterations per thread: %llu\n\n", NUM_THREADS, ITERATIONS);

    // 1. Run Unaligned Benchmark (Highly vulnerable to False Sharing)
    start = get_time_ms();
    for (int i = 0; i < NUM_THREADS; i++) {
        tids[i] = i;
        pthread_create(&threads[i], NULL, run_unaligned, &tids[i]);
    }
    for (int i = 0; i < NUM_THREADS; i++) {
        pthread_join(threads[i], NULL);
    }
    end = get_time_ms();
    double unaligned_duration = end - start;
    printf("❌ Unaligned (False Sharing Active): %.2f ms\n", unaligned_duration);

    // 2. Run Aligned Benchmark (Isolated via alignas(64))
    start = get_time_ms();
    for (int i = 0; i < NUM_THREADS; i++) {
        tids[i] = i;
        pthread_create(&threads[i], NULL, run_aligned, &tids[i]);
    }
    for (int i = 0; i < NUM_THREADS; i++) {
        pthread_join(threads[i], NULL);
    }
    end = get_time_ms();
    double aligned_duration = end - start;
    printf("✔ Aligned (Cache Line Isolated):     %.2f ms\n", aligned_duration);

    printf("\n🚀 Hardware speedup via cache alignment: %.2fx\n", unaligned_duration / aligned_duration);
    return 0;
}
"""

c_file = workspace_dir / 'cache_benchmark.c'
bin_file = workspace_dir / 'cache_benchmark'

c_file.write_text(c_benchmark_src, encoding='utf-8')

# Compile using optimized O3 parameters to represent real-world bare-metal execution
compile_cmd = ['gcc', '-O3', '-march=native', '-pthread', str(c_file), '-o', str(bin_file)]
res = subprocess.run(compile_cmd, capture_output=True, text=True)

if res.returncode == 0:
    # Run the native hardware benchmark
    run_res = subprocess.run([str(bin_file)], capture_output=True, text=True)
    print(run_res.stdout)
else:
    print(f"❌ Failed to compile: {res.stderr}")
````

````
import ctypes
import hashlib
import mmap
import os
from pathlib import Path
from collections import deque

class SHA256AddressParityPool:
    def __init__(self, shm_path: Path, pool_size: int = 512):
        self.shm_path = shm_path
        self.pool_size = pool_size
        self.node_bytes = 256
        self.substrate_size = 1536 * self.node_bytes

        # Double-ended queue for recycling allocated address slots
        self.recycled_pool = deque()
        self.allocated_slots = {}

        # Initialize mmap handle
        fd = os.open(str(shm_path), os.O_RDWR)
        self.mmap_buf = mmap.mmap(fd, self.substrate_size, mmap.MAP_SHARED, mmap.PROT_READ | mmap.PROT_WRITE)
        os.close(fd)

    def pre_load_parity_addresses(self, start_virtual_addr: int = 0x7F0000000000):
        """Pre-loads extensive list of SHA256 address parity signatures into VMM Nodes."""
        print(f"--- Pre-loading {self.pool_size} Parity Address Signatures into VMM Substrate ---")
        for i in range(self.pool_size):
            node_offset = i * self.node_bytes
            virtual_ptr = start_virtual_addr + node_offset

            # Generate physical memory address signature using SHA256
            payload = f"MEM_ADDR:{hex(virtual_ptr)}".encode('utf-8')
            sha256_sig = hashlib.sha256(payload).hexdigest()

            # Write directly to the payload segment of the mapped memory region
            # Struct layout: 56 bytes metrics + 200 bytes payload buffer
            payload_offset = node_offset + 56
            self.mmap_buf[payload_offset:payload_offset+64] = sha256_sig.encode('utf-8')

            # Add index to pool queue for recycling
            self.recycled_pool.append(i)

        print(f"✔ Successfully mapped, hashed, and pre-loaded {self.pool_size} nodes.")

    def lease_address_slot(self) -> int:
        """Acquires a preloaded address slot from the queue."""
        if not self.recycled_pool:
            raise IndexError("No available address slots in the recycled pool")
        slot_idx = self.recycled_pool.popleft()
        self.allocated_slots[slot_idx] = True
        return slot_idx

    def recycle_address_slot(self, slot_idx: int):
        """Recycles the address slot back into the active lease queue."""
        if slot_idx in self.allocated_slots:
            del self.allocated_slots[slot_idx]
            self.recycled_pool.append(slot_idx)

    def read_slot_signature(self, slot_idx: int) -> str:
        """Reads the stored SHA256 parity signature directly from mmap."""
        node_offset = slot_idx * self.node_bytes
        payload_offset = node_offset + 56
        return self.mmap_buf[payload_offset:payload_offset+64].decode('utf-8')

# Instantiate pool using existing RDMA substrate mapping file
target_vmm_file = Path('/dev/shm/sovereign_vmm_substrate_rdma.bin')
if not target_vmm_file.exists():
    target_vmm_file = Path('/tmp/autopoet_monolith_workspace/sovereign_vmm_substrate_rdma.bin')

parity_pool = SHA256AddressParityPool(target_vmm_file, pool_size=128)
parity_pool.pre_load_parity_addresses()

# Demo of Lease & Recycle Mechanism
slot_1 = parity_pool.lease_address_slot()
slot_2 = parity_pool.lease_address_slot()
print(f"\nLeased Slot {slot_1} Parity Address: {parity_pool.read_slot_signature(slot_1)}")
print(f"Leased Slot {slot_2} Parity Address: {parity_pool.read_slot_signature(slot_2)}")

parity_pool.recycle_address_slot(slot_1)
print(f"✔ Slot {slot_1} recycled back to lockless queue.")
````

````
import time
````
````
import os
import sys
import traceback
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path
import mmap
import hashlib
from collections import deque

# Configure paths to reuse the pre-populated binary
target_vmm_file = Path('/dev/shm/sovereign_vmm_substrate_rdma.bin')
if not target_vmm_file.exists():
    target_vmm_file = Path('/tmp/autopoet_monolith_workspace/sovereign_vmm_substrate_rdma.bin')

# Defining a unique class name to completely bypass namespace collisions with earlier cells
class HalfHashedAddressParityPool:
    def __init__(self, shm_path: Path, pool_size: int = 512, partition_id: int = 0):
        self.shm_path = shm_path
        self.pool_size = pool_size
        self.node_bytes = 256
        self.substrate_size = 1536 * self.node_bytes

        # Half-hashed partitioning: Split pool dynamically based on process parity (0=even, 1=odd)
        half_size = pool_size // 2
        if partition_id == 0:
            self.recycled_pool = deque(range(0, half_size))
        else:
            self.recycled_pool = deque(range(half_size, pool_size))

        self.allocated_slots = {}

        fd = os.open(str(shm_path), os.O_RDWR)
        self.mmap_buf = mmap.mmap(fd, self.substrate_size, mmap.MAP_SHARED, mmap.PROT_READ | mmap.PROT_WRITE)
        os.close(fd)

    def lease_address_slot(self) -> int:
        if not self.recycled_pool:
            raise IndexError("No available slots in process partition")
        slot_idx = self.recycled_pool.popleft()
        self.allocated_slots[slot_idx] = True
        return slot_idx

    def recycle_address_slot(self, slot_idx: int):
        if slot_idx in self.allocated_slots:
            del self.allocated_slots[slot_idx]
            self.recycled_pool.append(slot_idx)

    def read_slot_signature(self, slot_idx: int) -> str:
        node_offset = slot_idx * self.node_bytes
        payload_offset = node_offset + 56
        return self.mmap_buf[payload_offset:payload_offset+64].decode('utf-8')

# Worker initialization using half-hashed memory partitioning to avoid sibling collisions
def init_process_worker(vmm_path):
    global local_pool
    try:
        # Hash PID to assign the worker to either the Even (0) or Odd (1) address partition pool
        partition_id = os.getpid() % 2
        local_pool = HalfHashedAddressParityPool(vmm_path, pool_size=128, partition_id=partition_id)
    except Exception as e:
        print(f"[WORKER INIT ERROR] PID {os.getpid()}: {e}")
        local_pool = None

def execute_process_runs(worker_id, num_ops=100):
    pid = os.getpid()
    latencies = []
    success_count = 0
    err_msg = None

    global local_pool
    for i in range(num_ops):
        t0 = time.perf_counter_ns()
        try:
            if local_pool is None:
                raise RuntimeError("local_pool is uninitialized")
            slot_idx = local_pool.lease_address_slot()
            _ = local_pool.read_slot_signature(slot_idx)
            local_pool.recycle_address_slot(slot_idx)
            success_count += 1
        except Exception as e:
            if err_msg is None:
                err_msg = f"{type(e).__name__}: {str(e)}"
        t1 = time.perf_counter_ns()
        latencies.append((t1 - t0) / 1e6) # ms

    return worker_id, pid, success_count, latencies, err_msg

if __name__ == '__main__':
    print("=== Cross-Process Parallel Performance Benchmark (Half-Hashed Allocation) ===")

    NUM_PROCESSES = 4
    OPS_PER_PROCESS = 100
    TOTAL_EXPECTED_OPS = NUM_PROCESSES * OPS_PER_PROCESS

    t_start = time.perf_counter_ns()

    process_latencies = []
    total_successes = 0
    distinct_pids = set()
    errors = {}

    with ProcessPoolExecutor(max_workers=NUM_PROCESSES, initializer=init_process_worker, initargs=(target_vmm_file,)) as executor:
        futures = [executor.submit(execute_process_runs, j, OPS_PER_PROCESS) for j in range(NUM_PROCESSES)]
        for future in as_completed(futures):
            worker_id, pid, successes, latencies, err = future.result()
            total_successes += successes
            distinct_pids.add(pid)
            process_latencies.extend(latencies)
            if err:
                errors[pid] = err

    t_end = time.perf_counter_ns()
    wall_duration_ms = (t_end - t_start) / 1e6

    # Analytical metrics
    avg_lat = sum(process_latencies) / len(process_latencies) if process_latencies else 0
    min_lat = min(process_latencies) if process_latencies else 0
    max_lat = max(process_latencies) if process_latencies else 0
    process_throughput = (total_successes / wall_duration_ms) * 1000 if wall_duration_ms > 0 else 0

    print(f"Processes Spawning:  {NUM_PROCESSES} (Worker PIDs: {list(distinct_pids)})")
    if errors:
        print(f"Worker Errors:       {errors}")
    print(f"Total Transactions:  {total_successes} / {TOTAL_EXPECTED_OPS} ({(total_successes/TOTAL_EXPECTED_OPS)*100:.1f}% success rate)")
    print(f"Wall Time Duration:  {wall_duration_ms:.3f} ms")
    print(f"Average Latency:     {avg_lat:.4f} ms")
    print(f"Min / Max Latency:   {min_lat:.4f} ms / {max_lat:.4f} ms")
    print(f"Overall Throughput:  {process_throughput:.2f} transactions/sec")
````

````
import re
import time
from typing import Dict, List, Any

class JitRustPrecomputerVM:
    """
    An optimized JIT-inspired register-based VM that parses Rust-like syntax,
    supporting decimal and hexadecimal literal values, and executes transformations.
    """
    def __init__(self):
        self.registers = {"R0": 0, "R1": 0, "R2": 0, "R3": 0, "R4": 0, "R5": 0, "R6": 0, "R7": 0}
        self.instruction_cache = {}

    def precompile(self, rust_source: str) -> List[tuple]:
        """AST Pre-compiler and Tokenizer with Hexadecimal/Integer parsing support"""
        cache_key = hash(rust_source)
        if cache_key in self.instruction_cache:
            return self.instruction_cache[cache_key]

        compiled_ops = []
        lines = [line.strip() for line in rust_source.split('\n') if line.strip() and not line.startswith('//')]

        for line in lines:
            # Match assignments let Rx: u16 = value; (decimal or hex, or register)
            m_let = re.match(r"let\s+(R[0-7]):\s*u16\s*=\s*(0x[0-9A-Fa-f]+|\d+|R[0-7]);", line)
            if m_let:
                reg, val = m_let.groups()
                compiled_ops.append(("MOV", reg, val))
                continue

            # Match addition/XOR: Rx ^= Ry or value; (decimal or hex, or register)
            m_xor = re.match(r"(R[0-7])\s*\^=\s*(0x[0-9A-Fa-f]+|\d+|R[0-7]);", line)
            if m_xor:
                reg, val = m_xor.groups()
                compiled_ops.append(("XOR", reg, val))
                continue

            # Match shifts: Rx <<= Ry or value; (decimal or hex, or register)
            m_shl = re.match(r"(R[0-7])\s*<<=\s*(0x[0-9A-Fa-f]+|\d+|R[0-7]);", line)
            if m_shl:
                reg, val = m_shl.groups()
                compiled_ops.append(("SHL", reg, val))
                continue

        self.instruction_cache[cache_key] = compiled_ops
        return compiled_ops

    def execute(self, compiled_ops: List[tuple]) -> Dict[str, int]:
        """JIT-inspired fast-execution loop on virtual registers with base conversion"""
        for op, reg, val in compiled_ops:
            if val in self.registers:
                source_val = self.registers[val]
            else:
                # Parse Hexadecimal or base-10 values dynamically
                source_val = int(val, 16) if val.lower().startswith("0x") else int(val)

            if op == "MOV":
                self.registers[reg] = source_val & 0xFFFF
            elif op == "XOR":
                self.registers[reg] = (self.registers[reg] ^ source_val) & 0xFFFF
            elif op == "SHL":
                self.registers[reg] = (self.registers[reg] << source_val) & 0xFFFF

        return self.registers

# Demonstrating the updated In-Memory Rust Preprocessing VM
rust_code = """
let R0: u16 = 0xABCD;
let R1: u16 = 0x1002;
R0 ^= R1;
R0 <<= 1;
"""

vm = JitRustPrecomputerVM()
ops = vm.precompile(rust_code)
t0 = time.perf_counter_ns()
registers_out = vm.execute(ops)
t1 = time.perf_counter_ns()

print("=== JIT Preprocessor VM Verification ===")
print(f"Compiled AST ops: {ops}")
print(f"Final Register State: {registers_out}")
print(f"JIT Execution Duration: {(t1 - t0)/1e3:.3f} microseconds")
````

````
import os
import sys
import time
import torch
import numpy as np
from multiprocessing import Pool, RawArray

# Global container pointers for workers to access memory directly without serialization
var_dict = {}

def init_process_worker(shape, dtype, shared_a_name, shared_b_name, shared_res_name):
    """Initializes the process worker by mapping the raw shared arrays into numpy views."""
    global var_dict
    var_dict['shape'] = shape
    var_dict['dtype'] = dtype
    # Resolve array buffers from shared memory
    var_dict['a'] = np.frombuffer(globals()[shared_a_name], dtype=dtype)
    var_dict['b'] = np.frombuffer(globals()[shared_b_name], dtype=dtype)
    var_dict['res'] = np.frombuffer(globals()[shared_res_name], dtype=dtype)

def process_slice_worker(worker_id, slice_size, data_size):
    """Worker execution task. Computes the Galois Field multiplication on its assigned process slice."""
    global var_dict
    start_idx = worker_id * slice_size
    end_idx = min(start_idx + slice_size, data_size)

    # Wrap shared memory region into PyTorch tensors directly without memory copies
    t_a = torch.from_numpy(var_dict['a'][start_idx:end_idx].astype(np.int32))
    t_b = torch.from_numpy(var_dict['b'][start_idx:end_idx].astype(np.int32))

    # Executing Vectorized Galois Field GF(2^16) Multiplication
    res = torch.zeros_like(t_a, dtype=torch.int32)
    p = t_a.clone()
    b_temp = t_b.clone()
    poly = 0x1002B

    for _ in range(16):
        mask = (b_temp & 1).to(torch.bool)
        res = torch.where(mask, res ^ p, res)
        b_temp = b_temp >> 1
        p = p << 1
        overflow_mask = (p & 0x10000).to(torch.bool)
        p = torch.where(overflow_mask, p ^ poly, p)

    # Write computation back to zero-copy shared output buffer
    var_dict['res'][start_idx:end_idx] = res.to(torch.uint16).numpy()

def orchestrate_process_hardware_aware_engine(data_size: int = 5_000_000):
    """
    Refactored hardware-aware acceleration engine utilizing a Process Pool.
    This bypasses the Python GIL completely during concurrent computation.
    """
    print(f"=== Initiating Process-Pool Hardware-Aware Orchestrator (Data Size: {data_size:,}) ===")

    cpu_cores = os.cpu_count() or 1
    cuda_available = torch.cuda.is_available()

    print(f"[System Check] Available CPU Cores: {cpu_cores}")
    print(f"[System Check] PyTorch CUDA Capability: {cuda_available}")

    if cuda_available:
        print("\n🚀 Mode: GPU-CPU Co-Processing Acceleration Activated!")
        device = torch.device('cuda')
        tensor_a = torch.randint(1, 65535, (data_size,), dtype=torch.int32).pin_memory()
        tensor_b = torch.randint(1, 65535, (data_size,), dtype=torch.int32).pin_memory()

        t_start = time.perf_counter()
        dev_a = tensor_a.to(device, non_blocking=True)
        dev_b = tensor_b.to(device, non_blocking=True)

        # Vectorized GPU Execution
        res = torch.zeros_like(dev_a, dtype=torch.int32)
        p = dev_a.clone()
        b_temp = dev_b.clone()
        poly = 0x1002B
        for _ in range(16):
            mask = (b_temp & 1).to(torch.bool)
            res = torch.where(mask, res ^ p, res)
            b_temp = b_temp >> 1
            p = p << 1
            overflow_mask = (p & 0x10000).to(torch.bool)
            p = torch.where(overflow_mask, p ^ poly, p)

        torch.cuda.synchronize()
        t_end = time.perf_counter()

        duration_ms = (t_end - t_start) * 1000
        print(f"✔ GPU computation successfully completed in {duration_ms:.3f} ms.")
        print(f"⚡ Effective Throughput: {data_size / (t_end - t_start) / 1e6:.2f} MOPs/sec")
    else:
        print("\n⚡ Mode: CPU-to-CPU Process Pool Concurrency Activated!")
        workers = max(2, cpu_cores)
        slice_size = int(np.ceil(data_size / workers))
        print(f"[Process Partitioning] Spawning {workers} independent worker processes processing slices of {slice_size:,} nodes.")

        # Declare Shared Raw Arrays to bypass inter-process serialization overheads
        shared_a = RawArray('H', data_size)
        shared_b = RawArray('H', data_size)
        shared_res = RawArray('H', data_size)

        # Populate inputs locally
        np_a = np.frombuffer(shared_a, dtype=np.uint16)
        np_b = np.frombuffer(shared_b, dtype=np.uint16)
        np_a[:] = np.random.randint(1, 65535, size=data_size, dtype=np.uint16)
        np_b[:] = np.random.randint(1, 65535, size=data_size, dtype=np.uint16)

        # Expose shared memory names globally in parent space for initialization references
        global shared_a_buf, shared_b_buf, shared_res_buf
        shared_a_buf = shared_a
        shared_b_buf = shared_b
        shared_res_buf = shared_res

        t_start = time.perf_counter()
        # Initialize the Process Pool with pointer access mapped to shared matrices
        with Pool(processes=workers, initializer=init_process_worker,
                  initargs=((data_size,), np.uint16, 'shared_a_buf', 'shared_b_buf', 'shared_res_buf')) as pool:
            pool.starmap(process_slice_worker, [(i, slice_size, data_size) for i in range(workers)])
        t_end = time.perf_counter()

        duration_ms = (t_end - t_start) * 1000
        print(f"✔ Process-based computation completed in {duration_ms:.3f} ms.")
        print(f"⚡ Effective Throughput: {data_size / (t_end - t_start) / 1e6:.2f} MOPs/sec")

orchestrate_process_hardware_aware_engine(data_size=5_000_000)
````

````
%%writefile cython_galois.pyx
# cython: boundscheck=False
# cython: wraparound=False
# cython: cdivision=True

import numpy as np
cimport numpy as cnp

# Dynamic Cython C-Implementation of the Galois Field GF(2^16) Multiplication
cdef unsigned short c_gf16_mul_element(unsigned short a, unsigned short b) noexcept nogil:
    cdef unsigned int res = 0
    cdef unsigned int p = a
    cdef unsigned int b_temp = b
    cdef unsigned int poly = 0x1002B
    cdef int i

    for i in range(16):
        if b_temp & 1:
            res ^= p
        b_temp >>= 1
        p <<= 1
        if p & 0x10000:
            p ^= poly

    return <unsigned short>res

# Vectorized array loop bypassing the GIL completely
def cython_gf16_mul_array(cnp.ndarray[unsigned short, ndim=1] arr_a, cnp.ndarray[unsigned short, ndim=1] arr_b):
    cdef int size = arr_a.shape[0]
    cdef cnp.ndarray[unsigned short, ndim=1] arr_res = np.zeros(size, dtype=np.uint16)
    cdef int i

    with nogil:
        for i in range(size):
            arr_res[i] = c_gf16_mul_element(arr_a[i], arr_b[i])

    return arr_res
````

```markdown
### 🧠 Memory Alignment & Buffer Boundary Profiling
Because the underlying Cython module compiles down to direct hardware-addressed C arrays, the memory alignment (e.g., 64-byte L1 cache-line boundaries) of input NumPy arrays plays a key role in processing speed.

Let's write a memory alignment diagnostic suite to profile memory boundaries of our datasets and test memory alignment impact on performance.
```

````
import ctypes
import numpy as np
import cython_galois_parameterized

def get_alignment_offset(array: np.ndarray) -> int:
    """Calculates byte alignment offset relative to 64-byte boundaries."""
    address = array.__array_interface__['data'][0]
    return address % 64

def create_aligned_array(size: int, dtype, alignment: int = 64) -> np.ndarray:
    """Allocates a memory block explicitly aligned to hardware cache boundaries."""
    raw_bytes = size * np.dtype(dtype).itemsize
    buffer = bytearray(raw_bytes + alignment)
    address = ctypes.addressof(ctypes.c_char.from_buffer(buffer))
    offset = (alignment - (address % alignment)) % alignment

    # Slice out the aligned segment and view as target numpy type
    return np.frombuffer(buffer, dtype=dtype, count=size, offset=offset)

# Diagnostic profiling execution
print("=== Memory Alignment Profile ===\n")

# Standard allocation (typical memory alignments are random based on OS allocator)
arr_standard = np.random.randint(1, 65535, size=10_000_000, dtype=np.uint16)
print(f"[Standard Array] Memory Address: {hex(arr_standard.__array_interface__['data'][0])}")
print(f"[Standard Array] 64-Byte Alignment Offset: {get_alignment_offset(arr_standard)} bytes")

# Aligned allocation
arr_aligned_a = create_aligned_array(10_000_000, np.uint16, 64)
print(f"\n[Aligned Array]  Memory Address: {hex(arr_aligned_a.__array_interface__['data'][0])}")
print(f"[Aligned Array]  64-Byte Alignment Offset: {get_alignment_offset(arr_aligned_a)} bytes")
````
