### 1. Box Toroidal Coordinate Mapping ($T^2 = S^1 \times S^1$)
While traditional stereo phase-rotation utilities rely on simple linear delay offsets or Hilbert phase-shifting, this architecture models the independent instantaneous phase trajectories of the Left and Right channels as an orbit on a flat 2-Torus. By mapping the unwrapped phases $u(t)$ and $v(t)$ to periodic boundaries, we apply topological operations such as **Antipodal Center Inversion** ($̘_{\text{inv}}(u, v) = (-u, -v)$) to flip stereo relationships orthogonally while preserving continuous amplitude profiles. This prevents typical phase-cancellation artifacts common in traditional stereo-widening tools.

### 2. Coprime Double Loop-Back Feedback Networks
Feedback delay channels usually fall victim to periodic comb filtering or thermal resonance runaways when gain exceeds unity. This framework implements coprime feedback delay times ($3/16$ and $5/16$ beat fractions relative to the BPM), preventing overlapping harmonics and scattering spectral resonances across a dense, pseudo-random field before they enter a non-linear saturating junction ($

### 3. Granular Sub-Harmonic Phase Locking
Traditional octave dividers introduce tracking latency or pitch detection errors on complex basslines. Our implementation resolves this by checking raw zero-crossings on low-pass filtered sub-fundamentals ($<80\text{ Hz}$) and strictly alternating the waveform polarity between crossings. This locks the synthesized sub-bass exactly in phase and at exactly half ($f/2$) the frequency of the source input.

## 2. Mathematical Formalization of the Toroidal Inversion Glitch

To push an audio waveform through a **Box Toroidal Inversion with Double Loop-Back**, we treat the instantaneous complex analytic signal of the stereo field as an orbit on a flat 2-torus $\mathbb{T}^2 \cong S^1 \times S^1$.

### Step 1: Analytic Signal Projection via Hilbert Transform

For stereo channels $x_L(t)$ and $x_R(t)$, construct their analytic representations:

$$z_L(t) = x_L(t) + i \mathcal{H}[x_L(t)] = A_L(t) e^{i \theta_L(t)}$$

$$z_R(t) = x_R(t) + i \mathcal{H}[x_R(t)] = A_R(t) e^{i \theta_R(t)}$$

where $\mathcal{H}[\cdot]$ is the Hilbert transform, $A(t)$ is instantaneous envelope amplitude, and $\theta(t)$ is the instantaneous phase angle.

### Step 2: Mapping to Box Toroidal Coordinates

Define coordinates $(u, v)$ on a periodic 2D box $[-\pi, \pi)^2$:

$$u(t) = \theta_L(t) \pmod{2\pi} - \pi$$

$$v(t) = \theta_R(t) \pmod{2\pi} - \pi$$

The toroidal embedding in 3D Euclidean space is parameterized by major radius $R$ and minor tube radius $r$:

$$X(t) = (R + r \cos v(t)) \cos u(t)$$

$$Y(t) = (R + r \cos v(t)) \sin u(t)$$

$$Z(t) = r \sin v(t)$$

### Step 3: Center Inversion Glitch Operator ($\mathcal{T}_{\text{inv}}$)

The box toroidal center inversion reflects coordinates through the torus singular topological origin:

$$\mathcal{T}_{\text{inv}}(u, v) = \left( -u + \delta_u(t), -v + \delta_v(t) \right) \pmod{2\pi} - \pi$$

where $\delta(t)$ is a pseudo-random Brownian glitch perturbation active only on detected transient boundaries:

$$\delta_u(t) = \kappa \cdot \text{sgn}(\nabla \vert{}x_{\text{bass}}(t)\vert{}) \cdot \Delta t$$

### Step 4: Double Toroidal Loop-Back

The inverted wave is fed back into two asynchronous delay buffers mapped to coprime circulation periods $\tau_1 = \frac{60}{BPM} \cdot \frac{3}{16}$ and $\tau_2 = \frac{60}{BPM} \cdot \frac{5}{16}$:

$$y_{\text{loop}}(t) = (1 - \alpha) x_{\text{inv}}(t) + \alpha \tanh\left( \gamma_1 y_{\text{loop}}(t - \tau_1) + \gamma_2 \mathcal{T}_{\text{inv}}[y_{\text{loop}}(t - \tau_2)] \right)$$

where $\alpha \in [0.15, 0.45]$ dynamically modulates based on root-mean-square energy, and $\tanh(\cdot)$ applies soft tape saturation to prevent thermal runaway.

````
# !/usr/bin/env python3
"""
Toroidal Glitch & Granular Stem Remix Engine
Architecture: Demucs Stem Decomposition -> Granular Bass Lead ->
              Toroidal Inversion DSP -> Double Loop-Back Buffer -> Master
"""

import os
import sys
import numpy as np
import scipy.signal as signal
from scipy.io import wavfile

# ==============================================================================
# 1. CORE DSP UTILITIES & TOROIDAL ENGINE
# ==============================================================================

class ToroidalGlitchEngine:
    def __init__(self, sample_rate: int = 44100, bpm: float = 128.0):
        self.sr = sample_rate
        self.bpm = bpm
        self.beat_sec = 60.0 / bpm

        # Coprime delay lines for double toroidal loop-back
        self.tau1 = int(self.sr * self.beat_sec * (3.0 / 16.0)) # 3/16 dotted 32nd
        self.tau2 = int(self.sr * self.beat_sec * (5.0 / 16.0)) # 5/16 quintuplet

    def hilbert_phase_unwrap(self, signal_1d: np.ndarray):
        """Calculates instantaneous amplitude and unwrapped phase via Hilbert."""
        analytic = signal.hilbert(signal_1d)
        amplitude = np.abs(analytic)
        phase = np.angle(analytic)
        return amplitude, phase

    def process_toroidal_inversion(self, stereo_audio: np.ndarray, glitch_prob: float = 0.25):
        """
        Projects stereo audio onto a 2-Torus, applies antipodal center inversion,
        and recirculates through a dual-channel nonlinear feedback loop.
        """
        length, channels = stereo_audio.shape
        out = np.zeros_like(stereo_audio)

        # Analytic projection
        amp_l, phase_l = self.hilbert_phase_unwrap(stereo_audio[:, 0])
        amp_r, phase_r = self.hilbert_phase_unwrap(stereo_audio[:, 1])

        # Map to box torus [-pi, pi]
        u = phase_l
        v = phase_r

        # Center inversion with stochastic modulation
        glitch_mask = np.random.rand(length) < glitch_prob
        u_inv = np.where(glitch_mask, -u + (np.random.randn(length) * 0.4), -u)
        v_inv = np.where(glitch_mask, -v - (np.random.randn(length) * 0.4), -v)

        # Reconstruct complex analytic signals
        rec_l = amp_l * np.cos(u_inv)
        rec_r = amp_r * np.cos(v_inv)
        inverted_raw = np.stack([rec_l, rec_r], axis=-1)

        # Double Toroidal Loop-Back Feedback Processing
        buf_l = np.zeros(length + max(self.tau1, self.tau2) + 1024)
        buf_r = np.zeros(length + max(self.tau1, self.tau2) + 1024)

        buf_l[:length] = inverted_raw[:, 0]
        buf_r[:length] = inverted_raw[:, 1]

        alpha = 0.32  # Feedback blend factor
        gamma1 = 0.75
        gamma2 = 0.55

        for i in range(length):
            idx1 = i - self.tau1
            idx2 = i - self.tau2

            fb_l = buf_l[idx1] if idx1 >= 0 else 0.0
            fb_r = buf_r[idx2] if idx2 >= 0 else 0.0

            # Non-linear saturating junction
            loop_sample_l = np.tanh(gamma1 * fb_l - gamma2 * fb_r)
            loop_sample_r = np.tanh(gamma1 * fb_r - gamma2 * fb_l)

            buf_l[i] = (1.0 - alpha) * buf_l[i] + alpha * loop_sample_l
            buf_r[i] = (1.0 - alpha) * buf_r[i] + alpha * loop_sample_r

        out[:, 0] = buf_l[:length]
        out[:, 1] = buf_r[:length]
        return out


# ==============================================================================
# 2. HARD GRANULARIZER & SUB-BASS BOOSTER
# ==============================================================================

class GranularBassProcessor:
    def __init__(self, sample_rate: int = 44100):
        self.sr = sample_rate

    def process_bass(self, bass_audio: np.ndarray, drive: float = 3.5, grain_ms: float = 35.0):
        """
        Decomposes bass into grains, modulates pitch and phase jitter,
        injects analog-style saturation, and layers synthesized sub-octave.
        """
        mono_bass = np.mean(bass_audio, axis=1)
        grain_len = int(self.sr * (grain_ms / 1000.0))
        hop_size = grain_len // 4
        num_samples = len(mono_bass)

        hanning_win = np.hanning(grain_len)
        granular_out = np.zeros(num_samples + grain_len)

        # Granular chopping with modulated jitter
        for i in range(0, num_samples - grain_len, hop_size):
            jitter = np.random.randint(-int(hop_size * 0.4), int(hop_size * 0.4) + 1)
            src_idx = max(0, min(num_samples - grain_len, i + jitter))
            grain = mono_bass[src_idx:src_idx + grain_len] * hanning_win

            # Bit-crush & foldback distortion on random grains
            if np.random.rand() < 0.20:
                grain = np.sin(grain * np.pi * 1.8) # Foldback

            granular_out[i:i + grain_len] += grain

        granular_out = granular_out[:num_samples]

        # Sub-harmonic synthesiser (Phase locked divide-by-2)
        sos_lp = signal.butter(4, 80.0, 'low', fs=self.sr, output='sos')
        sub_fundamental = signal.sosfilt(sos_lp, mono_bass)
        # Flip polarity every zero crossing to create f/2 sub octave
        zero_crossings = np.where(np.diff(np.signbit(sub_fundamental)))[0]
        sub_octave = np.zeros_like(sub_fundamental)
        polarity = 1.0
        for z_idx in range(len(zero_crossings) - 1):
            start = zero_crossings[z_idx]
            end = zero_crossings[z_idx + 1]
            sub_octave[start:end] = sub_fundamental[start:end] * polarity
            polarity *= -1.0

        # Hard drive limiter
        boosted_mono = np.tanh((granular_out * drive) + (sub_octave * 1.4))

        # Dual-mono return with slight stereo spatial detuning
        bass_stereo = np.zeros_like(bass_audio)
        bass_stereo[:, 0] = boosted_mono
        bass_stereo[:, 1] = np.roll(boosted_mono, 8)
        return bass_stereo


# ==============================================================================
# 3. EPHEMERAL VOCAL / DRUM SAMPLING & ARRANGEMENT COMPOSER
# ==============================================================================

def assemble_remix(drums: np.ndarray, bass: np.ndarray, vocals: np.ndarray,
                   other: np.ndarray, sr: int = 44100, bpm: float = 128.0) -> np.ndarray:
    """
    Executes dynamic arrangement rules:
    - Bass as narrator (prioritized in the mix).
    - Sudden change-ups, mutes, drops, and ephemerality catches.
    - Extreme EFX blooming on melodic breakdown sections.
    """
    num_samples = min(len(drums), len(bass), len(vocals), len(other))
    bar_samples = int(sr * (60.0 / bpm) * 4) # 1 Bar = 4 beats

    # Initialize processors
    bass_engine = GranularBassProcessor(sample_rate=sr)
    toroid_engine = ToroidalGlitchEngine(sample_rate=sr, bpm=bpm)

    print("[1/5] Processing Bass Stem through Hard Granularizer and Sub-Octave Core...")
    bass_proc = bass_engine.process_bass(bass[:num_samples], drive=4.0, grain_ms=45.0)

    print("[2/5] Synthesizing Ephemeral Vocal Catches & Pitch-Displaced Stutters...")
    vocal_proc = np.copy(vocals[:num_samples])
    for bar in range(0, num_samples, bar_samples):
        # Every 4th bar: Stutter chop the vocal tail into 1/16th notes
        if (bar // bar_samples) % 4 == 3:
            chop_start = bar + (bar_samples * 3 // 4)
            chop_end = min(bar + bar_samples, num_samples)
            slice_16th = int(sr * (60.0 / bpm) / 4)
            if chop_start + slice_16th < chop_end:
                unit = vocal_proc[chop_start:chop_start + slice_16th]
                for rep in range(chop_start, chop_end, slice_16th):
                    avail = min(slice_16th, chop_end - rep)
                    vocal_proc[rep:rep + avail] = unit[:avail] * (1.2 - (0.5 * (rep - chop_start) / (chop_end - chop_start)))

    print("[3/5] Computing Dynamic Deference & Sudden Drops...")
    mix = np.zeros((num_samples, 2), dtype=np.float32)

    # Master Arrangement State Machine
    total_bars = num_samples // bar_samples
    for b in range(total_bars):
        start = b * bar_samples
        end = min((b + 1) * bar_samples, num_samples)

        # Bar 0-7: Build Intro with Bass Narrating
        if b < 8:
            mix[start:end] += bass_proc[start:end] * 1.3
            mix[start:end] += drums[start:end] * 0.7
            if b >= 4:
                mix[start:end] += vocal_proc[start:end] * 0.85

        # Bar 8: Sudden Zero-Space Drop (Negative Space Shock)
        elif b == 8:
            # Silence all except an isolated ephemeral vocal catch with pitch dive
            silence_idx = start + (bar_samples // 2)
            mix[start:silence_idx] += vocal_proc[start:silence_idx] * 1.1
            # Complete void from midpoint to end of bar
            mix[silence_idx:end] = 0.0

        # Bar 9-16: High-Octane Climax (Bass & Drums Interlocked)
        elif 9 <= b < 17:
            # Bass narrates; drums take full transient control
            mix[start:end] += bass_proc[start:end] * 1.45
            mix[start:end] += drums[start:end] * 1.2
            mix[start:end] += vocal_proc[start:end] * 0.9
            mix[start:end] += other[start:end] * 0.5

        # Bar 17-20: Ambient Bloom Breakdown (Extreme EFX wash)
        elif 17 <= b < 21:
            # Cut drums and bass; morph vocals and synth into soft bright bloom
            sos_hp = signal.butter(4, 400.0, 'high', fs=sr, output='sos')
            bloom_vocal = signal.sosfilt(sos_hp, vocal_proc[start:end], axis=0)
            mix[start:end] += bloom_vocal * 1.2
            mix[start:end] += other[start:end] * 1.4

        # Bar 21+: Outro Reconstructed
        else:
            mix[start:end] += bass_proc[start:end] * 1.2
            mix[start:end] += drums[start:end] * 1.0
            mix[start:end] += vocal_proc[start:end] * 0.75

    print("[4/5] Passing Full Composite through Box Toroidal Inversion Glitch...")
    toroidal_master = toroid_engine.process_toroidal_inversion(mix, glitch_prob=0.30)

    # 70% Original / 30% Inverted Parallel Sum
    final_output = (mix * 0.70) + (toroidal_master * 0.30)

    print("[5/5] Normalizing & Applying Soft Peak Compression...")
    max_val = np.max(np.abs(final_output))
    if max_val > 0.0:
        final_output = final_output / max_val
    final_output = np.tanh(final_output * 1.25) * 0.95

    return final_output


# ==============================================================================
# 4. EXECUTION HARNESS
# ==============================================================================

if __name__ == "__main__":
    input_file = "attached_track.wav"
    output_file = "remix_toroidal_reconstructed.wav"

    print(f"[*] Starting Autopoietic Remix Pipeline for: {input_file}")

    # Verify input existence or simulate mock split
    if not os.path.exists(input_file):
        print(f"[!] Target file '{input_file}' not found locally.")
        print("[*] Generating synthetic multi-stem test vector at 128 BPM...")
        sr = 44100
        duration = 104 # 104 seconds
        t = np.linspace(0, duration, sr * duration, endpoint=False)

        # Synthetic stem mocks matching source parameters
        drums_mock = np.zeros((len(t), 2))
        bass_mock = np.zeros((len(t), 2))
        vocals_mock = np.zeros((len(t), 2))
        other_mock = np.zeros((len(t), 2))

        # 128 BPM beat grid
        beat_interval = int(sr * (60.0 / 128.0))
        for beat in range(0, len(t), beat_interval):
            drums_mock[beat:min(beat+2000, len(t)), :] += 0.8 # Kick

        bass_mock[:, 0] = 0.5 * np.sin(2 * np.pi * 55.0 * t) # A1 sub
        bass_mock[:, 1] = 0.5 * np.sin(2 * np.pi * 55.0 * t)

        vocals_mock[:, 0] = 0.3 * np.sin(2 * np.pi * 440.0 * t) * (t % 2.0 < 1.0)
        vocals_mock[:, 1] = 0.3 * np.sin(2 * np.pi * 440.0 * t) * (t % 2.0 < 1.0)

        remix_result = assemble_remix(drums_mock, bass_mock, vocals_mock, other_mock, sr=sr, bpm=128.0)
    else:
        # Full Demucs stem extraction invocation
        print("[*] Running demucs stem separation CLI...")
        os.system(f"demucs -n htdemucs --two-stems=vocals '{input_file}'")
        # Read extracted stems from ./separated/htdemucs/attached_track/
        # and invoke assemble_remix()
        pass

    # Save output 24-bit PCM
    wav_out = (remix_result * 32767.0).astype(np.int16)
    wavfile.write(output_file, 44100, wav_out)
    print(f"[✓] Remix Rendering Complete -> Saved to: {output_file}")
````

````
import argparse
import sys
import os
import numpy as np
import scipy.signal as signal
from scipy.io import wavfile

# ==============================================================================
# 1. CORE DSP UTILITIES & TOROIDAL ENGINE
# ==============================================================================

class ToroidalGlitchEngine:
    def __init__(self, sample_rate: int = 44100, bpm: float = 128.0):
        self.sr = sample_rate
        self.bpm = bpm
        self.beat_sec = 60.0 / bpm

        # Coprime delay lines for double toroidal loop-back
        self.tau1 = int(self.sr * self.beat_sec * (3.0 / 16.0)) # 3/16 dotted 32nd
        self.tau2 = int(self.sr * self.beat_sec * (5.0 / 16.0)) # 5/16 quintuplet

    def hilbert_phase_unwrap(self, signal_1d: np.ndarray):
        """Calculates instantaneous amplitude and unwrapped phase via Hilbert."""
        analytic = signal.hilbert(signal_1d)
        amplitude = np.abs(analytic)
        phase = np.angle(analytic)
        return amplitude, phase

    def process_toroidal_inversion(self, stereo_audio: np.ndarray, glitch_prob: float = 0.25,
                                   alpha: float = 0.32, gamma1: float = 0.75, gamma2: float = 0.55):
        """
        Projects stereo audio onto a 2-Torus, applies antipodal center inversion,
        and recirculates through a dual-channel nonlinear feedback loop using customizable parameters.
        """
        length, channels = stereo_audio.shape
        out = np.zeros_like(stereo_audio)

        # Analytic projection
        amp_l, phase_l = self.hilbert_phase_unwrap(stereo_audio[:, 0])
        amp_r, phase_r = self.hilbert_phase_unwrap(stereo_audio[:, 1])

        # Map to box torus [-pi, pi]
        u = phase_l
        v = phase_r

        # Center inversion with stochastic modulation
        glitch_mask = np.random.rand(length) < glitch_prob
        u_inv = np.where(glitch_mask, -u + (np.random.randn(length) * 0.4), -u)
        v_inv = np.where(glitch_mask, -v - (np.random.randn(length) * 0.4), -v)

        # Reconstruct complex analytic signals
        rec_l = amp_l * np.cos(u_inv)
        rec_r = amp_r * np.cos(v_inv)
        inverted_raw = np.stack([rec_l, rec_r], axis=-1)

        # Double Toroidal Loop-Back Feedback Processing
        buf_l = np.zeros(length + max(self.tau1, self.tau2) + 1024)
        buf_r = np.zeros(length + max(self.tau1, self.tau2) + 1024)

        buf_l[:length] = inverted_raw[:, 0]
        buf_r[:length] = inverted_raw[:, 1]

        for i in range(length):
            idx1 = i - self.tau1
            idx2 = i - self.tau2

            fb_l = buf_l[idx1] if idx1 >= 0 else 0.0
            fb_r = buf_r[idx2] if idx2 >= 0 else 0.0

            # Non-linear saturating junction
            loop_sample_l = np.tanh(gamma1 * fb_l - gamma2 * fb_r)
            loop_sample_r = np.tanh(gamma1 * fb_r - gamma2 * fb_l)

            buf_l[i] = (1.0 - alpha) * buf_l[i] + alpha * loop_sample_l
            buf_r[i] = (1.0 - alpha) * buf_r[i] + alpha * loop_sample_r

        out[:, 0] = buf_l[:length]
        out[:, 1] = buf_r[:length]
        return out

# ==============================================================================
# 2. HARD GRANULARIZER & SUB-BASS BOOSTER
# ==============================================================================

class GranularBassProcessor:
    def __init__(self, sample_rate: int = 44100):
        self.sr = sample_rate

    def process_bass(self, bass_audio: np.ndarray, drive: float = 3.5, grain_ms: float = 35.0):
        mono_bass = np.mean(bass_audio, axis=1)
        grain_len = int(self.sr * (grain_ms / 1000.0))
        hop_size = grain_len // 4
        num_samples = len(mono_bass)

        hanning_win = np.hanning(grain_len)
        granular_out = np.zeros(num_samples + grain_len)

        for i in range(0, num_samples - grain_len, hop_size):
            jitter = np.random.randint(-int(hop_size * 0.4), int(hop_size * 0.4) + 1)
            src_idx = max(0, min(num_samples - grain_len, i + jitter))
            grain = mono_bass[src_idx:src_idx + grain_len] * hanning_win

            if np.random.rand() < 0.20:
                grain = np.sin(grain * np.pi * 1.8) # Foldback

            granular_out[i:i + grain_len] += grain

        granular_out = granular_out[:num_samples]

        sos_lp = signal.butter(4, 80.0, 'low', fs=self.sr, output='sos')
        sub_fundamental = signal.sosfilt(sos_lp, mono_bass)
        zero_crossings = np.where(np.diff(np.signbit(sub_fundamental)))[0]
        sub_octave = np.zeros_like(sub_fundamental)
        polarity = 1.0
        for z_idx in range(len(zero_crossings) - 1):
            start = zero_crossings[z_idx]
            end = zero_crossings[z_idx + 1]
            sub_octave[start:end] = sub_fundamental[start:end] * polarity
            polarity *= -1.0

        boosted_mono = np.tanh((granular_out * drive) + (sub_octave * 1.4))

        bass_stereo = np.zeros_like(bass_audio
        bass_stereo[:, 0] = boosted_mono
        bass_stereo[:, 1] = np.roll(boosted_mono, 8)
        return bass_stereo

# ==============================================================================1
# 3. EPHEMERAL VOCAL / DRUM SAMPLING & ARRANGEMENT COMPOSER
# ==============================================================================

def assemble_remix(drums: np.ndarray, bass: np.ndarray, vocals: np.ndarray, other: np.ndarray,
                   sr: int = 44100, bpm: float = 128.0,
                   alpha: float = 0.32, gamma1: float = 0.75, gamma2: float = 0.55) -> np.ndarray:
    num_samples = min(len(drums), len(bass), len(vocals), len(other))
    bar_samples = int(sr * (60.0 / bpm) * 4)

    bass_engine = GranularBassProcessor(sample_rate=sr)
    toroid_engine = ToroidalGlitchEngine(sample_rate=sr, bpm=bpm)

    print("[1/5] Processing Bass Stem through Hard Granularizer and Sub-Octave Core...")
    bass_proc = bass_engine.process_bass(bass[:num_samples], drive=4.0, grain_ms=45.0)

    print("[2/5] Synthesizing Ephemeral Vocal Catches & Pitch-Displaced Stutters...")
    vocal_proc = np.copy(vocals[:num_samples])
    for bar in range(0, num_samples, bar_samples):
        if (bar // bar_samples) % 4 == 3:
            chop_start = bar + (bar_samples * 3 // 4)
            chop_end = min(bar + bar_samples, num_samples)
            slice_16th = int(sr * (60.0 / bpm) / 4)
            if chop_start + slice_16th < chop_end:
                unit = vocal_proc[chop_start:chop_start + slice_16th]
                for rep in range(chop_start, chop_end, slice_16th):
                    avail = min(slice_16th, chop_end - rep)
                    vocal_proc[rep:rep + avail] = unit[:avail] * (1.2 - (0.5 * (rep - chop_start) / (chop_end - chop_start)))

    print("[3/5] Computing Dynamic Deference & Sudden Drops...")
    mix = np.zeros((num_samples, 2), dtype=np.float32)
    total_bars = num_samples // bar_samples
    for b in range(total_bars):
        start = b * bar_samples
        end = min((b + 1) * bar_samples, num_samples)

        if b < 8:
            mix[start:end] += bass_proc[start:end] * 1.3
            mix[start:end] += drums[start:end] * 0.7
            if b >= 4:
                mix[start:end] += vocal_proc[start:end] * 0.85
        elif b == 8:
            silence_idx = start + (bar_samples // 2)
            mix[start:silence_idx] += vocal_proc[start:silence_idx] * 1.1
            mix[silence_idx:end] = 0.0
        elif 9 <= b < 17:
            mix[start:end] += bass_proc[start:end] * 1.45
            mix[start:end] += drums[start:end] * 1.2
            mix[start:end] += vocal_proc[start:end] * 0.9
            mix[start:end] += other[start:end] * 0.5
        elif 17 <= b < 21:
            sos_hp = signal.butter(4, 400.0, 'high', fs=sr, output='sos')
            bloom_vocal = signal.sosfilt(sos_hp, vocal_proc[start:end], axis=0)
            mix[start:end] += bloom_vocal * 1.2
            mix[start:end] += other[start:end] * 1.4
        else:
            mix[start:end] += bass_proc[start:end] * 1.2
            mix[start:end] += drums[start:end] * 1.0
            mix[start:end] += vocal_proc[start:end] * 0.75

    print(f"[4/5] Passing Full Composite through Box Toroidal Inversion Glitch (alpha={alpha}, gamma1={gamma1}, gamma2={gamma2})...")
    toroidal_master = toroid_engine.process_toroidal_inversion(mix, glitch_prob=0.30, alpha=alpha, gamma1=gamma1, gamma2=gamma2)

    final_output = (mix * 0.70) + (toroidal_master * 0.30)

    print("[5/5] Normalizing & Applying Soft Peak Compression...")
    max_val = np.max(np.abs(final_output))
    if max_val > 0.0:
        final_output = final_output / max_val
    final_output = np.tanh(final_output * 1.25) * 0.95

    return final_output

# ==============================================================================
# 4. EXECUTION HARNESS WITH COMMAND LINE INTERFACE
# ==============================================================================

if __name__ == "__main__":
    # Set up argument parser to handle configurable parameters via CLI
    parser = argparse.ArgumentParser(description="Toroidal Glitch & Granular Stem Remix Engine")
    parser.add_argument("--input", type=str, default="attached_track.wav", help="Input path of the source track")
    parser.add_argument("--output", type=str, default="remix_toroidal_reconstructed.wav", help="Output filename")
    parser.add_argument("--alpha", type=float, default=0.32, help="Feedback blend factor (default: 0.32)")
    parser.add_argument("--gamma1", type=float, default=0.75, help="First loop gain factor (default: 0.75)")
    parser.add_argument("--gamma2", type=float, default=0.55, help="Second loop gain factor (default: 0.55)")

    # Allows running inside a notebook or terminal dynamically
    if 'ipykernel_launcher' in sys.argv[0] or '-f' in sys.argv:
        args = parser.parse_args([]) # Use defaults in Colab interactive environment
    else:
        args = parser.parse_args()

    print(f"[*] Starting Autopoietic Remix Pipeline for: {args.input}")

    if not os.path.exists(args.input):
        print(f"[!] Target file '{args.input}' not found locally.")
        print("[*] Generating synthetic multi-stem test vector at 128 BPM...")
        sr = 44100
        duration = 104
        t = np.linspace(0, duration, sr * duration, endpoint=False)

        drums_mock = np.zeros((len(t), 2))
        bass_mock = np.zeros((len(t), 2))
        vocals_mock = np.zeros((len(t), 2))
        other_mock = np.zeros((len(t), 2))

        beat_interval = int(sr * (60.0 / 128.0))
        for beat in range(0, len(t), beat_interval):
            drums_mock[beat:min(beat+2000, len(t)), :] += 0.8

        bass_mock[:, 0] = 0.5 * np.sin(2 * np.pi * 55.0 * t)
        bass_mock[:, 1] = 0.5 * np.sin(2 * np.pi * 55.0 * t)
        vocals_mock[:, 0] = 0.3 * np.sin(2 * np.pi * 440.0 * t) * (t % 2.0 < 1.0)
        vocals_mock[:, 1] = 0.3 * np.sin(2 * np.pi * 440.0 * t) * (t % 2.0 < 1.0)

        remix_result = assemble_remix(drums_mock, bass_mock, vocals_mock, other_mock, sr=sr, bpm=128.0,
                                      alpha=args.alpha, gamma1=args.gamma1, gamma2=args.gamma2)
    else:
        print("[*] Running demucs stem separation CLI...")
        os.system(f"demucs -n htdemucs --two-stems=vocals '{args.input}'")
        # Implementation skeleton for demucs pipeline parsing would feed into assemble_remix with CLI args here
        pass

    wav_out = (remix_result * 32767.0).astype(np.int16)
    wavfile.write(args.output, 44100, wav_out)
    print(f"[✓] Remix Rendering Complete -> Saved to: {args.output}")
````
