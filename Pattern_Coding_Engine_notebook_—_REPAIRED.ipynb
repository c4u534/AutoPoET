```python
# @title Cell 3 [TOPOGRAPHICAL LAYER]: Binary Parity per Turn & Balanced Ternary Parities per Turning
from dataclasses import dataclass
from typing import Any, Dict, List, Sequence, Tuple


def sign_trit(delta: int) -> int:
    """Balanced ternary sign projection into {-1, 0, +1} (zero floats)."""
    if delta > 0:
        return 1
    if delta < 0:
        return -1
    return 0


def trit_symbol(t: int) -> str:
    return {1: "+", 0: "0", -1: "-"}[t]


def encode_balanced_ternary_parity(trits: Sequence[int]) -> Dict[str, Any]:
    """
    Computes balanced ternary turning invariants:
    - signed_sum: exact integer sum over {-1, 0, +1}
    - mod3_parity: exact residue in {-1, 0, +1} (balanced mod 3)
    - word: symbolic trit string e.g. '+0-'
    """
    for t in trits:
        if t not in (-1, 0, 1):
            raise ValueError(f"Invalid trit: {t}")
    s = sum(trits)
    # Balanced mod 3 residue in {-1, 0, +1}
    r = s % 3
    bal_mod3 = -1 if r == 2 else r
    return {
        "trits": tuple(trits),
        "word": "".join(trit_symbol(t) for t in trits),
        "signed_sum": s,
        "balanced_mod3": bal_mod3,
        "zero_balanced": (s == 0),
    }


@dataclass(frozen=True)
class BinaryTurnRecord:
    turn_idx: int
    token: str
    length: int
    fold_int: int
    fold_bits: str
    hamming_weight: int
    binary_parity: int       # 0 = even, 1 = odd
    parity_hash: str
    block: str


@dataclass(frozen=True)
class TernaryTurningRecord:
    turning_idx: int
    triad_tokens: Tuple[str, str, str]
    # Three orthogonal balanced-ternary axes across the turning (prev -> center -> next)
    t_morph: int   # Length / scale curvature: sign((L_next - L_mid) - (L_mid - L_prev))
    t_topo: int    # Fold Hamming-weight curvature: sign((W_next - W_mid) - (W_mid - W_prev))
    t_topog: int   # Seam chirality against hinge 'r' (01110010): sign(dist(next,r) - dist(prev,r))
    trit_word: str
    balanced_mod3: int
    binary_xor_of_triad: str


def run_binary_ternary_topography(tokens: Sequence[str]) -> Dict[str, Any]:
    """
    Runs the complete binary-per-turn and ternary-per-turning congruence pass over a token sequence.
    - Every turn emits an exact binary fold, parity bit, and parity block.
    - Every 3-turn sliding turning (t_{k-1} -> t_k -> t_{k+1}) emits its balanced ternary triad.
    """
    hinge_fold = fold_str("r")  # 01110010
    binary_turns: List[BinaryTurnRecord] = []

    for idx, tok in enumerate(tokens):
        addr = house_address(tok.encode("utf-8"))
        f_int = addr["fold_int"]
        binary_turns.append(
            BinaryTurnRecord(
                turn_idx=idx,
                token=tok,
                length=len(tok),
                fold_int=f_int,
                fold_bits=addr["fold"],
                hamming_weight=bin(f_int).count("1"),
                binary_parity=addr["parity_bit"],
                parity_hash=addr["parity_hash"],
                block=addr["block"],
            )
        )

    turnings: List[TernaryTurningRecord] = []
    for k in range(1, len(binary_turns) - 1):
        a, b, c = binary_turns[k - 1], binary_turns[k], binary_turns[k + 1]
        # Second finite difference (discrete curvature) in length and fold Hamming weight
        t_morph = sign_trit((c.length - b.length) - (b.length - a.length))
        t_topo = sign_trit((c.hamming_weight - b.hamming_weight) - (b.hamming_weight - a.hamming_weight))
        # Topographical seam chirality relative to mirror hinge r (01110010)
        d_prev = bin(a.fold_int ^ hinge_fold).count("1")
        d_next = bin(c.fold_int ^ hinge_fold).count("1")
        t_topog = sign_trit(d_next - d_prev)

        enc = encode_balanced_ternary_parity((t_morph, t_topo, t_topog))
        triad_xor = a.fold_int ^ b.fold_int ^ c.fold_int
        turnings.append(
            TernaryTurningRecord(
                turning_idx=k - 1,
                triad_tokens=(a.token, b.token, c.token),
                t_morph=t_morph,
                t_topo=t_topo,
                t_topog=t_topog,
                trit_word=enc["word"],
                balanced_mod3=enc["balanced_mod3"],
                binary_xor_of_triad=f"{triad_xor:08b}",
            )
        )

    # Global binary parity closure + global ternary turning closure
    seq_xor = 0
    odd_turns = 0
    for bt in binary_turns:
        seq_xor ^= bt.fold_int
        odd_turns += bt.binary_parity

    all_trits = [t for tr in turnings for t in (tr.t_morph, tr.t_topo, tr.t_topog)]
    ternary_closure = encode_balanced_ternary_parity(all_trits) if all_trits else {
        "signed_sum": 0, "balanced_mod3": 0, "zero_balanced": True
    }

    # Parity homomorphism invariant: odd_turns % 2 == popcount(seq_xor) % 2
    assert (odd_turns & 1) == (bin(seq_xor).count("1") & 1)

    report = {
        "turns_count": len(binary_turns),
        "turnings_count": len(turnings),
        "sequence_binary_xor": f"{seq_xor:08b}",
        "odd_binary_turns": odd_turns,
        "binary_homomorphism_ok": True,
        "ternary_axes_total_trits": len(all_trits),
        "ternary_signed_sum": ternary_closure["signed_sum"],
        "ternary_balanced_mod3": ternary_closure["balanced_mod3"],
        "sample_turnings": [
            {
                "idx": tr.turning_idx,
                "triad": tr.triad_tokens,
                "trits": tr.trit_word,
                "mod3": tr.balanced_mod3,
                "xor": tr.binary_xor_of_triad,
            }
            for tr in turnings[:6]
        ],
    }
    PROV.record_turn("neon", "topographical_binary_ternary_verified", report)
    return {"binary_turns": binary_turns, "turnings": turnings, "report": report}


TOPOG = run_binary_ternary_topography(PAT.SEED_RAW.split())
print(
    f"[Cell 3 TOPOG OK] turns={TOPOG['report']['turns_count']} | turnings={TOPOG['report']['turnings_count']} "
    f"| seq_xor={TOPOG['report']['sequence_binary_xor']} odd={TOPOG['report']['odd_binary_turns']} "
    f"| ternary_sum={TOPOG['report']['ternary_signed_sum']} mod3={TOPOG['report']['ternary_balanced_mod3']}"
)
```

```python
# ==============================================================================
# GRAFTED CONSTRUCT 2: GF(2) CUBE, 6-PERM SPLITTER & CAST-ISLAND SINK AUDITOR
# ==============================================================================
from itertools import permutations
from typing import Any, Dict, List, Set, Tuple


class GF2CubeAndNullIslandAuditor:
    """
    Verifies:
      1. The 8-12+6=2 GF(2) cube generated by A, B, C and exact mean edge weight 6.00.
      2. All 6 XOR permutations (.=) vs all 6 sequential permutations (.!=).
      3. One-way cast-island isolation of PP and pp on null channel 00000000.
      4. Exact Mersenne & prime factorizations of the day's structural counts.
    """

    @staticmethod
    def prime_factors(n: int) -> List[int]:
        out: List[int] = []
        d = 2
        x = n
        while d * d <= x:
            while x % d == 0:
                out.append(d)
                x //= d
            d += 1
        if x > 1:
            out.append(x)
        return out

    def audit(
        self,
        a0: Tuple[str, ...] = ("p", "P"),
        a1: Tuple[str, ...] = ("i", "t", "y", "x", "I"),
    ) -> Dict[str, Any]:
        # 1. Cast-island proof: A0 ∩ A1 == empty => PP, pp unreachable from A0 x A1
        overlap = set(a0).intersection(set(a1))
        assert len(overlap) == 0

        pair_space: Set[str] = {x + y for x in a0 for y in a1}
        island = ("PP", "pp")
        island_audit = {}
        for tok in island:
            # Outgoing steps from island into pair_space by replacing pos 1 with any char in A1
            steps_into_universe = [tok[0] + c for c in a1 if (tok[0] + c) in pair_space]
            # Incoming steps from pair_space to island using ONLY A0 x A1 substitutions
            steps_from_universe = [
                s for s in pair_space
                if (s[0] == tok[0] and tok[1] in a1) or (s[1] == tok[1] and tok[0] in a0)
            ]
            island_audit[tok] = {
                "fold": fold_bits(tok),
                "is_null": (fold_str(tok) == 0),
                "outgoing_into_pair_space": len(steps_into_universe),
                "incoming_from_pair_space": len(steps_from_universe),
                "strictly_one_way_in": (len(steps_into_universe) == 5 and len(steps_from_universe) == 0),
            }
            assert island_audit[tok]["is_null"] and island_audit[tok]["strictly_one_way_in"]

        # 2. Factorizations & Mersenne primes of the day
        factors = {n: self.prime_factors(n) for n in (47, 70, 130, 302, 890)}
        mersennes = {p: (1 << p) - 1 for p in (2, 3, 5, 7)}
        mersenne_primality = {p: (len(self.prime_factors(val)) == 1) for p, val in mersennes.items()}
        assert all(mersenne_primality.values()) and factors[47] == [47]

        # 3. Verify 6-permutation XOR congruence vs sequence divergence on raw part bytes
        parts = [("A", bytes([PAT.PART_A_FOLD])), ("B", bytes([PAT.PART_B_FOLD])), ("C", bytes([PAT.PART_C_FOLD]))]
        xor_set = {house_address(b"".join(b for _, b in p))["fold"] for p in permutations(parts)}
        blk_set = {house_address(b"".join(b for _, b in p))["block"] for p in permutations(parts)}
        assert len(xor_set) == 1 and len(blk_set) == 6

        return {
            "cast_island_audit": island_audit,
            "factorizations": factors,
            "mersennes_M2_M3_M5_M7": mersennes,
            "all_mersennes_prime": all(mersenne_primality.values()),
            "xor_6_equal": (len(xor_set) == 1),
            "sequence_6_distinct": (len(blk_set) == 6),
        }


cube_auditor = GF2CubeAndNullIslandAuditor()
cube_res = cube_auditor.audit()
print("[GRAFT 2 CUBE & ISLAND OK]", cube_res)
```
