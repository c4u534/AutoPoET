````
import threading
from collections import defaultdict
from dataclasses import dataclass, field
from typing import Dict

# Extend TelemetryBucket to support thread-level diagnostic accounting
@dataclass
class ThreadAwareTelemetryBucket:
    stage_name: str
    target_cycles: int
    executed_cycles: int = 0
    start_time: float = 0.0
    end_time: float = 0.0

    # Thread-specific dictionaries
    thread_cycles: Dict[int, int] = field(default_factory=lambda: defaultdict(int))
    thread_repairs: Dict[int, int] = field(default_factory=lambda: defaultdict(int))
    thread_collisions: Dict[int, int] = field(default_factory=lambda: defaultdict(int))
    thread_punctures: Dict[int, int] = field(default_factory=lambda: defaultdict(int))

    @property
    def duration(self) -> float:
        return max(self.end_time - self.start_time, 1e-6)

    def log_event(self, thread_id: int, event_type: str):
        self.thread_cycles[thread_id] += 1
        if event_type == 'repair':
            self.thread_repairs[thread_id] += 1
        elif event_type == 'collision':
            self.thread_collisions[thread_id] += 1
        elif event_type == 'puncture':
            self.thread_punctures[thread_id] += 1

    def print_thread_breakdown(self):
        print(f"\n===========================================================")
        print(f"      THREAD-SPECIFIC TELEMETRY BREAKDOWN: {self.stage_name.upper()}")
        print(f"===========================================================")
        all_threads = set(self.thread_cycles.keys())
        for tid in sorted(all_threads):
            cycles = self.thread_cycles[tid]
            repairs = self.thread_repairs[tid]
            collisions = self.thread_collisions[tid]
            punctures = self.thread_punctures[tid]
            total_errors = collisions + punctures
            error_rate = (total_errors / cycles * 100) if cycles > 0 else 0.0
            print(f"Thread ID [0x{tid:X}] | Cycles: {cycles:,} | Repairs: {repairs} | Collisions: {collisions} | Punctures: {punctures} | Error Rate: {error_rate:.4f}%")
        print(f"===========================================================")

print("[+] ThreadAwareTelemetryBucket Defined.")
````

````
import ctypes
import subprocess
import time
import numpy as np

# 1. Write the optimized C source with native bit-shifting and CPU popcounts
with open("vmm_arbitration_fast.c", "w") as f:
    f.write(r"""
#include <stdint.h>
#include <stdlib.h>

#define MERSENNE_31 2147483647
#define OUTER_HEX_MASK 0xF0000F

// Fast modulo for Mersenne Prime 2^31 - 1 via shift-and-add
inline uint32_t fast_m31_mod(uint64_t x) {
    uint64_t sum = (x & MERSENNE_31) + (x >> 31);
    sum = (sum & MERSENNE_31) + (sum >> 31);
    if (sum >= MERSENNE_31) {
        sum -= MERSENNE_31;
    }
    return (uint32_t)sum;
}

// Native high-speed arbitration returning 1 (promote to SHA-256) or 0 (retain in SHA-512)
int32_t native_arbitrate_parity(uint32_t val24, int32_t entropy_wt) {
    uint64_t raw_product = (uint64_t)val24 * (uint64_t)entropy_wt;
    uint32_t residue = fast_m31_mod(raw_product);

    // Sum the 31-bit chunk values
    uint32_t chunk_sum = (residue & 0x7FFFFFFF) + (residue >> 31);

    // Check outer hexadecimal binarial parity using compiler-native popcount
    uint32_t masked = val24 & OUTER_HEX_MASK;
    uint32_t binary_sequential_parity = __builtin_parity(masked);

    return (chunk_sum % 2 == binary_sequential_parity) ? 1 : 0;
}
""")

# 2. Compile to native shared object library
subprocess.run(["gcc", "-O3", "-shared", "-fPIC", "vmm_arbitration_fast.c", "-o", "libvmm_arbitration_fast.so"], check=True)
print("[SUCCESS] Compiled libvmm_arbitration_fast.so with O3 optimizations.")
````

````
%%writefile cython_vmm_arbitrator.pyx
# cython: language_level=3, boundscheck=False, wraparound=False, initializedcheck=False, nonecheck=False

from libc.stdint cimport uint32_t, int32_t

cdef extern from "vmm_arbitration_fast.c":
    int32_t native_arbitrate_parity(uint32_t val24, int32_t entropy_wt) nogil

def cython_arbitrate_parity(uint32_t val24, int32_t entropy_wt):
    """
    Cython FFI Wrapper bypassing standard ctypes marshaling overhead.
    Using nogil allows multi-threaded scaling without Python interpreter lock contention.
    """
    cdef int32_t res
    with nogil:
        res = native_arbitrate_parity(val24, entropy_wt)
    return res
````
