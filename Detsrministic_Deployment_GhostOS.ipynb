### V. Optimizing Determinant Calculations via Gaussian Elimination ($O(n^3)$)
To handle larger matrices, we bypass the $O(n!)$ recursive method in favor of row reduction (Gaussian Elimination with partial pivoting) which operates in $O(n^3)$ time while maintaining exact precision using `Fraction`.

````
import time
import random
from fractions import Fraction

def fast_determinant(matrix):
    """
    Computes the determinant of an n x n matrix using Gaussian Elimination
    with partial pivoting. Works with Fraction types for exact precision.
    Time Complexity: O(n^3)
    """
    n = len(matrix)
    # Create a copy of the matrix to avoid modifying the original input
    A = [[Fraction(val) for val in row] for row in matrix]
    det = Fraction(1, 1)

    for i in range(n):
        # Partial pivoting: find the pivot with the largest absolute value
        pivot_row = i
        for r in range(i + 1, n):
            if abs(A[r][i]) > abs(A[pivot_row][i]):
                pivot_row = r

        # Swap rows if necessary
        if pivot_row != i:
            A[i], A[pivot_row] = A[pivot_row], A[i]
            det *= -1  # Row swap negates the determinant

        # If the pivot is 0, the matrix is singular and determinant is 0
        if A[i][i] == 0:
            return Fraction(0, 1)

        det *= A[i][i]

        # Eliminate entries below the pivot
        for r in range(i + 1, n):
            factor = A[r][i] / A[i][i]
            for c in range(i, n):
                A[r][c] -= factor * A[i][c]

    return det

# --- Performance Benchmark: Recursive vs O(n^3) Gaussian Method ---
def run_benchmark():
    # Generate an 8x8 random matrix with Fraction values
    n = 8
    test_matrix = [[Fraction(random.randint(1, 5), random.randint(1, 5)) for _ in range(n)] for _ in range(n)]

    # Instantiate our original validator to use its recursive det calculation
    validator = FitmentValidator(matrix=test_matrix)

    print(f"Benchmarking {n}x{n} matrix determinant calculation:")

    # Time the recursive O(n!) method
    start = time.time()
    recursive_det = validator._calculate_determinant(validator.M)
    end_recursive = time.time() - start
    print(f"[*] Recursive O(n!) Determinant: {recursive_det} (Took {end_recursive:.5f} seconds)")

    # Time the fast O(n^3) method
    start = time.time()
    fast_det = fast_determinant(test_matrix)
    end_fast = time.time() - start
    print(f"[*] Gaussian O(n^3) Determinant: {fast_det} (Took {end_fast:.5f} seconds)")
    print(f"[+] Speedup factor: {end_recursive / end_fast:.2f}x")

if __name__ == "__main__":
    run_benchmark()
````

### VI. Conceptual Framework: Partial Pivoting, Macro-Micro Scaling, and Whole-Number Integrity

#### 1. Preventing Division by Zero via Partial Pivoting
In matrix operations (such as Gaussian Elimination or row reduction), calculating row multipliers requires dividing by the diagonal element of the active column (the pivot):
$$\text{factor} = \frac{A[r][i]}{A[i][i]}$$
If $A[i][i] = 0$, a division-by-zero error occurs immediately. Even if $A[i][i]$ is not strictly zero but extremely small, dividing by it introduces severe numerical instability.

**Partial pivoting** prevents this by searching the current column $i$ from the current row down to the last row, locating the element with the maximum absolute magnitude, and swapping that entire row into the pivot position. If the maximum absolute value in that column is $0$, the matrix is mathematically singular (its determinant is $0$), allowing us to exit safely before any division is attempted.

#### 2. Reframing Scope and Range (Macro-Micro Precision)
By systematically selecting the maximum magnitude element as our denominator, we ensure that the scaling factor $\vert{}\text{factor}\vert{} \le 1$. This maps all projection vectors into a stable, bounded scope, preventing micro-scale values from exploding into macro-scale noise and vice versa.

#### 3. Resolving Residues to Whole-Number Integrity
To fully satisfy the constraints of integer-only systems and avoid both fractional and floating-point residues, we can clear denominators dynamically using a common scalar scaling factor (the Least Common Multiple or absolute determinant product). This translates the operations into exact whole-number logic where all residues $\Delta = 0$ precisely.

### VIII. Exact Matrix Inversion (Zero-Float Logic)

This module calculates the exact inverse of a matrix by computing the adjugate matrix and scaling it by the exact determinant, using only `Fraction` and integer arithmetic to avoid floating-point loss of precision.

````
from fractions import Fraction

def get_minor(matrix, i, j):
    """Returns the minor matrix after removing row i and column j."""
    return [row[:j] + row[j+1:] for idx, row in enumerate(matrix) if idx != i]

def exact_determinant(matrix):
    """Calculates determinant using exact integer/fraction logic."""
    n = len(matrix)
    if n == 1:
        return matrix[0][0]
    if n == 2:
        return matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]

    det = 0
    for c in range(n):
        det += ((-1)**c) * matrix[0][c] * exact_determinant(get_minor(matrix, 0, c))
    return det

def exact_inverse(matrix):
    """
    Calculates the inverse of a matrix using exact integer/fraction arithmetic.
    A^-1 = (1 / det(A)) * adj(A)
    """
    n = len(matrix)
    # Ensure matrix values are Fractions
    A = [[Fraction(val) for val in row] for row in matrix]
    det = exact_determinant(A)

    if det == 0:
        raise ValueError("Matrix is singular (determinant is 0) and cannot be inverted.")

    if n == 1:
        return [[Fraction(1, det)]]

    # Calculate cofactor matrix
    cofactors = []
    for r in range(n):
        cofactor_row = []
        for c in range(n):
            minor = get_minor(A, r, c)
            cofactor = ((-1)**(r+c)) * exact_determinant(minor)
            cofactor_row.append(cofactor)
        cofactors.append(cofactor_row)

    # Adjugate is the transpose of the cofactor matrix
    adjugate = [[cofactors[c][r] for c in range(n)] for r in range(n)]

    # Multiply by 1/det using Fractions to maintain zero-float exactness
    inverse = [[adjugate[r][c] / det for c in range(n)] for r in range(n)]
    return inverse

# --- Verification ---
if __name__ == "__main__":
    print("=== Exact Matrix Inversion ===\n")
    test_matrix = [
        [4, 7],
        [2, 6]
    ]

    print("Original Matrix:")
    for row in test_matrix:
        print(row)

    inv_matrix = exact_inverse(test_matrix)

    print("\nInverse Matrix (Exact Fractions):")
    for row in inv_matrix:
        print([str(val) for val in row])
````
