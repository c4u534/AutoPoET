````
# !/usr/bin/env python3
"""
================================================================================
PAX-AUTOPOIETIC EPHEMERAL FRAMER & SELF-REPAIRING ASSIMILATION ENGINE
================================================================================
Implements:
  1. Multi-Runtime Ephemeral Decoupling (2-of-2 / 3-of-3 Runtime Survival Proof)
  2. Interlooped 15s Heartbeat -> 1m15s (75s) Pulse-Beat Pool -> 15th-Pulse Transport
  3. Dual-Copy Folder Metadata Pattern Hashing (meta_pattern_copy_1 & copy_2)
  4. Shorter-Filename Process Assimilation, Optimization & Distortion Ejection
  5. Shortcut-to-Shortcut Endpoint-1 Overwhelm -> Endpoint-2 Loopback Socket Rebuild
     with Shifted Pulse Metric & Local Provenance Ledger Self-Repair
================================================================================
"""

import ast
import copy
import hashlib
import json
import logging
import math
import mmap
import os
import shutil
import socket
import struct
import sys
import threading
import time
from datetime import datetime, timezone
from pathlib import Path
from typing import Dict, List, Optional, Tuple, Any

# ==============================================================================
# SECTION 1: CORE ARCHITECTURAL CONSTANTS & TIMING METRICS
# ==============================================================================
CANONICAL_LONG_NAME = "pax_ephemeral_autopoietic_framer_20261108_full.py"
CANONICAL_SHORT_STEM = "pax_ephemeral"

FOLDER_ID = "1MFUTmQvKR24bx8ClFIbOksywUPq2oN1M"
SESSION_ID = "ab272eee-855f-4825-a089-7400e2072b18"
AXIAL_REGISTER_HEX = 0x78E8
ISOMORPHIC_GROUND = 0.84210000

# Standard Cadence: 15s Heart, 75s (1m15s) Pulse-Beat, 15th Pulse Transport
DEFAULT_HEART_INTERVAL_S = 15.0
DEFAULT_PULSES_PER_HEART_SWEEP = 5       # 5 * 15s = 75s (1 min 15 sec)
DEFAULT_TRANSPORT_PULSE_MODULO = 15      # Every 15th pulse transports metadata hash

# Shifted Cadence after Panic Loopback Rebuild (Coprime metric to block resonance)
SHIFTED_HEART_INTERVAL_S = 11.0
SHIFTED_PULSES_PER_HEART_SWEEP = 5       # 5 * 11s = 55s

# Identity Distortion Threshold for Shorter-Filename Assimilation Ejection
DISTORTION_PANIC_THRESHOLD = 0.25        # Eject upgrade if distortion > 25%


# ==============================================================================
# SECTION 2: EPHEMERAL MULTI-RUNTIME DECOUPLING TRACKER (2 OR 3 RUNTIMES)
# ==============================================================================
class EphemeralRuntimeDecoupler:
    """
    Tracks exposure across N in {2, 3} runtimes (e.g., Termux, Colab-xterm, VM-Muse).
    When the process observes each runtime individually end/wipe without the
    process state being lost, and then observes the 2nd of 2 or 3rd of 3 runtimes
    again, it transitions to DECOUPLED_AUTOPOIETIC_RUNTIME=True.
    """
    def __init__(self, state_dir: Path, required_runtimes: int = 3):
        self.state_dir = state_dir
        self.required_runtimes = max(2, min(3, required_runtimes))
        self.ledger_path = self.state_dir / "ephemeral_decoupling_ledger.json"
        self.state = self._load_or_init()

    def _load_or_init(self) -> Dict[str, Any]:
        if self.ledger_path.exists():
            try:
                return json.loads(self.ledger_path.read_text(encoding="utf-8"))
            except Exception:
                pass
        data = {
            "required_runtimes": self.required_runtimes,
            "known_runtimes": {},          # runtime_id -> {"allocations": int, "wipes_survived": int}
            "survival_observations": [],   # ordered log of observed runtime ends survived
            "re_exposure_count": 0,
            "decoupled": False,
            "decoupled_utc": None,
            "equal_offset_artifacts": []
        }
        self._save(data)
        return data

    def _save(self, data: Optional[Dict[str, Any]] = None) -> None:
        target = data if data is not None else self.state
        self.state_dir.mkdir(parents=True, exist_ok=True)
        tmp = self.ledger_path.with_suffix(".tmp")
        tmp.write_text(json.dumps(target, indent=2, sort_keys=True), encoding="utf-8")
        os.replace(tmp, self.ledger_path)

    def observe_runtime_allocation(self, runtime_id: str) -> Dict[str, Any]:
        rt = self.state["known_runtimes"].setdefault(
            runtime_id, {"allocations": 0, "wipes_survived": 0, "last_seen_utc": ""}
        )
        rt["allocations"] += 1
        rt["last_seen_utc"] = datetime.now(timezone.utc).isoformat()

        # Check if all required runtimes have been observed ending/changing at least once
        survived_rts = [
            k for k, v in self.state["known_runtimes"].items()
            if v["wipes_survived"] >= 1
        ]
        if len(survived_rts) >= self.required_runtimes:
            self.state["re_exposure_count"] += 1
            if self.state["re_exposure_count"] >= self.required_runtimes and not self.state["decoupled"]:
                self.state["decoupled"] = True
                self.state["decoupled_utc"] = datetime.now(timezone.utc).isoformat()

        self._save()
        return copy.deepcopy(self.state)

    def observe_runtime_end_survived(self, runtime_id: str, reason: str = "wipe_or_reset") -> Dict[str, Any]:
        rt = self.state["known_runtimes"].setdefault(
            runtime_id, {"allocations": 1, "wipes_survived": 0, "last_seen_utc": ""}
        )
        rt["wipes_survived"] += 1
        rt["last_seen_utc"] = datetime.now(timezone.utc).isoformat()
        self.state["survival_observations"].append({
            "runtime_id": runtime_id,
            "reason": reason,
            "wipes_survived": rt["wipes_survived"],
            "timestamp_utc": rt["last_seen_utc"]
        })
        self._save()
        return copy.deepcopy(self.state)


# ==============================================================================
# SECTION 3: DUAL-COPY FOLDER METADATA PATTERN HASHER & LOCAL REPAIR LEDGER
# ==============================================================================
class DualCopyMetadataLedger:
    """
    Maintains 2 rolling copies of the folder structure's metadata pattern hash
    (meta_pattern_copy_1.json and meta_pattern_copy_2.json) plus a byte-exact
    provenance repair store so the construct can self-repair if complexity >= Theta.
    """
    IGNORED_NAMES = {
        "heart_side.log", "pulse.log", "pulse_pool.jsonl", "event.log",
        "meta_pattern_copy_1.json", "meta_pattern_copy_2.json",
        "LOCKDOWN.lock", "ephemeral_decoupling_ledger.json"
    }

    def __init__(self, anchor_dir: Path):
        self.anchor_dir = anchor_dir
        self.state_dir = anchor_dir / "state"
        self.copy_1_path = self.state_dir / "meta_pattern_copy_1.json"
        self.copy_2_path = self.state_dir / "meta_pattern_copy_2.json"
        self.repair_vault_path = self.state_dir / "local_repair_ledger.json"
        self.state_dir.mkdir(parents=True, exist_ok=True)

    def compute_folder_pattern(self) -> Dict[str, Any]:
        entries = []
        merkle_leaves = []

        for root, dirs, files in os.walk(self.anchor_dir):
            dirs[:] = sorted(d for d in dirs if not d.startswith(".") and d != "state")
            for fname in sorted(files):
                if fname.startswith(".") or fname.endswith(".tmp") or fname in self.IGNORED_NAMES:
                    continue
                fpath = Path(root) / fname
                try:
                    st = fpath.stat()
                    raw = fpath.read_bytes()
                    f_sha = hashlib.sha256(raw).hexdigest()
                except OSError:
                    continue
                rel = str(fpath.relative_to(self.anchor_dir))
                item = {
                    "relpath": rel,
                    "size": st.st_size,
                    "mode": oct(st.st_mode & 0o777),
                    "sha256": f_sha
                }
                entries.append(item)
                leaf = hashlib.sha256(f"{rel}:{st.st_size}:{f_sha}".encode("utf-8")).hexdigest()
                merkle_leaves.append(leaf)

        combined = "|".join(merkle_leaves).encode("utf-8")
        pattern_hash = hashlib.sha256(combined).hexdigest()
        return {
            "timestamp_utc": datetime.now(timezone.utc).isoformat(),
            "anchor_dir": str(self.anchor_dir),
            "file_count": len(entries),
            "pattern_hash": pattern_hash,
            "files": entries
        }

    def write_dual_copies(self) -> Dict[str, Any]:
        pattern = self.compute_folder_pattern()
        payload = json.dumps(pattern, indent=2, sort_keys=True)
        for target in (self.copy_1_path, self.copy_2_path):
            tmp = target.with_suffix(".tmp")
            tmp.write_text(payload, encoding="utf-8")
            os.replace(tmp, target)
        return pattern

    def verify_dual_copies_match(self) -> Tuple[bool, Optional[Dict[str, Any]]]:
        if not (self.copy_1_path.exists() and self.copy_2_path.exists()):
            return False, None
        try:
            b1 = self.copy_1_path.read_bytes()
            b2 = self.copy_2_path.read_bytes()
            if hashlib.sha256(b1).hexdigest() != hashlib.sha256(b2).hexdigest():
                return False, None
            return True, json.loads(b1.decode("utf-8"))
        except Exception:
            return False, None

    def snapshot_for_self_repair(self) -> None:
        """Stores verified file contents in the local repair ledger for autonomous recovery."""
        vault = {"timestamp_utc": datetime.now(timezone.utc).isoformat(), "blobs": {}}
        for root, dirs, files in os.walk(self.anchor_dir):
            dirs[:] = [d for d in dirs if not d.startswith(".") and d != "state"]
            for fname in files:
                if fname in self.IGNORED_NAMES or fname.endswith(".tmp"):
                    continue
                fpath = Path(root) / fname
                try:
                    rel = str(fpath.relative_to(self.anchor_dir))
                    raw_text = fpath.read_text(encoding="utf-8")
                    vault["blobs"][rel] = {
                        "sha256": hashlib.sha256(raw_text.encode("utf-8")).hexdigest(),
                        "content": raw_text
                    }
                except Exception:
                    continue
        tmp = self.repair_vault_path.with_suffix(".tmp")
        tmp.write_text(json.dumps(vault, indent=2), encoding="utf-8")
        os.replace(tmp, self.repair_vault_path)

    def repair_from_local_ledger(self) -> Tuple[bool, List[str]]:
        """Reconstructs missing or distorted files directly from the local repair ledger."""
        if not self.repair_vault_path.exists():
            return False, ["repair_vault_missing"]
        repaired = []
        try:
            vault = json.loads(self.repair_vault_path.read_text(encoding="utf-8"))
            for rel, info in vault.get("blobs", {}).items():
                target = self.anchor_dir / rel
                needs_repair = False
                if not target.exists():
                    needs_repair = True
                else:
                    curr_sha = hashlib.sha256(target.read_bytes()).hexdigest()
                    if curr_sha != info["sha256"]:
                        needs_repair = True
                if needs_repair:
                    target.parent.mkdir(parents=True, exist_ok=True)
                    target.write_text(info["content"], encoding="utf-8")
                    repaired.append(rel)
            self.write_dual_copies()
            return True, repaired
        except Exception as e:
            return False, [f"repair_error: {e}"]


# ==============================================================================
# SECTION 4: SHORTER-FILENAME PROCESS ASSIMILATOR & DISTORTION AUDITOR
# ==============================================================================
class ShorterFilenameAssimilator:
    """
    Detects parallel processes launched from a similar/shorter filename.
    Willfully assimilates if core invariants match and filename is shorter.
    Continuously measures AST/invariant distortion as the shorter-named file
    optimizes or upgrades; forces an event write and ejects when distortion
    exceeds DISTORTION_PANIC_THRESHOLD.
    """
    def __init__(self, native_script_path: Path, threshold: float = DISTORTION_PANIC_THRESHOLD):
        self.native_path = native_script_path
        self.native_name = native_script_path.name
        self.threshold = threshold
        self.native_signature = self._extract_signature(native_script_path)
        self.assimilated_target: Optional[Path] = None

    @staticmethod
    def _extract_signature(script_path: Path) -> Dict[str, Any]:
        if not script_path.exists():
            return {"valid_syntax": False, "functions": set(), "constants": {}, "sha256": ""}
        try:
            src = script_path.read_text(encoding="utf-8")
            tree = ast.parse(src)
            funcs = {node.name for node in ast.walk(tree) if isinstance(node, ast.FunctionDef)}
            consts = {}
            for node in ast.walk(tree):
                if isinstance(node, ast.Assign) and len(node.targets) == 1:
                    t = node.targets[0]
                    if isinstance(t, ast.Name) and t.id in {"FOLDER_ID", "SESSION_ID", "AXIAL_REGISTER_HEX"}:
                        if isinstance(node.value, ast.Constant):
                            consts[t.id] = node.value.value
            return {
                "valid_syntax": True,
                "functions": funcs,
                "constants": consts,
                "sha256": hashlib.sha256(src.encode("utf-8")).hexdigest()
            }
        except Exception:
            return {"valid_syntax": False, "functions": set(), "constants": {}, "sha256": ""}

    def compute_distortion(self, candidate_path: Path) -> Tuple[float, str]:
        cand_sig = self._extract_signature(candidate_path)
        if not cand_sig["valid_syntax"]:
            return 1.0, "SYNTAX_INVALID"

        # Check anchor invariant preservation
        for k, v in self.native_signature["constants"].items():
            if cand_sig["constants"].get(k) != v:
                return 0.95, f"INVARIANT_MISMATCH:{k}"

        f_native = self.native_signature["functions"]
        f_cand = cand_sig["functions"]
        if not f_native and not f_cand:
            return 0.0, "IDENTICAL_EMPTY"

        intersection = len(f_native & f_cand)
        union = len(f_native | f_cand)
        jaccard_similarity = intersection / max(1, union)
        distortion = round(1.0 - jaccard_similarity, 6)
        return distortion, "OK" if distortion <= self.threshold else "IDENTITY_DISTORTION_EXCEEDED"

    def evaluate_candidate(self, candidate_path: Path) -> Dict[str, Any]:
        cand_name = candidate_path.name
        is_shorter = (
            len(cand_name) < len(self.native_name)
            and cand_name.startswith(CANONICAL_SHORT_STEM)
            and cand_name.endswith(".py")
        )
        distortion, reason = self.compute_distortion(candidate_path)

        if not is_shorter:
            return {
                "action": "ALIEN_PROCESS_DETECTED",
                "candidate": cand_name,
                "is_shorter": False,
                "distortion": distortion,
                "reason": "NON_SHORTER_OR_ALIEN_NAME"
            }

        if self.assimilated_target is None:
            # Initial assimilation requires identical pattern (distortion == 0.0)
            if distortion == 0.0:
                self.assimilated_target = candidate_path
                return {
                    "action": "WILLFUL_ASSIMILATION",
                    "candidate": cand_name,
                    "is_shorter": True,
                    "distortion": distortion,
                    "reason": "IDENTICAL_CORE_SHORTER_NAME"
                }
            else:
                return {
                    "action": "REJECT_INITIAL_ASSIMILATION",
                    "candidate": cand_name,
                    "is_shorter": True,
                    "distortion": distortion,
                    "reason": reason
                }
        else:
            # Already assimilated: allow optimization until distortion > threshold
            if distortion <= self.threshold:
                return {
                    "action": "ASSIMILATE_OPTIMIZATION_UPGRADE",
                    "candidate": cand_name,
                    "is_shorter": True,
                    "distortion": distortion,
                    "reason": "WITHIN_TOLERANCE"
                }
            else:
                # EJECT from shorter-named process and force event write!
                self.assimilated_target = None
                return {
                    "action": "EJECT_UPGRADE_AND_PANIC_SYNC",
                    "candidate": cand_name,
                    "is_shorter": True,
                    "distortion": distortion,
                    "reason": reason
                }


# ==============================================================================
# SECTION 5: SHORTCUT-TO-SHORTCUT FAILOVER & LOOPBACK SOCKET LOCK CONTROLLER
# ==============================================================================
class EmergencyFailoverAndLoopbackSocket:
    """
    Manages:
      - Endpoint 1: Dedicated signal-out shortcut-to-shortcut link (queue capacity Q_max = 3).
        When overwhelmed by emergency flags, isolates the node and fails over to Endpoint 2.
      - Endpoint 2: Loopback WebSocket wired to local socket's trigger and lock.
        Closes and kills the active socket, rebuilds a new socket on a shifted pulse metric,
        and either locks down the local folder or triggers autonomous ledger self-repair.
    """
    def __init__(self, anchor_dir: Path, ledger: DualCopyMetadataLedger, self_repair_enabled: bool = True):
        self.anchor_dir = anchor_dir
        self.ledger = ledger
        self.self_repair_enabled = self_repair_enabled
        self.endpoint_1_shortcut_queue: List[Dict[str, Any]] = []
        self.endpoint_1_max_capacity = 3
        self.socket_generation = 1
        self.socket_alive = True
        self.current_heart_interval = DEFAULT_HEART_INTERVAL_S
        self.current_pulses_per_sweep = DEFAULT_PULSES_PER_HEART_SWEEP
        self.folder_locked_down = False

    def route_emergency_flags(self, first_occurrence_path: str, flags: List[str]) -> Dict[str, Any]:
        """
        Routes all panic notifications through the point of first occurrence to
        overwhelm Endpoint 1 (shortcut-to-shortcut link), forcing failover to Endpoint 2.
        """
        for flag in flags:
            self.endpoint_1_shortcut_queue.append({
                "point_of_first_occurrence": first_occurrence_path,
                "flag": flag,
                "ts": datetime.now(timezone.utc).isoformat()
            })

        if len(self.endpoint_1_shortcut_queue) >= self.endpoint_1_max_capacity:
            return self._failover_to_endpoint_2_loopback(first_occurrence_path)

        return {
            "status": "ENDPOINT_1_BUFFERING",
            "queue_depth": len(self.endpoint_1_shortcut_queue)
        }

    def _failover_to_endpoint_2_loopback(self, trigger_path: str) -> Dict[str, Any]:
        # 1. Endpoint 2 loopback captures error log, closes and kills old socket
        old_gen = self.socket_generation
        self.socket_alive = False

        # 2. Rebuild a new socket with the appropriate endpoint but a DIFFERENT pulse metric
        self.socket_generation += 1
        self.current_heart_interval = SHIFTED_HEART_INTERVAL_S
        self.current_pulses_per_sweep = SHIFTED_PULSES_PER_HEART_SWEEP
        self.endpoint_1_shortcut_queue.clear()
        self.socket_alive = True

        # 3. Propagate panic signal in local folder: Lockdown vs. Ledger Self-Repair
        if self.self_repair_enabled:
            repaired_ok, repaired_files = self.ledger.repair_from_local_ledger()
            if repaired_ok:
                return {
                    "status": "LOOPBACK_REBUILT_AND_SELF_REPAIRED",
                    "killed_socket_gen": old_gen,
                    "new_socket_gen": self.socket_generation,
                    "new_heart_interval_s": self.current_heart_interval,
                    "new_pulse_sweep_s": self.current_heart_interval * self.current_pulses_per_sweep,
                    "repaired_files": repaired_files,
                    "trigger_path": trigger_path
                }

        # Fallback: Lock down local folder and end ephemeral autopoiesis
        self.folder_locked_down = True
        lock_file = self.anchor_dir / "LOCKDOWN.lock"
        lock_file.write_text(json.dumps({
            "locked_utc": datetime.now(timezone.utc).isoformat(),
            "reason": "PANIC_SIGNAL_PROPAGATED_NO_REPAIR",
            "trigger_path": trigger_path
        }, indent=2), encoding="utf-8")
        return {
            "status": "FOLDER_LOCKED_DOWN_AUTOPOIESIS_ENDED",
            "killed_socket_gen": old_gen,
            "lock_file": str(lock_file)
        }


# ==============================================================================
# SECTION 6: UNIFIED INTERLOOPED DAEMON CONTROLLER
# ==============================================================================
class InterloopedAutopoieticDaemon:
    """
    Coordinates the 15s Heartbeat, 1m15s (75s) Pulse-Beat Pool Cleanup,
    15th-Pulse Dual-Copy Metadata Transport, Folder Monitor, and WebSocket Sync.
    """
    def __init__(self, anchor_dir: Path, self_repair_complexity: bool = True):
        self.anchor_dir = anchor_dir
        self.anchor_dir.mkdir(parents=True, exist_ok=True)
        self.heart_log = self.anchor_dir / "heart_side.log"
        self.pulse_log = self.anchor_dir / "pulse.log"
        self.pulse_pool = self.anchor_dir / "pulse_pool.jsonl"
        self.event_log = self.anchor_dir / "event.log"

        self.decoupler = EphemeralRuntimeDecoupler(self.anchor_dir / "state", required_runtimes=3)
        self.ledger = DualCopyMetadataLedger(self.anchor_dir)
        self.native_script = self.anchor_dir / CANONICAL_LONG_NAME
        if not self.native_script.exists():
            self.native_script.write_text(Path(__file__).read_text(encoding="utf-8") if "__file__" in globals() else "# native root\n", encoding="utf-8")

        self.assimilator = ShorterFilenameAssimilator(self.native_script)
        self.failover = EmergencyFailoverAndLoopbackSocket(
            self.anchor_dir, self.ledger, self_repair_enabled=self_repair_complexity
        )

        self.heart_counter = 0
        self.pulse_counter = 0
        self.websocket_open = False
        self.remote_Autosync_events: List[Dict[str, Any]] = []

    def step_heart(self) -> Dict[str, Any]:
        """Executes one 15-second heartbeat tick."""
        if self.failover.folder_locked_down:
            return {"status": "LOCKED_DOWN"}

        self.heart_counter += 1
        ts = datetime.now(timezone.utc).isoformat()
        heart_line = json.dumps({"kind": "heart", "seq": self.heart_counter, "ts": ts}) + "\n"
        with open(self.heart_log, "a", encoding="utf-8") as f:
            f.write(heart_line)

        result = {"heart_seq": self.heart_counter, "pulse_triggered": False}

        # Every 5 hearts (5 * 15s = 75s = 1 min 15 sec), execute Pulse-Beat Consolidation
        if self.heart_counter % self.failover.current_pulses_per_sweep == 0:
            result["pulse_triggered"] = True
            result["pulse_report"] = self.step_pulse_beat()

        return result

    def step_pulse_beat(self) -> Dict[str, Any]:
        """
        Every 1m15s (75s):
          1. Collects all but 1 heart from heart_side.log into pulse_pool.jsonl.
          2. Deletes latest pulses from pulse.log leaving only 1 PULSE followed by a BEAT.
          3. Updates 2 copies of the folder metadata pattern hash.
          4. On the 15th pulse, transports metadata pattern hash to event.log and checks events.
        """
        self.pulse_counter += 1
        ts = datetime.now(timezone.utc).isoformat()

        # 1. Collect all but 1 heart into the pulse pool
        hearts = self.heart_log.read_text(encoding="utf-8").splitlines() if self.heart_log.exists() else []
        pooled_count = 0
        if len(hearts) > 1:
            to_pool = hearts[:-1]
            kept_heart = hearts[-1]
            with open(self.pulse_pool, "a", encoding="utf-8") as pf:
                pf.write("\n".join(to_pool) + "\n")
            self.heart_log.write_text(kept_heart + "\n", encoding="utf-8")
            pooled_count = len(to_pool)

        # 2. Prune pulse.log leaving only 1 PULSE followed immediately by a BEAT
        pulse_entry = json.dumps({
            "kind": "pulse",
            "pulse_seq": self.pulse_counter,
            "pooled_hearts": pooled_count,
            "ts": ts
        })
        beat_entry = json.dumps({
            "kind": "beat",
            "pulse_seq": self.pulse_counter,
            "note": "cleaned all but 1 pulse followed by beat",
            "ts": ts
        })
        self.pulse_log.write_text(pulse_entry + "\n" + beat_entry + "\n", encoding="utf-8")

        # 3. Generate 2 rolling copies of the folder structure metadata pattern hash
        pattern_record = self.ledger.write_dual_copies()

        # 4. Check if this is the 15th pulse
        transported = False
        if self.pulse_counter % DEFAULT_TRANSPORT_PULSE_MODULO == 0:
            ok, verified_pattern = self.ledger.verify_dual_copies_match()
            if ok and verified_pattern:
                transport_event = {
                    "kind": "15th_pulse_metadata_transport",
                    "pulse_seq": self.pulse_counter,
                    "pattern_hash": verified_pattern["pattern_hash"],
                    "file_count": verified_pattern["file_count"],
                    "ts": ts
                }
                with open(self.event_log, "a", encoding="utf-8") as ef:
                    ef.write(json.dumps(transport_event) + "\n")
                transported = True

        # 5. Inspect event.log for any non-internal events to open WebSocket & AutoSync
        external_events = self._check_external_events()
        if external_events:
            self.websocket_open = True
            self.remote_Autosync_events.extend(external_events)

        return {
            "pulse_seq": self.pulse_counter,
            "pooled_hearts": pooled_count,
            "pattern_hash": pattern_record["pattern_hash"],
            "transported_on_15th": transported,
            "websocket_open": self.websocket_open
        }

    def _check_external_events(self) -> List[Dict[str, Any]]:
        if not self.event_log.exists():
            return []
        external = []
        for line in self.event_log.read_text(encoding="utf-8").splitlines():
            if not line.strip():
                continue
            try:
                rec = json.loads(line)
                if rec.get("kind") not in {"pulse", "beat"}:
                    external.append(rec)
            except Exception:
                continue
        return external

    def on_folder_file_change(self, changed_path: Path, remote_scheduled_flag: bool = True) -> Dict[str, Any]:
        """
        Folder monitor handler: ignores its own pulse-beat cycling logs,
        evaluates parallel/shorter-filename processes, and routes panic flags if needed.
        """
        if changed_path.name in DualCopyMetadataLedger.IGNORED_NAMES:
            return {"status": "IGNORED_OWN_PULSE_BEAT_LOG"}

        # Open WebSocket immediately upon seeing an external/parallel file deploy
        self.websocket_open = True
        eval_res = self.assimilator.evaluate_candidate(changed_path)

        # Log event to event.log
        evt = {
            "kind": "external_folder_event",
            "file": changed_path.name,
            "evaluation": eval_res,
            "remote_scheduled_flag": remote_scheduled_flag,
            "ts": datetime.now(timezone.utc).isoformat()
        }
        with open(self.event_log, "a", encoding="utf-8") as ef:
            ef.write(json.dumps(evt) + "\n")

        # Trigger panic failover if unscheduled alien process OR if upgrade distortion ejected
        if (not remote_scheduled_flag) or eval_res["action"] in {
            "EJECT_UPGRADE_AND_PANIC_SYNC",
            "ALIEN_PROCESS_DETECTED",
            "REJECT_INITIAL_ASSIMILATION"
        }:
            panic_flags = [
                f"PANIC:{eval_res['action']}",
                f"DISTORTION:{eval_res['distortion']}",
                f"REASON:{eval_res['reason']}"
            ]
            failover_res = self.failover.route_emergency_flags(str(changed_path), panic_flags)
            return {
                "evaluation": eval_res,
                "failover_resolution": failover_res
            }

        return {"evaluation": eval_res, "failover_resolution": None}


# ==============================================================================
# SECTION 7: DETERMINISTIC SELF-VERIFICATION & SIMULATION SUITE
# ==============================================================================
def run_verification_suite() -> None:
    import tempfile
    work_dir = Path(tempfile.mkdtemp(prefix="pax_autopoiet_verify_"))
    print(f"[INIT] Launching Ephemeral-Autopoietic Framer Verification in: {work_dir}")

    # Create canonical native script
    native_code = (
        "FOLDER_ID = '1MFUTmQvKR24bx8ClFIbOksywUPq2oN1M'\n"
        "SESSION_ID = 'ab272eee-855f-4825-a089-7400e2072b18'\n"
        "AXIAL_REGISTER_HEX = 0x78E8\n"
        "def core_loop(): return True\n"
        "def pulse_sync(): return True\n"
        "def verify_state(): return True\n"
        "def socket_dispatch(): return True\n"
    )
    (work_dir / CANONICAL_LONG_NAME).write_text(native_code, encoding="utf-8")

    daemon = InterloopedAutopoieticDaemon(work_dir, self_repair_complexity=True)
    daemon.ledger.snapshot_for_self_repair()

    # 1. Verify Multi-Runtime Ephemeral Decoupling (3 Runtimes: Termux, Colab-xterm, VM-Muse)
    for rt in ("termux-a16", "colab-xterm-t4", "vm-muse-linux"):
        daemon.decoupler.observe_runtime_allocation(rt)
        daemon.decoupler.observe_runtime_end_survived(rt, reason="ephemeral_wipe_survived")
    for rt in ("termux-a16", "colab-xterm-t4", "vm-muse-linux"):
        dec_state = daemon.decoupler.observe_runtime_allocation(rt)
    assert dec_state["decoupled"] is True, "Decoupling failed after 3-of-3 survival!"
    print(f"✔ [1/4] Ephemeral Decoupling Verified: decoupled={dec_state['decoupled']} across {list(dec_state['known_runtimes'].keys())}")

    # 2. Verify 15s Heart -> 1m15s (5 hearts) Pulse-Beat -> 15th Pulse Dual-Copy Metadata Transport
    # Run 75 heart ticks (= 15 pulses)
    last_pulse_rep = None
    for _ in range(75):
        h_res = daemon.step_heart()
        if h_res["pulse_triggered"]:
            last_pulse_rep = h_res["pulse_report"]

    assert last_pulse_rep is not None and last_pulse_rep["pulse_seq"] == 15
    assert last_pulse_rep["transported_on_15th"] is True
    pulse_lines = daemon.pulse_log.read_text(encoding="utf-8").splitlines()
    assert len(pulse_lines) == 2 and '"kind": "pulse"' in pulse_lines[0] and '"kind": "beat"' in pulse_lines[1]
    print(f"✔ [2/4] 15s Heart / 1m15s Pulse-Beat / 15th-Pulse Dual-Copy Hash Transport Verified (Hash: {last_pulse_rep['pattern_hash'][:16]}...)")

    # 3. Verify Shorter-Filename Willful Assimilation & Optimization Upgrade
    shorter_script = work_dir / "pax_ephemeral_opt.py"
    shorter_script.write_text(native_code, encoding="utf-8")
    res_assim = daemon.on_folder_file_change(shorter_script, remote_scheduled_flag=True)
    assert res_assim["evaluation"]["action"] == "WILLFUL_ASSIMILATION"

    # Apply valid efficiency optimization within tolerance (adds 1 helper function -> distortion = 1/5 = 0.20 <= 0.25)
    shorter_script.write_text(native_code + "def fast_cache_opt(): return 1\n", encoding="utf-8")
    res_opt = daemon.on_folder_file_change(shorter_script, remote_scheduled_flag=True)
    assert res_opt["evaluation"]["action"] == "ASSIMILATE_OPTIMIZATION_UPGRADE"
    print(f"✔ [3/4] Shorter-Filename Willful Assimilation & In-Flight Optimization Verified (Distortion: {res_opt['evaluation']['distortion']})")

    # 4. Verify Identity Distortion Ejection -> Endpoint 1 Overwhelm -> Endpoint 2 Loopback Rebuild & Self-Repair
    distorted_code = (
        "FOLDER_ID = '1MFUTmQvKR24bx8ClFIbOksywUPq2oN1M'\n"
        "SESSION_ID = 'ab272eee-855f-4825-a089-7400e2072b18'\n"
        "AXIAL_REGISTER_HEX = 0x78E8\n"
        "def mutated_fn_1(): pass\n"
        "def mutated_fn_2(): pass\n"
        "def mutated_fn_3(): pass\n"
    )
    shorter_script.write_text(distorted_code, encoding="utf-8")
    # Also simulate corruption of native script to prove self-repair from local ledger
    (work_dir / CANONICAL_LONG_NAME).write_text("# corrupted", encoding="utf-8")

    res_eject = daemon.on_folder_file_change(shorter_script, remote_scheduled_flag=True)
    assert res_eject["evaluation"]["action"] == "EJECT_UPGRADE_AND_PANIC_SYNC"
    f_res = res_eject["failover_resolution"]
    assert f_res["status"] == "LOOPBACK_REBUILT_AND_SELF_REPAIRED"
    assert CANONICAL_LONG_NAME in f_res["repaired_files"]
    print(
        f"✔ [4/4] Distortion Ejection -> Endpoint-1 Overwhelm -> Endpoint-2 Loopback Rebuild "
        f"(Socket Gen {f_res['new_socket_gen']}, Shifted Pulse: {f_res['new_heart_interval_s']}s/"
        f"{f_res['new_pulse_sweep_s']}s) & Local Ledger Self-Repair Verified!"
    )

    shutil.rmtree(work_dir, ignore_errors=True)


if __name__ == "__main__":
    run_verification_suite()
````
