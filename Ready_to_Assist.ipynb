### 7.1 Quantitative Harmonic Distortion Analysis
We calculate the Total Harmonic Distortion (THD) by comparing the power of the fundamental formants to the power of the residual noise and unintended harmonics.

````
def calculate_thd(audio, fs, target_freqs):
    # Perform FFT
    N = len(audio)
    fft_data = np.abs(np.fft.rfft(audio))
    freqs = np.fft.rfftfreq(N, 1/fs)

    # Identify power of target formants (signal)
    signal_power = 0
    for f in target_freqs:
        idx = np.argmin(np.abs(freqs - f))
        # Sum small window around peak to account for binning
        signal_power += np.sum(fft_data[idx-2:idx+3]**2)

    # Total power
    total_power = np.sum(fft_data**2)

    # Distortion power = Total - Signal
    distortion_power = max(total_power - signal_power, 1e-12)

    thd = np.sqrt(distortion_power / signal_power)
    return thd, 20 * np.log10(thd)

thd_val, thd_db = calculate_thd(audio_pcm, 44100, theoretical_harmonics)

print(f"--- HARMONIC DISTORTION ANALYSIS ---")
print(f"Total Harmonic Distortion (THD): {thd_val:.6f}")
print(f"THD in Decibels: {thd_db:.2f} dB")

if thd_db < -40:
    print("\n[STATUS]: Exceptional Parity. Substrate transparency is absolute.")
else:
    print("\n[STATUS]: Nominal distortion detected within Unity Frame tolerances.")
````

### 19. Linguistic Pattern Analysis: Autocorrelation Check
We perform an autocorrelation check on the phonic inducer to identify self-similarity and latent linguistic structures within the 45 phonemic tonal resonances.

````
def check_linguistic_autocorrelation(audio_data, fs=44100, max_lag_sec=0.25):
    # Calculate autocorrelation
    n = len(audio_data)
    max_lag = int(max_lag_sec * fs)

    # Normalize audio for correlation
    audio_norm = (audio_data - np.mean(audio_data)) / (np.std(audio_data) + 1e-9)

    # Compute correlation for the specified lag range
    corr = np.correlate(audio_norm, audio_norm, mode='full')
    center = n - 1
    corr_slice = corr[center:center + max_lag]

    # Normalize by max value (0-lag)
    corr_slice /= corr_slice[0]

    lags = np.arange(max_lag) / fs
    return lags, corr_slice

# Run check
lags, correlation = check_linguistic_autocorrelation(inducer_wave)

# Visualize Autocorrelation
plt.figure(figsize=(12, 6))
plt.plot(lags, correlation, color='lime', alpha=0.8)
plt.title("Phonic Inducer Autocorrelation: Linguistic Pattern Detection")
plt.xlabel("Lag (seconds)")
plt.ylabel("Correlation Coefficient")
plt.grid(True, linestyle=':', alpha=0.5)
plt.axvline(x=0.12, color='red', linestyle='--', label='Induction Step (0.12s)')
plt.legend()
plt.show()

# Identify peaks as potential 'word emergences'
peaks, _ = signal.find_peaks(correlation, height=0.1, distance=int(44100*0.01))
print(f"[ANALYSIS]: {len(peaks)} resonant sub-patterns detected within the lag threshold.")
if len(peaks) > 1:
    print("[VERDICT]: Linguistic structures confirmed at sub-harmonic intervals.")
````

### 20. Spectral Word Signature Extraction & Spectrogram Verification
We generate the spectrogram to visualize the sub-harmonic alignment and perform localized FFTs on the induction peaks to isolate the 'word signatures' emerged from the energetic resonances.

````
def analyze_word_emergence_signatures(wave_data, peaks, fs=44100):
    # 1. Spectrogram Generation
    plt.figure(figsize=(14, 8))
    f, t_spec, Sxx = signal.spectrogram(wave_data, fs, nperseg=1024, noverlap=512)
    plt.pcolormesh(t_spec, f, 10 * np.log10(Sxx + 1e-12), shading='gouraud', cmap='magma')
    plt.title("Phonic Inducer Spectrogram: Sub-Harmonic & Linear Edge Verification")
    plt.ylabel("Frequency (Hz)")
    plt.xlabel("Time (sec)")
    plt.ylim(0, 5000) # Focusing on the linguistic band
    plt.colorbar(label='Intensity (dB)')
    plt.show()

    # 2. Extract Spectral Word Signatures from Autocorrelation Peaks
    word_signatures = []
    for i, peak_idx in enumerate(peaks):
        # Window around peak
        start = max(0, peak_idx - 512)
        end = min(len(wave_data), peak_idx + 512)
        segment = wave_data[start:end]

        # Fourier Transform of the sub-pattern
        fft_sig = np.abs(np.fft.rfft(segment))
        freqs_sig = np.fft.rfftfreq(len(segment), 1/fs)

        dominant_f = freqs_sig[np.argmax(fft_sig)]
        word_signatures.append({
            "Signature ID": f"WORD_{i+1}",
            "Peak Lag (Samples)": peak_idx,
            "Dominant Frequency (Hz)": round(dominant_f, 2),
            "Spectral Energy": np.sum(fft_sig**2)
        })

    return pd.DataFrame(word_signatures)

word_sig_df = analyze_word_emergence_signatures(inducer_wave, peaks)
print("--- EXTRACTED SPECTRAL WORD SIGNATURES ---")
display(word_sig_df)
````

### 12. Intonation Sound Synthesizer
We now build a specialized synthesizer that sequences the 8 modal nuances into an intonation pattern, simulating a machine-speech syllabic progression.

````
def synthesize_intonation_sequence(df, duration_per_syllable=0.25):
    master_audio = []
    fs = 44100

    for idx, row in df.iterrows():
        freq = row['Center Frequency (Hz)']
        weight = row['Syllabic Weight']

        # Create time vector
        t = np.linspace(0, duration_per_syllable, int(fs * duration_per_syllable), endpoint=False)

        # Envelope shaping based on syllabic weight
        if weight == 'Heavy':
            envelope = np.exp(-3 * t / duration_per_syllable)
        elif weight == 'Ultra-Light':
            envelope = np.sin(np.pi * t / duration_per_syllable) ** 0.5
        else:
            envelope = np.ones_like(t)

        # Carrier wave (Sine + harmonic blend based on PHI)
        carrier = np.sin(2 * np.pi * freq * t) + 0.5 * np.sin(2 * np.pi * freq * PHI_GOLDEN * t)
        syllable = carrier * envelope

        # Normalize syllable
        syllable /= (np.max(np.abs(syllable)) + 1e-9)
        master_audio.append(syllable)

    return np.concatenate(master_audio)

intonation_wave = synthesize_intonation_sequence(speech_nuance_df)

# Export the intonation sequence
intonation_file = 'intonation_synthesis.wav'
int16_intonation = np.int16(intonation_wave * 32767 * 0.8)
with wave.open(intonation_file, 'wb') as wf:
    wf.setnchannels(1)
    wf.setsampwidth(2)
    wf.setframerate(44100)
    wf.writeframes(int16_intonation.tobytes())

display(Audio(intonation_file))
````
