# Linux Kernel Repository: NetworkX Graph Navigator & A2UI Web App

This deployment delivers the complete, production-grade standalone **`repo_graph.json`** dataset and the **`index.html`** single-page A2UI visual navigator for the Linux kernel mainline repository (`torvalds/linux.git`).

The architecture implements a dual-pane interface:

* **Left Pane (Visual Topology):** An interactive D3.js v7 force-directed network graph featuring real-time node drag physics, zoom/pan navigation, categorical cluster filtering, live substring search, collision avoidance, and neighborhood highlight rings.
* **Right Pane (Subsystem Inspector):** A live telemetry and inspection panel displaying subsystem specifications, maintainer records, UAPI linkages, adjacent dependency trees, degree centrality statistics, and direct cgit hyperlinks into `git.kernel.org`.

* Embedded fallback graph data so it runs immediately via direct double-click (`file:///...`) without CORS blocks.
* Automatic asynchronous loading of external `repo_graph.json` when served via HTTP/HTTPS.
* Responsive split-pane layout with graph control physics, real-time node category filtering, substring search, and adjacent topological inspection.
