## The two terminals, the midspace, and the name relations

Termux on the phone and Colab here (under xterm) are two ephemeral Linux
terminals running the *same* runtime with the *same* addressing. What they
share is the **midspace** — the middle space between the two terminals,
where they meet: the shared Drive folder (`/APlaceForPax/`), and in it the
navigation state file `state/nav-state.json`. Either terminal moves, the
state file records where it went; the other terminal reads the same file
and resumes at the same place.

**The name relations** (the user's framing, not a technical claim):
Termux = term + ux; xterm = x + term. The shared "term" is the terminal.
The directions congrue — the name relations ARE the directions:
`([term)[(ux)](term)]` — `[term)` is the terminal opening, `[(ux)]` is
the ux in brackets, `(term)` is the terminal in parentheses. The
terminals share "term"; they differ in the prefix/suffix (x vs ux).
That is how the names point the way between them.

**The VPN** (Tailscale, the tailnet) is how the terminals reach each
other. The VM daemon is the hub: it reaches the phone's device endpoint
through the runtime tunnel proxy (port 3130). Colab is NOT on the
tailnet, so the outbound device link from Colab is left unset; the VM
daemon is the hub for the phone link.

The user's framing of the VPN pathway (recorded as his): the VPN pathway
connects through his GitHub login back to him through his Google account,

## Launch + cleanup — the rollover (one chance in the ephemeral runtime)

This notebook is a **single .ipynb with a new separate file name** —
`autoruntime-colab-autopoet-nav-20261106-083425.ipynb` — and every contained file is named separately with
the same stamp `20261106-083425`:

- the navigation module: `autopoet_nav-20261106-083425.py`
- graphify: `graphify-20261106-083425.py`

The names changed because the Colab runtime is ephemeral, like the
Termux instance: there is only **one chance to establish the construct**
in an ephemeral runtime, and a new construct needs new names — the same
as the Termux rollover, where the phone-side files changed names
(`pax_device_link-20261106-063922.py` and the rest) so the new process
could take over from the old one. The former names are:

- former notebook: `autoruntime-colab-autopoet-nav.ipynb`
- former module: `autopoet_nav.py`
- former graphify: `graphify.py`

The code cell below does **both** — launch and cleanup — in this order:

1. **Deploy SHA.** It hashes this notebook's own `.ipynb` file when the
   file can be found (working dir, Drive mount, or the clone), and
   records it as the deploy SHA. If the file cannot be found, the deploy
   SHA is recorded as `unavailable` and the stamp is the identity — the
   cell says so plainly.
2. **Marker first.** It writes `state/rollover.json` (shared Drive folder
   and local mirror) with the stamp, the former name, the new name, the
   deploy SHA, and status `in-progress` — BEFORE launching or deleting
   anything. **Idempotency:** if the marker already says `complete` with
   this same stamp, the cell is a no-op (nothing re-launched, nothing
   re-deleted). If the marker says `in-progress` or `failed`, it resumes
   with the SAME stamp — no new stamp is generated on resume. (The
   stamp, not the SHA, gates the no-op: Colab re-saves the notebook with
   outputs as it runs, so the runtime SHA of the `.ipynb` differs from
   the build-time SHA by design; the stamp is the deployment identity,
   and the deploy SHA is recorded for audit.)
3. **Pre-flight.** It parse-checks (`ast.parse`) and SHA-checks the two
   stamped files in the working dir before anything is launched or
   deleted. If any check fails, it **stops before launching or cleaning
   up anything** and says so plainly.
4. **Launch.** It confirms the new stamped files are in place and
   importable, and prints that this notebook runs from the new stamped
   construct. The navigation, endpoints, congrues, and graph cells below
   then run from it.
5. **Cleanup.** It deletes **all prior files of the former name** from
   BOTH places they can live: the Colab runtime working directory and
   the shared Drive folder (top level of APlaceForPax, `state/`, and the
   `/Main` subfolders, bounded walk). The rule is the Termux rollover's
   rule: delete every match for each former root + `*` + extension —
   `autoruntime-colab-autopoet-nav*.ipynb`, `autopoet_nav*.py`,
   `graphify*.py` — **EXCEPT** the new stamped files, and **NEVER** the
   running notebook itself. (The rollover cell exists only in the new
   stamped notebook, so the running notebook is always the new stamped
   name, which is excluded by name.) The git clone (`./AutoPoET`) is a
   git working copy and is **not** touched — the repo is managed by git,
   and a former-name notebook file inside the repo is a historical
   artifact, not a runtime file.
6. **Carried over, not deleted.** The state files (`nav-state.json`,
   `nav-state-drive.json`, `nav-state-smartrepo.json`, `endpoints.json`,
   `congrue.json`, `congrue-latest.json`, `final-state.json`,
   `final-state-latest.json`) are carried over: they are rewritten with
   the new stamped names and the new stamp inside them where they
   reference file names, the same way the Termux rollover rewrites

   former roots.


```python
import ast, glob, hashlib, json, os
from datetime import datetime, timezone
# ---- Launch + cleanup: the rollover (Colab equivalent of the Termux rollover) ----
# One chance to establish this construct in the ephemeral runtime: the
# names changed (new stamped notebook, new stamped contained files),
# so this cell launches from the NEW names and cleans up ALL prior
# files of the FORMER name — the same rule as the Termux rollover.
STAMP = '20261106-083425'
FORMER_NOTEBOOK = 'autoruntime-colab-autopoet-nav.ipynb'
NEW_NOTEBOOK = 'autoruntime-colab-autopoet-nav-20261106-083425.ipynb'
NAV_STAMPED = 'autopoet_nav-20261106-083425.py'
GRAPHIFY_STAMPED = 'graphify-20261106-083425.py'
FORMER_PATTERNS = [  # (root, extension) — delete root + '*' + extension
    ('autoruntime-colab-autopoet-nav', '.ipynb'),
    ('autopoet_nav', '.py'),
    ('graphify', '.py'),
]
KEEP_NAMES = {NEW_NOTEBOOK, NAV_STAMPED, GRAPHIFY_STAMPED}  # never delete the new files
# State files are CARRIED OVER (rewritten with the new names/stamp), never deleted.
CARRYOVER_FILES = ['nav-state.json', 'nav-state-drive.json', 'nav-state-smartrepo.json',
                   'endpoints.json', 'congrue.json', 'congrue-latest.json',
                   'final-state.json', 'final-state-latest.json']
MARKER_NAME = 'rollover.json'
# --- deploy SHA: hash this notebook's own .ipynb when it can be found ---
def _find_self():
    for _d in [os.getcwd(), NESTING_1, NESTING_2]:
        if not _d:
            continue
        _p = os.path.join(_d, NEW_NOTEBOOK)
        if os.path.isfile(_p):
            return _p
    for _base in [NESTING_1, NESTING_2]:
        if not _base:
            continue
        for _root, _dirs, _files in os.walk(_base):
            _dirs[:] = [d for d in _dirs if not d.startswith('.')]
            if NEW_NOTEBOOK in _files:
                return os.path.join(_root, NEW_NOTEBOOK)
    return None
_self_path = _find_self()
if _self_path:
    DEPLOY_SHA = hashlib.sha256(open(_self_path, 'rb').read()).hexdigest()
    print('deploy SHA (this notebook file, found at %s): %s' % (_self_path, DEPLOY_SHA))
else:
    DEPLOY_SHA = 'unavailable (notebook file not readable at runtime)'
    print('deploy SHA: unavailable — the notebook file was not found in the working dir or the Drive mount (idempotency uses the stamp)')
# --- marker: read first; idempotent no-op when complete with this stamp ---
MARKER_SHARED = os.path.join(SHARED_DIR, MARKER_NAME)
MARKER_LOCAL = os.path.join(LOCAL_STATE_DIR, MARKER_NAME)
_marker = None
_marker_path = None
for _mp in [MARKER_SHARED, MARKER_LOCAL]:
    if os.path.isfile(_mp):
        try:
            _marker = json.load(open(_mp))
            _marker_path = _mp
            break
        except Exception:
            _marker = None
if _marker and _marker.get('status') == 'complete' and _marker.get('stamp') == STAMP:
    print('ROLLOVER ALREADY COMPLETE for stamp %s (marker %s) — no-op: nothing re-launched, nothing re-deleted.' % (STAMP, _marker_path))
    print('  marker deploy SHA:', _marker.get('deploy_sha'), '| this run deploy SHA:', DEPLOY_SHA)
    print('  (the deploy SHA may differ from the build-time SHA because Colab re-saves the notebook with outputs as it runs; the stamp is the deployment identity)')
    ROLLOVER_DONE = True
else:
    if _marker:
        print('resuming rollover with the SAME stamp %s (marker status: %s) — no new stamp is generated.' % (STAMP, _marker.get('status')))
    ROLLOVER_DONE = False
    # --- pre-flight: parse + SHA-check the stamped files BEFORE anything is launched or deleted ---
    _pre_ok = True
    for _fname, _expected_sha in [(NAV_STAMPED, _EXPECTED_NAV_SHA), (GRAPHIFY_STAMPED, GRAPHIFY_SHA256)]:
        _fpath = os.path.join(os.getcwd(), _fname)
        if not os.path.isfile(_fpath):
            print('PRE-FLIGHT FAIL: %s not present in the working directory' % _fname)
            _pre_ok = False
            continue
        _fb = open(_fpath, 'rb').read()
        _fsha = hashlib.sha256(_fb).hexdigest()
        if _fsha != _expected_sha:
            print('PRE-FLIGHT FAIL: %s SHA mismatch (got %s, expected %s)' % (_fname, _fsha, _expected_sha))
            _pre_ok = False
            continue
        try:
            ast.parse(_fb.decode('utf-8'))
            print('pre-flight: %s parses OK, SHA match' % _fname)
        except Exception as _e:
            print('PRE-FLIGHT FAIL: %s does not parse: %r' % (_fname, _e))
            _pre_ok = False
    if not _pre_ok:
        print('ROLLOVER STOPPED — pre-flight failed. Nothing was launched, nothing was deleted, no marker was written.')
        raise RuntimeError('rollover pre-flight failed')
    # --- marker (in-progress) BEFORE the launch-and-cleanup ---
    _marker_rec = {'stamp': STAMP, 'former_notebook': FORMER_NOTEBOOK, 'new_notebook': NEW_NOTEBOOK,
                   'former_files': ['autopoet_nav.py', 'graphify.py'],
                   'new_files': [NAV_STAMPED, GRAPHIFY_STAMPED],
                   'deploy_sha': DEPLOY_SHA, 'status': 'in-progress', 'writer': 'colab-xterm',
                   'updated_utc': datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ'),
                   'note': 'rollover: launch from the new stamped names, cleanup all prior files of the former name'}
    os.makedirs(SHARED_DIR, exist_ok=True)
    os.makedirs(LOCAL_STATE_DIR, exist_ok=True)
    json.dump(_marker_rec, open(MARKER_SHARED, 'w'), indent=2)
    json.dump(_marker_rec, open(MARKER_LOCAL, 'w'), indent=2)
    print('marker written (in-progress):', MARKER_SHARED, '| local mirror:', MARKER_LOCAL)
    # --- launch: confirm the new stamped files are in place and importable ---
    print('launch: new stamped files present — %s, %s (module stamp %s, graphify SHA match %s)' % (NAV_STAMPED, GRAPHIFY_STAMPED, nav.VERSION_STAMP, GRAPHIFY_SHA256 == nav.GRAPHIFY_SHA256))
    print('launch: this notebook (%s) runs from the new stamped construct; the navigation, endpoints, congrues, and graph cells below use it.' % NEW_NOTEBOOK)
    # --- cleanup: delete ALL prior files of the former name, both places ---
    # Rule (Termux rollover): delete every match for each former root + '*' +
    # extension, EXCEPT the new stamped files, and NEVER the running notebook
    # itself. State files are carried over, never deleted. The clone
    # (./AutoPoET) is a git working copy and is NOT touched.
    def _walk_files(_base, _max_depth):
        _found = []
        if not _base or not os.path.isdir(_base):
            return _found
        for _root, _dirs, _files in os.walk(_base):
            _rel = os.path.relpath(_root, _base)
            _depth = 0 if _rel == '.' else _rel.count(os.sep) + 1
            if _depth >= _max_depth:
                _dirs[:] = []
            _dirs[:] = [d for d in _dirs if not d.startswith('.') and d != 'node_modules']
            for _f in _files:
                _found.append(os.path.join(_root, _f))
        return _found
    def _top_files(_d):
        if not _d or not os.path.isdir(_d):
            return []
        return [os.path.join(_d, _f) for _f in os.listdir(_d) if os.path.isfile(os.path.join(_d, _f))]
    _all_files = []
    _all_files += _top_files(os.getcwd())
    _all_files += _top_files(os.path.join(os.getcwd(), 'state'))
    _all_files += _walk_files(NESTING_1, 2)
    _all_files += _walk_files(NESTING_2, 3)
    _deleted, _kept = [], []
    for _p in sorted(set(_all_files)):
        _base = os.path.basename(_p)
        _is_former = any(_base.startswith(_root) and _base.endswith(_ext) for _root, _ext in FORMER_PATTERNS)
        if not _is_former:
            continue
        if _base in KEEP_NAMES:
            _kept.append(_p)
            continue
        # Never delete a state file or the marker (they do not match the
        # former patterns, but guard anyway) and never delete a file
        # inside the clone.
        if _base in CARRYOVER_FILES or _base == MARKER_NAME:
            _kept.append(_p)
            continue
        if (os.path.sep + 'AutoPoET' + os.path.sep) in _p:
            _kept.append(_p)
            continue
        os.remove(_p)
        _deleted.append(_p)
    print('cleanup: deleted %d former-name file(s):' % len(_deleted))
    for _p in _deleted:
        print('  deleted:', _p)
    print('cleanup: kept %d matching new/carried-over file(s) (not deleted):' % len(_kept))
    for _p in _kept:
        print('  kept:', _p)
    # --- carry over the state files: rewrite the new names/stamp inside, never delete ---
    def _carry(_path):
        if not os.path.isfile(_path):
            return None
        try:
            _rec = json.load(open(_path))
        except Exception as _e:
            print('carry-over: %s could not be read as JSON (%r) — left as-is' % (_path, _e))
            return None
        if not isinstance(_rec, dict):
            return None
        _changed = []
        if 'version_stamp' in _rec and _rec['version_stamp'] != STAMP:
            _rec['version_stamp'] = STAMP
            _changed.append('version_stamp')
        if isinstance(_rec.get('nav_module'), dict) and _rec['nav_module'].get('file') == 'autopoet_nav.py':
            _rec['nav_module']['file'] = NAV_STAMPED
            _changed.append('nav_module.file')
        if isinstance(_rec.get('graphify'), dict) and _rec['graphify'].get('file') == 'graphify.py':
            _rec['graphify']['file'] = GRAPHIFY_STAMPED
            _changed.append('graphify.file')
        if isinstance(_rec.get('notebook_in_repo'), str) and FORMER_NOTEBOOK in _rec['notebook_in_repo']:
            _rec['notebook_in_repo'] = _rec['notebook_in_repo'].replace(FORMER_NOTEBOOK, NEW_NOTEBOOK)
            _changed.append('notebook_in_repo')
        if _changed:
            _tmp = _path + '.tmp'
            json.dump(_rec, open(_tmp, 'w'), indent=2, sort_keys=True)
            os.replace(_tmp, _path)
            print('carry-over: %s rewritten (%s)' % (_path, ', '.join(_changed)))
        else:
            print('carry-over: %s already current (nothing to rewrite)' % _path)
        return _path
    _carry_paths = []
    for _d in [SHARED_DIR, LOCAL_STATE_DIR]:
        for _fn in CARRYOVER_FILES:
            _carry_paths.append(os.path.join(_d, _fn))
    if NESTING_1:
        for _fn in ['congrue-latest.json', 'final-state-latest.json']:
            _carry_paths.append(os.path.join(NESTING_1, _fn))
    for _p in _carry_paths:
        _carry(_p)
    # --- marker (complete) ---
    _marker_rec['status'] = 'complete'
    _marker_rec['updated_utc'] = datetime.now(timezone.utc).strftime('%Y-%m-%dT%H:%M:%SZ')
    _marker_rec['deleted'] = _deleted
    _marker_rec['kept'] = _kept
    json.dump(_marker_rec, open(MARKER_SHARED, 'w'), indent=2)
    json.dump(_marker_rec, open(MARKER_LOCAL, 'w'), indent=2)
    print('marker updated (complete):', MARKER_SHARED)
    print('ROLLOVER COMPLETE for stamp %s: launched from %s + %s, deleted %d former-name file(s), carried over the state files.' % (STAMP, NAV_STAMPED, GRAPHIFY_STAMPED, len(_deleted)))
    ROLLOVER_DONE = True
```

## Navigate: the commands and the midspace state file

The navigation works on the local clone, through the `autopoet_nav`
functions:

- `list_dir` — list the current folder (dirs first, then files).
- `show_file` — show a file's contents (bounded: 80 lines, 64 KB; binary
  and oversize files are reported, not dumped).
- `enter(name)` — move into a subfolder. `up()` — move to the parent.
  `to_root()` — back to the root. `goto(rel)` — jump to a path inside
  the repo. `tree(depth=2)` — a bounded tree listing. `find(pattern)` —
  find files by name.
- No path ever escapes the repo root: `..` and absolute paths are
  refused.

After every move the state is written to the midspace — the shared
state file `state/nav-state.json` in the shared Drive folder — and to a
local mirror (`./state/nav-state.json`). The state file holds the current
path (relative to the repo root), the path taken, the repo head, and
which terminal wrote it last. The other terminal reads the same file and
resumes at the same place. That is how the two terminals navigate
together, and how navigation survives an ephemeral restart: the state
lives in the shared Drive folder, not in the runtime's memory.

## Addressings — one runtime, one addressing, and the ephemeral constructs

Every addressing the navigation uses, written to `state/endpoints.json`,
including the Drive mount paths, both nesting levels, the repo, the
midspace state file, and the VPN.

**The ephemeral constructs.** The Colab runtime is ephemeral; the VM is
ephemeral; the phone Termux instance is ephemeral. What survives a
runtime restart:

- The navigation state file (`state/nav-state.json` in the shared Drive
  folder) — the current path, the path taken, the repo head, the writer.
  A fresh runtime resumes from it (`nav.resume`), so navigation picks up
  where it left off.
- The endpoints file (`state/endpoints.json`), the local clone
  (`./AutoPoET`), and the logs — all files on disk, not in memory.

What is lost when the runtime restarts:

- The running processes (there are none started by this notebook — it
  starts no socket, so there is no port conflict with the runtime
  notebook) and the in-memory navigation object (`STATE`). The notebook
  must be re-run (cells in order) to rebuild them; the state file is
  what makes the rebuild land in the same place.

**The VPN addressings** (Tailscale, the tailnet):

- VM: IPv4 `100.92.198.68`, IPv6 `fd7a:115c:a1e0::4848:c645`.
- Phone: `christophers-a16.tail269cfa.ts.net`, IPv4 `100.107.226.36`,
  IPv6 `fd7a:115c:a1e0::3c01:e23a`.
- Device endpoint `ws://100.107.226.36:18765`; MCP redirect port 18766.
- Ordinary networking does NOT reach the tailnet. Tailnet TCP goes
  through the runtime tunnel proxy with only its port changed to 3130
  (the egress proxy itself is on 3128). The tunnel proxy URL is derived
  at runtime from HTTPS_PROXY (port changed to 3130) — never hardcoded,
  never written to a file, never logged, because it may carry
  credentials. The tunnel proxy itself does not require credentials.
  A host needs the tunnel proxy when it is CGNAT 100.64/10, RFC1918, or
  `.ts.net`; loopback and public addresses connect directly (proxy=None —
  the egress proxy never carries local traffic). The Tailscale client
  (`/opt/hatch/bin/tailscale`) is client-only: outgoing TCP only, no
  inbound.
- Colab is NOT on the tailnet: `PAX_DEVICE_ENDPOINT_URL` is left unset
