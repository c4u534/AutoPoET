````
import matplotlib.pyplot as plt
import numpy as np
import math
import hashlib
import networkx as nx
from dataclasses import dataclass, field, asdict
from typing import Dict, List, Tuple, Any

# 1. NOTEBOOK CALIBRATION CONSTANTS (ARGON GYRO INVARIANTS)
NOTEBOOK_METRICS = {
    "G0_CONGRUENCE": 0.84210000,
    "DELTA_MU_M": 5.5 - math.sqrt(10.0),
    "DELTA_VAR_GALOIS": 0.83772234,
    "THETA_LEAN_RAD": 0.01501,
    "HAMMER_THRESHOLD": 0.46,
    "CHERN_C1": 1.0,
    "SHADOW_OFFSET_RATIO": 8.0,
    "MERSENNE_PRIMES": (3, 7, 31, 127),
    "CROWN_TRACE": 15.0,
    "KB_T_LN2_NORM": 1e-4,
    "MODULARITY_BENCHMARK": 0.975256,
    "SVD_DRIFT_BOUND": 1e-12,
}

# 2. COMPLETE 113-ENTITY REGISTRY
ACTIVATED_ENTITIES_RAW = [
    ("adaptive-motive-skillset-generator", "skill", "motive_packaging", "icosahedron", 1, 0.35, 1.15),
    ("agentic-governance-supervisor-qa", "skill", "governance_qa", "cube", 0, -0.25, 1.00),
    ("aletheia-nami-substrate", "skill", "substrate_kernel", "torus", 1, -0.40, 0.92),
    ("adaptive-tasktree-swarm-orchestrator", "skill", "swarm_orchestration", "octahedron", 1, 0.45, 1.08),
    ("aletheia-maternal-ladder-monolith", "skill", "annealing_monolith", "pyramid", 1, -0.30, 0.88),
    ("and-skill-or", "skill", "logic_gate", "tetrahedron", 0, 0.05, 1.00),
    ("auto-spark-forrest-coding", "skill", "forest_coding", "dodecahedron", 1, 0.50, 1.12),
    ("autopoet-zerodrift-production-deployer", "skill", "production_deploy", "cube", 1, -0.15, 0.95),
    ("autopoietic-code-orchestrator", "skill", "code_synthesis", "octahedron", 1, 0.30, 1.04),
    ("autopoietic-modular-orchestrator", "skill", "swarm_orchestration", "dodecahedron", 1, 0.28, 1.05),
    ("backdoor-audit-compatibility-engineer", "skill", "governance_qa", "cube", -1, -0.45, 1.20),
    ("bounded-memory-swarm-orchestrator", "skill", "substrate_kernel", "torus", 0, -0.50, 0.85),
    ("chatlog-fullstack-assembler", "skill", "fullstack_assembly", "dodecahedron", 1, 0.40, 1.10),
    ("chiral-agency-orchestrator", "skill", "chiral_temporal", "helix", 1, 0.15, 0.98),
    ("co-colab", "skill", "colab_runtime", "icosahedron", 1, 0.32, 1.02),
    ("colab-adaptive-morphology-coder", "skill", "morphology_math", "icosahedron", 1, 0.25, 0.96),
    ("colab-construct-orchestrator", "skill", "colab_runtime", "octahedron", 1, 0.22, 1.03),
    ("colab-diagram-stack-synthesizer", "skill", "visual_canvas", "dodecahedron", 1, 0.38, 1.07),
    ("combinational-swarm-orchestrator", "skill", "swarm_orchestration", "icosahedron", 1, 0.42, 1.06),
    ("cross-constructive-morphology-engine", "skill", "morphology_math", "dodecahedron", 1, 0.35, 0.94),
    ("cross-hatched-modal-matrix-coder", "skill", "matrix_manifold", "cube", 0, -0.10, 0.91),
    ("decompositional-reverse-engineering", "skill", "reverse_deconstruct", "tetrahedron", -1, -0.55, 1.25),
    ("deep-research-project-orchestrator", "skill", "research_swarm", "octahedron", 1, 0.60, 1.14),
    ("deploy-spark", "skill", "spark_matrix", "cube", 1, 0.10, 0.90),
    ("deep-research-modal-visualizer", "skill", "visual_canvas", "icosahedron", 1, 0.48, 1.09),
    ("directional-motive-force-vector", "skill", "motive_packaging", "pyramid", 1, 0.55, 1.18),
    ("dual-image-colab-stack-synthesizer", "skill", "visual_canvas", "octahedron", 1, 0.30, 1.08),
    ("fastmcp-colab-bridge", "skill", "mcp_bridge", "torus", 1, 0.20, 0.97),
    ("firecrawl-fastmcp", "skill", "mcp_bridge", "pyramid", 1, 0.52, 1.11),
    ("forest-hive-swarm-substrate", "skill", "swarm_orchestration", "dodecahedron", 1, 0.24, 0.99),
    ("full-stack-application-deployer", "skill", "fullstack_assembly", "cube", 1, 0.36, 1.06),
    ("gem-opal-deployment-orchestrator", "skill", "gem_opal_meta", "octahedron", 1, 0.27, 1.04),
    ("gem-opal-skill-synthesizer", "skill", "gem_opal_meta", "icosahedron", 1, 0.44, 1.07),
    ("geomnami-qmv2-pantheon", "skill", "physics_logic", "torus", 1, -0.20, 0.89),
    ("geom", "skill", "physics_logic", "tetrahedron", 1, -0.15, 0.92),
    ("google-drive-indexer", "skill", "workspace_index", "cube", 0, -0.35, 1.01),
    ("indexd", "skill", "workspace_index", "cube", 0, -0.35, 1.01),
    ("inverted-whiteboard-canvas", "skill", "visual_canvas", "pyramid", -1, 0.18, 1.13),
    ("intelligence-canvas-framework", "skill", "visual_canvas", "dodecahedron", 1, 0.33, 1.02),
    ("literary-palette-swarm-orchestrator", "skill", "creative_palette", "icosahedron", 1, 0.65, 1.22),
    ("logi", "skill", "logic_gate", "tetrahedron", 0, -0.22, 0.96),
    ("machine-discretionary-means", "skill", "logic_gate", "octahedron", 1, 0.08, 0.95),
    ("magnitudinal-stem-swarm-synthesizer", "skill", "research_swarm", "dodecahedron", 1, 0.47, 1.05),
    ("mersenne-context-morphology-meter", "skill", "morphology_math", "torus", 0, -0.18, 0.87),
    ("mersenne-tensor-logic-crosscoder", "skill", "physics_logic", "helix", 1, 0.12, 0.90),
    ("morphogenetic-code-synthesizer", "skill", "code_synthesis", "icosahedron", 1, 0.58, 1.16),
    ("multi-observer-quantum-telemetry-orchestrator", "skill", "telemetry_quantum", "icosahedron", 0, 0.41, 1.09),
    ("multi-engine-swarm-orchestrator", "skill", "swarm_orchestration", "dodecahedron", 1, 0.39, 1.03),
    ("multi-tier-repo-cataloger", "skill", "workspace_index", "cube", 0, -0.28, 0.99),
    ("multi-tier-swarm-substrate", "skill", "swarm_orchestration", "octahedron", 1, 0.25, 0.98),
    ("multimodal-fullstack-deployer", "skill", "fullstack_assembly", "dodecahedron", 1, 0.43, 1.07),
    ("multiskill-arag-orchestrator", "skill", "rag_meta", "octahedron", 1, 0.31, 0.97),
    ("multiskill-morphological-crossbinder", "skill", "morphology_math", "dodecahedron", 1, 0.20, 0.91),
    ("natural-ethical-harmonics", "skill", "governance_qa", "sphere", 0, -0.12, 0.84),
    ("natural-forest-hive-substrate", "skill", "swarm_orchestration", "dodecahedron", 1, 0.22, 0.98),
    ("natural-temporal-deferment", "skill", "chiral_temporal", "sphere", -1, -0.60, 0.80),
    ("nested-matrix-isquared-manifold", "skill", "matrix_manifold", "torus", -1, -0.32, 0.93),
    ("nested-matrix-all-manifold", "skill", "matrix_manifold", "sphere", 0, 0.70, 0.86),
    ("nested-matrix-null-manifold", "skill", "matrix_manifold", "sphere", 0, -0.65, 0.82),
    ("nested-matrix-or-manifold", "skill", "matrix_manifold", "octahedron", 1, 0.26, 0.95),
    ("nested-matrix-of-matrices", "skill", "matrix_manifold", "cube", 0, 0.00, 0.88),
    ("nested-matrix-xor-manifold", "skill", "matrix_manifold", "tetrahedron", -1, 0.14, 0.94),
    ("nested-matrix-zero-manifold", "skill", "matrix_manifold", "torus", 0, -0.42, 0.85),
    ("networkx-airtable-canvas-mapper", "skill", "graph_nav", "octahedron", 1, 0.34, 1.04),
    ("networkx-repo-graph-navigator", "skill", "graph_nav", "octahedron", 1, 0.29, 1.02),
    ("orchestrator", "skill", "gem_opal_meta", "dodecahedron", 1, 0.16, 0.93),
    ("polyglot-deterministic-swarm-coder", "skill", "code_synthesis", "cube", 1, -0.20, 0.89),
    ("prompt-triad-synthesizer", "skill", "rag_meta", "pyramid", 0, 0.37, 1.06),
    ("pursuant-deferred-determination", "skill", "chiral_temporal", "sphere", -1, -0.52, 0.83),
    ("qmv2-physics-logic-coder", "skill", "physics_logic", "tetrahedron", 1, -0.14, 0.91),
    ("qmv2", "skill", "physics_logic", "tetrahedron", 1, -0.18, 0.90),
    ("rag-multiskill-prompt-initializer", "skill", "rag_meta", "icosahedron", 1, 0.46, 0.99),
    ("recursive-deep-context-swarm", "skill", "research_swarm", "helix", 1, 0.62, 1.17),
    ("recursive-rag-prompt-orchestrator", "skill", "rag_meta", "helix", 1, 0.28, 0.98),
    ("recursive-url-repo-mapper", "skill", "graph_nav", "helix", 1, 0.49, 1.10),
    ("rigorous-math-physics-derivator", "skill", "physics_logic", "tetrahedron", 0, -0.30, 0.87),
    ("royal-primality-matriarch-lineage", "skill", "annealing_monolith", "pyramid", 1, -0.08, 0.86),
    ("scholarly-skillset-architect", "skill", "motive_packaging", "cube", 0, -0.16, 0.95),
    ("semantic-scholar-orchestrator", "skill", "research_swarm", "octahedron", 1, 0.54, 1.08),
    ("session-checkpoint-persister", "skill", "workspace_index", "cube", 0, -0.48, 0.92),
    ("sim000-autopoietic-tensile-engine", "skill", "physics_logic", "torus", 1, 0.00, 0.84),
    ("skill-chain-morphology-analyzer", "skill", "morphology_math", "helix", 0, 0.11, 0.91),
    ("skill-helper-demo", "skill", "motive_packaging", "octahedron", 1, 0.25, 1.00),
    ("skill-outcome-projector", "skill", "motive_packaging", "pyramid", 1, 0.40, 1.05),
    ("skill-sequence-colab-compiler", "skill", "colab_runtime", "cube", 1, 0.19, 0.94),
    ("spark-documentconversation-coder", "skill", "fullstack_assembly", "octahedron", 1, 0.27, 1.03),
    ("spark-gemmultigem-opal", "skill", "gem_opal_meta", "icosahedron", 1, 0.38, 1.05),
    ("system-prompt-curator", "skill", "rag_meta", "cube", -1, -0.24, 0.96),
    ("swarm", "skill", "swarm_orchestration", "dodecahedron", 1, 0.35, 1.00),
    ("telemetry-graphing-sentinel", "skill", "telemetry_quantum", "octahedron", 0, 0.05, 0.93),
    ("topographical-chatlog-collator", "skill", "workspace_index", "cube", 0, -0.38, 0.97),
    ("transparency-emboss-engine", "skill", "visual_canvas", "pyramid", -1, 0.22, 1.12),
    ("tri-phase-skillset-packager", "skill", "motive_packaging", "pyramid", 1, 0.15, 0.94),
    ("tripartite-harmonic-divergence", "skill", "physics_logic", "helix", 0, 0.31, 0.96),
    ("tripartite-radial-research-swarm", "skill", "research_swarm", "icosahedron", 1, 0.56, 1.11),
    ("unified-autopoietic-swarm-substrate", "skill", "substrate_kernel", "dodecahedron", 1, 0.18, 0.92),
    ("universal-code-constructor", "skill", "code_synthesis", "cube", 1, 0.10, 0.90),
    ("universal-pseudo-ide-deployer", "skill", "colab_runtime", "cube", 1, 0.24, 0.99),
    ("universal-skill-actuator", "skill", "rag_meta", "sphere", 1, 0.30, 0.88),
    ("Canvs Mcp Server", "tool", "visual_canvas", "dodecahedron", 1, 0.42, 1.06),
    ("Firecrawl Fastmcp", "tool", "mcp_bridge", "pyramid", 1, 0.58, 1.13),
    ("Airtable", "tool", "workspace_index", "cube", 0, -0.22, 0.98),
    ("Gemini Notebook", "tool", "rag_meta", "sphere", 0, -0.15, 0.86),
    ("GitMCP", "tool", "mcp_bridge", "octahedron", 1, 0.18, 0.94),
    ("GitMCP-AllRepos", "tool", "mcp_bridge", "icosahedron", 1, 0.48, 1.05),
    ("Google Docs", "tool", "workspace_index", "cube", 0, -0.10, 0.97),
    ("Google Drive", "tool", "substrate_kernel", "cube", 0, -0.40, 0.90),
    ("Google Keep", "tool", "telemetry_quantum", "tetrahedron", 0, 0.12, 0.99),
    ("Google Sheets", "tool", "workspace_index", "cube", 0, -0.25, 0.96),
    ("Linear", "tool", "governance_qa", "pyramid", 1, -0.18, 0.95),
    ("Miro", "tool", "visual_canvas", "dodecahedron", 1, 0.45, 1.08),
    ("Sentry MCP", "tool", "telemetry_quantum", "tetrahedron", -1, -0.35, 0.91),
    ("Canva", "tool", "visual_canvas", "icosahedron", 1, 0.50, 1.14),
]

SHAPE_GEOMETRY = {
    "tetrahedron":  (4,  6,  4,  2, 0.7796, 0.1179),
    "cube":         (6,  12, 8,  2, 1.0000, 1.0000),
    "octahedron":   (8,  12, 6,  2, 0.9474, 0.4714),
    "pyramid":      (5,  8,  5,  2, 0.8200, 0.2357),
    "dodecahedron": (12, 30, 20, 2, 0.9045, 7.6631),
    "icosahedron":  (20, 30, 12, 2, 0.8364, 2.1817),
    "torus":        (16, 32, 16, 0, 0.8650, 1.9739),
    "helix":        (13, 26, 13, 0, 0.8120, 1.3000),
    "sphere":       (32, 64, 34, 2, 0.7405, 4.1888),
}

PHASE_BETA = {"solid": 0.15, "fluid": 0.45, "plasma": 0.95, "null-stasis": 0.02}

@dataclass
class EntityProfile:
    idx: int
    name: str
    kind: str
    cluster: str
    shape: str
    chirality: int
    expansion_bias: float
    impedance: float
    sha256_prefix: str
    laplacian_spectrum: List[float]
    boundary_normal: List[float]
    base_congruence: float
    relational_value_index: float
    solo_morph_rate: float
    solo_warp_rating: float
    frictional_heat: float
    phase_state: str
    null_drain_heat: float

class SkillToolMorphologyOrchestrator:
    def __init__(self, entities_raw: List[Tuple[Any, ...]]):
        self.entities_raw = entities_raw
        self.profiles: List[EntityProfile] = []
        self.interlock_matrix: np.ndarray = np.zeros((len(entities_raw), len(entities_raw)))

    @staticmethod
    def _deterministic_spectral_vec(name: str, cluster: str, shape: str) -> Tuple[np.ndarray, np.ndarray, str]:
        h = hashlib.sha256(f"{name}|{cluster}|{shape}".encode("utf-8")).hexdigest()
        vals = [int(h[i:i+4], 16) / 65535.0 for i in range(0, 24, 4)]
        spec = np.sort(np.array(vals[:6], dtype=np.float64))
        spec[0] = 0.0
        raw_n = np.array([
            math.cos(2 * math.pi * vals[0]) * math.sin(math.pi * vals[1]),
            math.sin(2 * math.pi * vals[0]) * math.sin(math.pi * vals[1]),
            math.cos(math.pi * vals[1]),
        ], dtype=np.float64)
        norm = np.linalg.norm(raw_n)
        unit_n = raw_n / (norm if norm > 1e-12 else 1.0)
        return spec, unit_n, h[:12]

    def analyze_workflow_input(self, workflow_prompt: str) -> Dict[str, Any]:
        tokens = workflow_prompt.lower().split()
        expand_kw = sum(1 for t in tokens if any(k in t for k in ("expand", "grow", "bloom", "build", "swarm", "generate", "pack")))
        contract_kw = sum(1 for t in tokens if any(k in t for k in ("contract", "compress", "prune", "lock", "audit", "bound", "zero")))
        radial_velocity = (expand_kw - contract_kw) / max(1.0, float(expand_kw + contract_kw))
        if abs(radial_velocity) < 0.1:
            phase = "null-stasis"
        elif radial_velocity > 0.35:
            phase = "plasma"
        elif radial_velocity > 0.0:
            phase = "fluid"
        else:
            phase = "solid"
        h = hashlib.sha256(workflow_prompt.encode("utf-8")).hexdigest()
        target_spec = np.sort(np.array([int(h[i:i+4], 16) / 65535.0 for i in range(0, 24, 4)]))
        target_spec[0] = 0.0
        return {
            "token_count": len(tokens),
            "radial_velocity": round(radial_velocity, 6),
            "motion_regime": "expanding" if radial_velocity > 0 else ("contracting" if radial_velocity < 0 else "isometric"),
            "target_phase": phase,
            "target_spectrum": target_spec,
        }

    def evaluate_all_entities(self, workflow_profile: Dict[str, Any]) -> List[EntityProfile]:
        self.profiles = []
        v_r = workflow_profile["radial_velocity"]
        target_phase = workflow_profile["target_phase"]
        target_spec = workflow_profile["target_spectrum"]
        G0 = NOTEBOOK_METRICS["G0_CONGRUENCE"]
        theta_lean = NOTEBOOK_METRICS["THETA_LEAN_RAD"]
        q_landauer = NOTEBOOK_METRICS["KB_T_LN2_NORM"]

        for idx, (name, kind, cluster, shape, chirality, exp_bias, impedance) in enumerate(self.entities_raw):
            faces, edges, verts, euler_chi, pack_eff, base_vol = SHAPE_GEOMETRY[shape]
            spec, normal, h_pref = self._deterministic_spectral_vec(name, cluster, shape)
            euler_norm = 1.0 if euler_chi in (0, 2) else 0.75
            base_congruence = (0.55 * pack_eff + 0.30 * euler_norm + 0.15 * (1.0 / (1.0 + abs(impedance - 1.0))))
            spec_sim = math.exp(-float(np.sum((spec - target_spec) ** 2)))
            motion_align = 1.0 - 0.5 * abs(exp_bias - v_r)
            rvi = (0.5 * spec_sim + 0.5 * motion_align) * math.exp(-abs(base_congruence - G0))
            eff_velocity = abs(v_r + exp_bias)
            curvature_kappa = (edges / max(1.0, float(faces))) / 2.5
            solo_morph = round(eff_velocity * (1.0 + curvature_kappa / NOTEBOOK_METRICS["SHADOW_OFFSET_RATIO"]), 6)
            phase_xi = 1.0 + PHASE_BETA[target_phase]
            grad_norm = float(np.linalg.norm(np.diff(spec)))
            solo_warp = round(math.sqrt(grad_norm ** 2 + (chirality * math.sin(theta_lean) * phase_xi) ** 2), 6)
            comp_shear = NOTEBOOK_METRICS["DELTA_VAR_GALOIS"] if (v_r + exp_bias) < 0 else 0.0
            eta_friction = abs(impedance - 1.0) + 0.08 * (1.0 - pack_eff)
            heat = q_landauer + eta_friction * (solo_morph ** 2) * (1.0 + PHASE_BETA[target_phase]) * (1.0 + comp_shear)
            null_drain = max(0.0, heat - 0.05)
            self.profiles.append(EntityProfile(
                idx=idx, name=name, kind=kind, cluster=cluster, shape=shape,
                chirality=chirality, expansion_bias=exp_bias, impedance=impedance,
                sha256_prefix=h_pref, laplacian_spectrum=spec.tolist(),
                boundary_normal=normal.tolist(), base_congruence=round(base_congruence, 6),
                relational_value_index=round(rvi, 6), solo_morph_rate=solo_morph,
                solo_warp_rating=solo_warp, frictional_heat=round(heat, 6),
                phase_state=target_phase, null_drain_heat=round(null_drain, 6),
            ))
        return self.profiles

    def compute_interlocking_matrix(self) -> np.ndarray:
        n = len(self.profiles)
        mat = np.zeros((n, n), dtype=np.float64)
        for i in range(n):
            pi = self.profiles[i]
            si = np.array(pi.laplacian_spectrum)
            ni = np.array(pi.boundary_normal)
            for j in range(i + 1, n):
                pj = self.profiles[j]
                sj = np.array(pj.laplacian_spectrum)
                nj = np.array(pj.boundary_normal)
                spec_res = math.exp(-float(np.sum((si - sj) ** 2)))
                mate_lock = 0.5 * (1.0 - float(np.dot(ni, nj)))
                gamma = abs(pj.impedance - pi.impedance) / max(1e-9, abs(pj.impedance + pi.impedance))
                admittance = 1.0 - gamma
                chiral_bonus = 0.12 if (pi.chirality * pj.chirality <= 0) else 0.05
                cross_kind_bonus = 0.08 if (pi.kind != pj.kind) else 0.0
                therm_pen = abs(pi.frictional_heat - pj.frictional_heat) / (1.0 + pi.frictional_heat + pj.frictional_heat)
                fit = 0.35 * spec_res + 0.25 * mate_lock + 0.30 * admittance + chiral_bonus + cross_kind_bonus - 0.15 * therm_pen
                fit = max(0.0, min(1.0, fit))
                mat[i, j] = mat[j, i] = fit
        self.interlock_matrix = mat
        return mat

    def evaluate_chain(self, indices: List[int]) -> Dict[str, Any]:
        k = len(indices)
        members = [self.profiles[i] for i in indices]
        if k == 1:
            m = members[0]
            _, _, _, _, pack_eff, base_vol = SHAPE_GEOMETRY[m.shape]
            return {
                "depth": 1,
                "members": [m.name],
                "kinds": [m.kind],
                "bond_strength": round(m.relational_value_index, 6),
                "completed_cycles": 0,
                "packing_density": round(pack_eff * math.exp(-m.solo_warp_rating), 6),
                "packed_volume": round(base_vol, 4),
                "total_heat": round(m.frictional_heat, 6),
                "mean_warp": round(m.solo_warp_rating, 6),
                "sto_pass": m.relational_value_index >= NOTEBOOK_METRICS["HAMMER_THRESHOLD"],
                "composite_shape": m.shape,
            }
        edge_fits = [self.interlock_matrix[indices[m], indices[m + 1]] for m in range(k - 1)]
        geo_mean_fit = float(np.exp(np.mean(np.log(np.maximum(edge_fits, 1e-9)))))
        G = nx.Graph()
        for a in range(k):
            G.add_node(indices[a])
            for b in range(a + 1, k):
                w = self.interlock_matrix[indices[a], indices[b]]
                if w >= NOTEBOOK_METRICS["HAMMER_THRESHOLD"]:
                    G.add_edge(indices[a], indices[b], weight=w)
        cycles = nx.cycle_basis(G)
        cycle_boost = 1.0 + 0.06 * min(len(cycles), k) / float(k)
        bond_strength = min(1.0, geo_mean_fit * cycle_boost)
        vols = [SHAPE_GEOMETRY[m.shape][5] for m in members]
        effs = [SHAPE_GEOMETRY[m.shape][4] for m in members]
        mean_warp = float(np.mean([m.solo_warp_rating for m in members]))
        total_heat = float(np.sum([m.frictional_heat for m in members]))
        effective_vol = float(np.sum(vols))
        hull_vol = effective_vol / max(0.50, float(np.mean(effs)) * (0.75 + 0.25 * bond_strength))
        packing_density = (effective_vol / hull_vol) * math.exp(-0.25 * mean_warp)
        if k == 10 and len(cycles) >= 3:
            comp_shape = "10-Strata Interlocked Toroidal-Crown Column"
        elif len(cycles) >= 1:
            comp_shape = f"Closed-{k}-Ring Helical Polytope"
        else:
            comp_shape = f"Open-{k}-Link Geodesic Chain"
        return {
            "depth": k,
            "members": [m.name for m in members],
            "kinds": [m.kind for m in members],
            "bond_strength": round(bond_strength, 6),
            "completed_cycles": len(cycles),
            "packing_density": round(packing_density, 6),
            "packed_volume": round(effective_vol, 4),
            "total_heat": round(total_heat, 6),
            "mean_warp": round(mean_warp, 6),
            "sto_pass": bond_strength >= NOTEBOOK_METRICS["HAMMER_THRESHOLD"],
            "composite_shape": comp_shape,
        }

    def build_1_to_10_library(self, records_per_level: int = 10) -> Dict[int, List[Dict[str, Any]]]:
        library: Dict[int, List[Dict[str, Any]]] = {}
        ranked_ids = [p.idx for p in sorted(self.profiles, key=lambda x: x.relational_value_index, reverse=True)]
        library[1] = [self.evaluate_chain([idx]) for idx in ranked_ids[:records_per_level]]
        anchor_pool = ranked_ids[:records_per_level]
        for depth in range(2, 11):
            level_combos = []
            for seed_idx in anchor_pool:
                chain = [seed_idx]
                used = {seed_idx}
                while len(chain) < depth:
                    last = chain[-1]
                    best_cand = None
                    best_score = -1.0
                    for cand in range(len(self.profiles)):
                        if cand in used:
                            continue
                        local_fit = self.interlock_matrix[last, cand]
                        loop_fit = float(np.mean([self.interlock_matrix[c, cand] for c in chain]))
                        tool_incentive = 0.04 if (depth >= 4 and self.profiles[cand].kind == "tool") else 0.0
                        score = 0.55 * local_fit + 0.45 * loop_fit + tool_incentive
                        if score > best_score:
                            best_score = score
                            best_cand = cand
                    chain.append(best_cand)
                    used.add(best_cand)
                level_combos.append(self.evaluate_chain(chain))
            level_combos.sort(key=lambda r: r["bond_strength"] * r["packing_density"], reverse=True)
            library[depth] = level_combos
        return library

# Initialize orchestrator globally
sample_workflow = (
    "Analyze workflow morphology, pack interlocking skill-tool chains into 10-skill "
    "vertical shapes, bound topological heat and warp, and sync across GitMCP, Drive, "
    "Airtable, Miro, Linear, and Sentry MCP."
)
orch = SkillToolMorphologyOrchestrator(ACTIVATED_ENTITIES_RAW)
wf = orch.analyze_workflow_input(sample_workflow)
profiles = orch.evaluate_all_entities(wf)
imat = orch.compute_interlocking_matrix()
lib = orch.build_1_to_10_library(records_per_level=10)

# Extract metrics from the 10-skill vertical shapes level (Depth 10)
level_10_shapes = lib[10]
bond_strengths = [s["bond_strength"] for s in level_10_shapes]
packing_densities = [s["packing_density"] for s in level_10_shapes]
total_heats = [s["total_heat"] for s in level_10_shapes]
labels = [f"Shape {i+1}" for i in range(len(level_10_shapes))]

x = np.arange(len(labels))
width = 0.25

fig, ax1 = plt.subplots(figsize=(12, 6), facecolor="#111111")
ax1.set_facecolor("#111111")

# Plot Bond Strength and Packing Density on the left y-axis
rects1 = ax1.bar(x - width/2, bond_strengths, width, label="Bond Strength", color="#00ffcc", alpha=0.85)
rects2 = ax1.bar(x + width/2, packing_densities, width, label="Packing Density", color="#ffcc00", alpha=0.85)
ax1.set_ylabel("Metrics (0 to 1.0)", color="white")
ax1.set_xlabel("10-Skill Interlocking Vertical Shapes", color="white")
ax1.tick_params(colors="white")
ax1.set_ylim(0, 1.2)

# Plot Total Heat on the right y-axis
ax2 = ax1.twinx()
rects3 = ax2.plot(x, total_heats, color="#ff3366", marker="o", linewidth=2.5, label="Total Heat (Thermal Load)")
ax2.set_ylabel("Total Thermal Heat", color="white")
ax2.tick_params(colors="white")

# Titles & Legends
plt.title("Comparison of the 10-Skill Interlocking Vertical Shapes", color="white", fontsize=14, pad=15)
ax1.set_xticks(x)
ax1.set_xticklabels(labels, color="white", rotation=45)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, facecolor="#111111", edgecolor="white", loc="upper right", labelcolor="white")

plt.grid(True, which="both", linestyle="--", color="gray", alpha=0.3)
plt.tight_layout()
plt.show()
````
