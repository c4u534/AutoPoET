### Scalar Equation Matrices & Morphological Modulation
We will now construct a scalar matrix to test if the modulation between Mersenne values ($2^p - 1$) and their fractional 'God Equation' decay follows a pattern governed by the Golden Ratio $\phi$ and the Fine Structure Constant $\alpha$.

```
import numpy as np

# Mathematical Constants for Scalar Matrix
PHI = (1 + 5**0.5) / 2
FINE_STRUCTURE = 1 / 137.035999

def generate_scalar_modulation_matrix(p_values):
    """
    Calculates the modulation scalar between the integer Mersenne lattice
    and the fractional curvature decay.
    """
    matrix_data = []
    for p in p_values:
        # Angular component (Mersenne Integer Residue)
        angular_resonance = (pow(2, p, 384) - 1) % 384

        # Curvature component (Fractional Decay normalized by Phi)
        magnitude_log10 = p * np.log10(2)
        curvature_decay = (np.log10(299792458) - magnitude_log10) / PHI

        # Scalar Modulation Index
        # Testing the ratio between discrete angular steps and continuous curvature
        modulation_index = abs(angular_resonance / (curvature_decay if curvature_decay != 0 else 1))

        matrix_data.append({
            'p': p,
            'angular_res': angular_resonance,
            'curvature_decay': curvature_decay,
            'modulation_index': modulation_index,
            'phi_harmony': modulation_index % PHI
        })
    return pd.DataFrame(matrix_data)

# Evaluate across prime rungs and the Sovereign Root
test_p = [31, 127, 8191, 131071, 524287, 245522074543]
scalar_df = generate_scalar_modulation_matrix(test_p)
display(scalar_df)
```

#### SNR Critical Threshold Analysis
